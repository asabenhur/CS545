# BERT: Our Encoder, Pretrained

Look back at what we built in the previous notebook.  `TransformerClassifierScratch` is an *encoder-only* transformer: token embeddings plus positional encoding, a stack of encoder blocks (multi-head self-attention + a position-wise FFN, each wrapped in Add & Norm), and a readout from a `<cls>` slot into a linear classifier.  We trained it from scratch on 25,000 IMDb reviews and it worked — but it is a narrow specialist, and its knowledge of English is only whatever those reviews could teach it.  Rare words never got enough gradient to mean anything (we saw this in the RNN module), and nothing it learned transfers to any other task.

**BERT** (Bidirectional Encoder Representations from Transformers; Devlin et al., 2018) is that exact architecture — encoder-only, `<cls>` readout and all — with two differences: it is *bigger* (12 layers and a 768-dimensional model instead of our 2 and 100, with a subword vocabulary), and, crucially, it is *pretrained* on billions of words of general text before it ever sees a labeled task.  The `<cls>` token we adopted in notebook 1 was BERT's invention; using its output as the sequence summary is exactly what BERT does.

This notebook makes that concrete.  Rather than describe BERT, we will take a real, pretrained one and (1) watch the objective it was pretrained on, then (2) use it to beat our from-scratch IMDb classifier — without training BERT at all.  We finish by looking beyond encoder-only models toward the large language models that are the next topic.

## The pretraining objective: masked language modeling

Start with the objective, because it is what turns a plain encoder into BERT.  How can a model learn language with *no* labels?  BERT's answer is **masked language modeling** (MLM): take ordinary text, hide a random token behind a `[MASK]` symbol, and train the encoder to predict what was hidden — using the words on *both* sides as context (this bidirectionality is what the "B" in BERT stands for).  Every sentence in every book becomes free supervision, so the training data is effectively unlimited.

One masked example, from input to prediction:

```python
# render as crisp SVG (also set in the setup cell below)
try:
    from matplotlib_inline.backend_inline import set_matplotlib_formats
    set_matplotlib_formats('svg')
except ImportError:
    pass

import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

BLUE, GREEN, IO = '#cdddec', '#d0e3d4', '#ededed'
MASKC, AMBER, TX = '#f0c674', '#c0662f', '#333333'

fig, (axL, axR) = plt.subplots(1, 2, figsize=(9, 4),
                               gridspec_kw={'width_ratios': [1.55, 1]})

# --- Panel A: masked language modeling ---
toks = ['<cls>', 'I', '[MASK]', 'this', 'red', 'car']
BW = 0.86
for i, t in enumerate(toks):
    c = MASKC if t == '[MASK]' else IO
    axL.add_patch(FancyBboxPatch((i - BW/2, -0.28), BW, 0.56,
                  boxstyle='round,pad=0.02', facecolor=c, edgecolor='none'))
    axL.text(i, 0, t, ha='center', va='center', fontsize=8, color=TX,
             fontweight='bold' if t == '[MASK]' else 'normal')
axL.text(2.5, -0.62, 'original word "love" hidden behind [MASK]',
         ha='center', va='top', fontsize=7.5, style='italic', color=AMBER)

axL.add_patch(FancyBboxPatch((-0.5, 0.95), 5.85, 0.7, boxstyle='round,pad=0.02',
              facecolor=BLUE, edgecolor='none'))
axL.text(2.42, 1.3, 'Transformer encoder  (bidirectional self-attention)',
         ha='center', va='center', fontsize=8, color=TX)
for i in range(len(toks)):
    axL.annotate('', xy=(i, 0.95), xytext=(i, 0.30),
                 arrowprops=dict(arrowstyle='-', color='0.6', lw=0.9))
axL.annotate('', xy=(2, 2.35), xytext=(2, 1.65),
             arrowprops=dict(arrowstyle='-|>', color=AMBER, lw=1.8))
for j in [0, 1, 3, 4, 5]:                       # both-sides context into the mask
    axL.annotate('', xy=(2, 1.28), xytext=(j, 1.28),
                 arrowprops=dict(arrowstyle='->', color='0.5', lw=0.8, alpha=0.35,
                                 connectionstyle='arc3,rad=-0.25'))
axL.add_patch(FancyBboxPatch((0.7, 2.35), 2.6, 1.15, boxstyle='round,pad=0.02',
              facecolor=GREEN, edgecolor='none'))
axL.text(2.0, 3.32, 'softmax over vocabulary', ha='center', va='center',
         fontsize=7.5, color=TX)
for k, (w, p) in enumerate([('love', 0.36), ('liked', 0.22), ('hated', 0.15)]):
    y = 3.02 - k * 0.24
    axL.add_patch(plt.Rectangle((1.05, y - 0.06), 1.6 * p / 0.36, 0.12,
                  facecolor=AMBER if k == 0 else '0.6', edgecolor='none'))
    axL.text(1.0, y, w, ha='right', va='center', fontsize=7, color=TX)
axL.text(2.0, 3.7, 'predict the hidden token   ->   "love"', ha='center',
         va='bottom', fontsize=8, color=AMBER, fontweight='bold')
axL.set_xlim(-1.0, 5.7); axL.set_ylim(-1.1, 4.15); axL.axis('off')
axL.set_title('Masked language modeling', fontsize=9)

# --- Panel B: attention pattern, tokens on both axes ---
n = len(toks)
for r in range(n):
    for c in range(n):
        axR.add_patch(plt.Rectangle((c, n - 1 - r), 1, 1, facecolor=BLUE,
                      edgecolor='white', lw=1.2))
axR.set_xlim(0, n); axR.set_ylim(0, n); axR.set_aspect('equal')
axR.set_xticks([i + 0.5 for i in range(n)])
axR.set_yticks([i + 0.5 for i in range(n)])
axR.set_xticklabels(toks, fontsize=6.5, rotation=45, ha='right')
axR.set_yticklabels(toks[::-1], fontsize=6.5)
axR.tick_params(length=0)
axR.set_xlabel('attended to  (key)', fontsize=8)
axR.set_ylabel('each token  (query)', fontsize=8)
axR.set_title('Encoder attention', fontsize=9)
plt.tight_layout();
```

## Getting a pretrained model

Enough concept — let's watch it work.  We use the Hugging Face `transformers` library, the standard way to load pretrained models, and **DistilBERT**: a compressed BERT that keeps \~97% of the accuracy at \~60% of the size, so it runs comfortably on a laptop.  The first call downloads the weights (\~250 MB) and caches them; later calls are instant.

The `fill-mask` pipeline runs the pretrained model and reports its top guesses for a masked position — exactly the objective pictured above, now on a model that has already done the training.

```python
# render figures as crisp vector graphics (SVG) in the notebook
try:
    from matplotlib_inline.backend_inline import set_matplotlib_formats
    set_matplotlib_formats('svg')
except ImportError:
    pass

# one-time install if needed:  pip install transformers
import os
import re
import tarfile
import urllib.request

import numpy as np
import torch
from torch import nn
from matplotlib import pyplot as plt

device = ('cuda' if torch.cuda.is_available()
          else 'mps' if torch.backends.mps.is_available() else 'cpu')
print(f'Using {device} device')

MODEL_NAME = "distilbert-base-uncased"
```

```python
from transformers import pipeline

fill = pipeline('fill-mask', model=MODEL_NAME, device=device)
# device is 'mps' on Apple Silicon; switch to device=-1 (CPU) if MPS errors out

for sentence in ['this movie was absolutely [MASK].',
                 'the acting was [MASK] and the plot made no sense.',
                 'a [MASK] is a small burrowing mammal.']:
    print(sentence)
    for p in fill(sentence, top_k=5):
        print(f'    {p["token_str"]:12s} {p["score"]:.3f}')
    print()
```

Two things are worth noticing.  First, the predictions are *sensible*: BERT completes "absolutely ___" with words like *fantastic*, *hilarious*, *terrible* — it has learned which words fit, and even that evaluative slot wants an evaluative word.  Second, it fills "a ___ is a small burrowing mammal" with *mole* (or *rabbit*, *rodent*), pulling on context from *both* directions — exactly the disambiguation-from-context idea from notebook 2, now learned from data at scale.  None of this required a labeled dataset; it fell out of predicting masked words over a huge corpus.

That is the whole trick of pretraining: a task you can generate for free (fill the blank) forces the model to build general representations of language, which we can then repurpose.

### How BERT was actually trained

The `fill-mask` demo is the finished product; here is the recipe that produced it, because several of the choices are instructive.

**Data.**  BERT was pretrained on the **BooksCorpus** (\~800M words) plus **English Wikipedia** (\~2.5B words) — about **3.3 billion words of unlabeled text**.  No annotation: the supervision is manufactured from the text itself, which is what lets pretraining use essentially unlimited data.

**Objective 1 — masked language modeling, and the 80/10/10 trick.**  BERT selects **15%** of the tokens in each sequence for prediction.  It would be natural to replace all of them with `[MASK]`, but that creates a subtle train/serve mismatch: `[MASK]` never appears in the real text you later fine-tune on, so a model that only ever predicts *at* `[MASK]` symbols is being trained for an input it will not see again.  BERT's fix: of the selected tokens, **80% become `[MASK]`, 10% are replaced by a random token, and 10% are left unchanged** — and the model is asked to predict the original in all three cases.  Because it cannot tell which token was corrupted and which is genuine, it is forced to build a useful representation at *every* position, not just decode the `[MASK]` symbol.

**Objective 2 — next-sentence prediction (and its quiet retirement).**  Original BERT was trained on a second task at the same time: given two segments packed as `[CLS] A [SEP] B [SEP]`, predict whether B genuinely followed A in the corpus, the idea being to teach inter-sentence relationships for downstream pair tasks (NLI, QA).  Later work (**RoBERTa**) showed NSP contributes little and *dropped it entirely* with no loss — and **DistilBERT, the model we load, has no NSP either**.  This is exactly why its raw `[CLS]` vector is an unremarkable summary until a task trains it (the pooling comparison below).

**Tokenization.**  BERT uses **WordPiece** subword tokens with a \~30,000-token vocabulary, so a rare or unseen word is split into known pieces (`playing` -> `play` + `##ing`).  There is effectively **no out-of-vocabulary problem** — a pointed contrast with the fixed word vocabulary and `<unk>` bucket we lived with in the RNN module.

**Size and optimization.**  BERT-base is 12 layers, hidden size 768, 12 heads, **110M parameters** (BERT-large: 24/1024/16, 340M).  It was trained with **Adam**, a peak learning rate of 1e-4 under a **warmup-then-linear-decay** schedule, weight decay, dropout 0.1, GELU activations, and batches of 256 sequences x 512 tokens for **\~1 million steps** (\~40 passes over the corpus) — roughly four days on TPUs.  That warmup+decay schedule is the same one we had to add to make *fine-tuning* stable a few cells down; it is standard for training transformers.

**And what we actually load — DistilBERT.**  DistilBERT is a 6-layer, **66M-parameter** student trained by *knowledge distillation*: it learns to match BERT-base's output distribution (soft targets) alongside the MLM objective, dropping NSP.  It keeps \~97% of BERT-base's benchmark score at \~60% of the size and about twice the speed — which is why the cells here run on a laptop.

**A few subtleties worth knowing.**

* **Contextual, not static, embeddings.**  Unlike GloVe (one fixed vector per word, as the CNN used), BERT gives each token a vector that depends on the *whole* sentence — *bank* in "river bank" and "savings bank" come out different.  That contextualization is the entire reason to run the encoder rather than just look words up in a table.
* **Learned positional embeddings, not sinusoids.**  In notebook 4 we built *sinusoidal* position encodings for their relative-offset and extrapolation properties; BERT instead simply **learns** a separate position vector for each of its 512 slots.  Simpler and effective, but it hard-caps inputs at 512 tokens and does not extrapolate beyond them — a real limitation that later models revisit (RoPE, ALiBi).
* **Bidirectional means BERT cannot generate.**  Because every token attends to the future as well as the past, BERT is built to *read* a whole input at once, not to write text left to right.  That is precisely why the next topic — generative LLMs — switches to *causal* (decoder-only) attention, where a token may see only what came before it.

## Reusing BERT for sentiment

Here is the payoff.  BERT has already learned to turn a piece of text into a rich vector — so instead of learning that from 25,000 reviews as we did in notebook 5, we can *borrow* it.  The simplest way, and the one we use here, is **feature extraction**: run each review through the frozen (untrained-by-us) BERT, read a single fixed vector out of its output for each review, and train only a small linear classifier on that.  BERT's weights never change; all the learning happens in one `Linear` layer.  (Which vector to read out — BERT's `[CLS]` slot, or an average over the tokens — turns out to matter, and we compare the two.)

First, the IMDb reviews as raw text (BERT brings its own tokenizer, so we skip the vocabulary we built earlier):

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

BLUE, GREEN, IO = '#cdddec', '#d0e3d4', '#ededed'
MASKC, AMBER, TX = '#f0c674', '#c0662f', '#333333'

fig, ax = plt.subplots(figsize=(6.8, 4.4))
toks = ['[CLS]', 'a', 'brilliant', 'film']
BW = 0.9
for i, t in enumerate(toks):
    c = MASKC if t == '[CLS]' else IO
    ax.add_patch(FancyBboxPatch((i - BW/2, -0.28), BW, 0.56,
                 boxstyle='round,pad=0.02', facecolor=c, edgecolor='none'))
    ax.text(i, 0, t, ha='center', va='center', fontsize=8, color=TX,
            fontweight='bold' if t == '[CLS]' else 'normal')
for i in range(len(toks)):
    ax.annotate('', xy=(i, 0.85), xytext=(i, 0.30),
                arrowprops=dict(arrowstyle='-', color='0.6', lw=0.9))
ax.add_patch(FancyBboxPatch((-0.55, 0.85), 4.1, 0.75, boxstyle='round,pad=0.02',
             facecolor=BLUE, edgecolor='none'))
ax.text(1.5, 1.22, 'pretrained BERT encoder', ha='center', va='center',
        fontsize=8.5, color=TX)
ax.annotate('', xy=(0, 2.35), xytext=(0, 1.6),
            arrowprops=dict(arrowstyle='-|>', color=AMBER, lw=1.8))
ax.text(0.15, 1.97, 'the [CLS] vector\n(768-dim summary)', ha='left',
        va='center', fontsize=7, color=AMBER)
ax.add_patch(FancyBboxPatch((-0.9, 2.35), 1.8, 0.6, boxstyle='round,pad=0.02',
             facecolor=GREEN, edgecolor='none'))
ax.text(0, 2.65, 'Linear head', ha='center', va='center', fontsize=8, color=TX)
ax.annotate('', xy=(0, 3.5), xytext=(0, 2.95),
            arrowprops=dict(arrowstyle='-|>', color='0.55', lw=1.5))
ax.text(0, 3.62, 'neg / pos', ha='center', va='bottom', fontsize=8.5, color=TX)
ax.text(3.75, 1.22, 'feature extraction:\nweights FROZEN,\ntrain only the head',
        ha='left', va='center', fontsize=7, color='#3a6ea5')
ax.text(3.75, 2.62, 'fine-tuning:\ngradients flow\ninto BERT too', ha='left',
        va='center', fontsize=7, color='#b5651d')
#ax.annotate('', xy=(-0.78, 1.1), xytext=(-0.78, 2.6),
#            arrowprops=dict(arrowstyle='<->', color='#b5651d', lw=1.0,
#                            linestyle='dashed', alpha=0.7))
ax.set_xlim(-1.7, 6.8); ax.set_ylim(-0.9, 3.95); ax.axis('off')
ax.set_title('Reusing a pretrained BERT for classification', fontsize=9)
plt.tight_layout();
```

```python
def download_imdb(data_dir='./data'):
    url = 'https://d2l-data.s3-accelerate.amazonaws.com/aclImdb_v1.tar.gz'
    os.makedirs(data_dir, exist_ok=True)
    archive = os.path.join(data_dir, 'aclImdb_v1.tar.gz')
    if not os.path.exists(archive):
        print('downloading', url, '...'); urllib.request.urlretrieve(url, archive)
    extracted = os.path.join(data_dir, 'aclImdb')
    if not os.path.exists(extracted):
        with tarfile.open(archive) as tar:
            tar.extractall(data_dir)
    return extracted

def read_imdb(data_dir, is_train):
    data, labels = [], []
    for label in ('pos', 'neg'):
        folder = os.path.join(data_dir, 'train' if is_train else 'test', label)
        for file in os.listdir(folder):
            with open(os.path.join(folder, file), 'rb') as f:
                data.append(f.read().decode('utf-8').replace('<br />', ' '))
                labels.append(1 if label == 'pos' else 0)
    return data, labels

def subsample(texts, labels, n, seed=0):
    """A balanced random subset of size n (encoding all 25k on CPU is slow)."""
    rng = np.random.default_rng(seed)
    idx = rng.permutation(len(texts))[:n]
    return [texts[i] for i in idx], [labels[i] for i in idx]

data_dir = download_imdb()
train_texts, train_labels = subsample(*read_imdb(data_dir, True), n=2000)
test_texts, test_labels = subsample(*read_imdb(data_dir, False), n=2000)
print(len(train_texts), 'train /', len(test_texts), 'test reviews')
```

Now the encoder.  We load DistilBERT and its tokenizer, put it in eval mode, and run the reviews through it *without* gradients — BERT's weights never change.  The only open question is how to turn each review's *sequence* of token vectors into one fixed vector for the classifier.  Two choices are standard, and we compute both:

* the **`[CLS]` vector** — the final-layer output at position 0, the summary slot BERT (and our own `<cls>`) reserve for exactly this purpose; and
* the **mean** of the final-layer token vectors, ignoring padding — a plain average over the whole review.

Both are 768-dimensional.  Which makes the better *frozen* feature is not obvious — and the answer is the lesson of this section.

```python
from transformers import AutoTokenizer, AutoModel

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
bert = AutoModel.from_pretrained(MODEL_NAME).to(device).eval()

@torch.no_grad()
def embed(texts, batch_size=16, max_len=256):
    """Frozen DistilBERT features, two poolings at once.
    Returns (cls, mean), each of shape (len(texts), 768)."""
    cls_feats, mean_feats = [], []
    for i in range(0, len(texts), batch_size):
        enc = tokenizer(texts[i:i + batch_size], padding=True, truncation=True,
                        max_length=max_len, return_tensors='pt').to(device)
        h = bert(**enc).last_hidden_state               # (B, T, 768)
        cls_feats.append(h[:, 0].cpu())                 # the [CLS] position
        mask = enc['attention_mask'].unsqueeze(-1)      # (B, T, 1)
        mean = (h * mask).sum(1) / mask.sum(1).clamp(min=1)   # padding-masked mean
        mean_feats.append(mean.cpu())
    return torch.cat(cls_feats), torch.cat(mean_feats)

cls_train, mean_train = embed(train_texts)
cls_test, mean_test = embed(test_texts)
y_train = torch.tensor(train_labels)
y_test = torch.tensor(test_labels)
print('feature shape:', cls_train.shape)   # (2000, 768) for each pooling
```

Each review is now a point in a 768-dimensional space that BERT arranged — in two versions, one per pooling.  All that remains is to find a boundary between positive and negative with a single linear layer.  We train the *same* probe on each feature set and compare; nothing about BERT is updated.

```python
def train_probe(X_train, X_test, epochs=500, lr=1e-3):
    """Train a linear classifier on fixed features; return test accuracy."""
    clf = nn.Linear(X_train.shape[1], 2)
    opt = torch.optim.Adam(clf.parameters(), lr=lr)
    loss_fn = nn.CrossEntropyLoss()
    for _ in range(epochs):
        opt.zero_grad()
        loss_fn(clf(X_train), y_train).backward()
        opt.step()
    with torch.no_grad():
        return (clf(X_test).argmax(1) == y_test).float().mean().item()

acc_cls = train_probe(cls_train, cls_test)
acc_mean = train_probe(mean_train, mean_test)
print(f'[CLS] feature        test accuracy: {acc_cls:.3f}')
print(f'mean-pooled feature  test accuracy: {acc_mean:.3f}')
```

The two poolings come out essentially **tied** (both near \~0.83–0.84; your exact numbers printed above).  That is worth pausing on, because it corrects a tempting misconception.

You may have read that *mean-pooling beats `[CLS]`* for BERT — and it does, but in a **different** setting: comparing raw, frozen vectors directly by cosine similarity (the Sentence-BERT result), where `[CLS]` is a poor sentence vector.  Here we are training a *supervised* linear probe, and a trained linear layer can read the sentiment direction out of either pooling about equally well — so the choice washes out.  The lesson is less "use mean" than **don't assume a finding from one regime (unsupervised similarity) carries over to another (supervised probing).**  (DistilBERT's `[CLS]` is also not specially trained — it dropped BERT's next-sentence objective — but for a trained probe that turns out not to matter either.)

More telling is *where* both land.  About 0.83–0.84 is roughly the linear-separability ceiling of *frozen* DistilBERT features for this task, and it is close to what our from-scratch model reached in notebook 5.  So on accuracy, frozen features do not really *beat* training from scratch — what they buy is **data efficiency and simplicity**: we got there with 2,000 reviews and a single `Linear` layer, where the from-scratch model needed all 25,000 and a full transformer training loop.  That is genuine transfer — the representation was paid for once, during pretraining — but it is an efficiency win, not an accuracy win.  (A linear probe converges to the best linear boundary in a *fixed* feature space, so more epochs cannot push past this; a nonlinear head buys little for the same reason — the ceiling is the feature, not the classifier.)

The accuracy win comes from **fine-tuning**: letting gradients flow back into DistilBERT so the representation itself adapts to sentiment — and `[CLS]`, finally given a task signal, becomes the summary it was designed to be.  The optional cell below does this the *realistic* way — AdamW with a learning-rate **warmup and linear decay**, gradient clipping, and more training data than the frozen probe used.  All three matter: fine-tuning is data-hungry, and without the schedule it is unstable — a naive constant-rate fine-tune can actually get *worse* after the first epoch.  It is the heavy path (comfortable on a GPU/MPS, slow on a CPU), but it is where pretraining decisively pays off.

```python
from transformers import (AutoModelForSequenceClassification,
                          get_linear_schedule_with_warmup)

# OPTIONAL, and the slow path: fine-tune DistilBERT end-to-end, the realistic
# way -- AdamW with weight decay, a learning-rate *warmup + linear decay*, and
# gradient clipping.  (Skipping the schedule is why a naive fine-tune can get
# WORSE in later epochs.)  Fine-tuning also needs more data than the frozen
# probe, so we draw a larger sample; set n_ft = 25000 on a GPU for the full
# result.  On CPU this is slow -- a GPU / Apple-MPS is strongly recommended.
n_ft = 5000
ft_texts, ft_labels = subsample(*read_imdb(data_dir, is_train=True),
                                n=n_ft, seed=1)
ft_y = torch.tensor(ft_labels)

ft_model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME, num_labels=2).to(device)
opt = torch.optim.AdamW(ft_model.parameters(), lr=2e-5, weight_decay=0.01)
loss_fn = nn.CrossEntropyLoss()
bs, epochs = 16, 3
total_steps = (len(ft_texts) + bs - 1) // bs * epochs
scheduler = get_linear_schedule_with_warmup(
    opt, num_warmup_steps=int(0.1 * total_steps),
    num_training_steps=total_steps)

def encode(texts):
    return tokenizer(texts, padding=True, truncation=True,
                     max_length=256, return_tensors='pt').to(device)

for epoch in range(epochs):
    ft_model.train()
    order = torch.randperm(len(ft_texts)).tolist()
    for i in range(0, len(order), bs):
        idx = order[i:i + bs]
        loss = loss_fn(ft_model(**encode([ft_texts[j] for j in idx])).logits,
                       ft_y[idx].to(device))
        opt.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(ft_model.parameters(), 1.0)
        opt.step()
        scheduler.step()
    ft_model.eval()
    correct = 0
    with torch.no_grad():
        for i in range(0, len(test_texts), bs):
            pred = ft_model(**encode(test_texts[i:i + bs])).logits.argmax(1).cpu()
            correct += (pred == y_test[i:i + bs]).sum().item()
    print(f'epoch {epoch + 1}: fine-tuned test accuracy '
          f'{correct / len(test_texts):.3f}')
```

Done properly, fine-tuning climbs into roughly the \~0.89–0.92 range on this 5,000-review sample — higher still with the full 25,000 — finally clearing both the frozen probe *and* the strong TF-IDF baseline from the recurrent module's dataset notebook.  That gain is exactly the representation *adapting* to the task instead of being read off frozen; it is the standard recipe in practice, where feature extraction is the fast, cheap approximation and fine-tuning is the real thing.  (Watch the per-epoch numbers *rise and hold* now, rather than degrade — that is the schedule doing its job.)

## Beyond BERT: toward large language models

BERT is one of three ways to use the transformer, distinguished only by *which half* of the original encoder–decoder they keep and how attention is allowed to flow:

* **Encoder-only** (BERT): bidirectional attention — every token sees every other — trained to fill in blanks.  Ideal for *understanding* a whole input at once: classification, tagging, retrieval.  This is what we built and just reused.
* **Encoder–decoder** (e.g. T5): an encoder reads the input, a decoder generates an output sequence, useful when the output is itself text, as in translation or summarization.
* **Decoder-only** (e.g. the GPT family): a single stack with *causal* attention — each token may attend only to earlier ones — trained simply to predict the next token.  Because "predict the next token" lets the model *generate* text, and because this objective scales smoothly with more data and parameters, decoder-only models are the architecture behind today's **large language models**.

The step from BERT to an LLM is mostly a change of objective and scale, not of mechanism: the attention, the residual blocks, the layer norms are the same pieces we built by hand.  What changes is that the model is trained to continue text rather than to fill blanks, and is grown by orders of magnitude — at which point new behaviors (following instructions, in-context learning, chain-of-thought reasoning) begin to appear.  That is where the course goes next.

## Summary

* BERT is the encoder-only transformer we built, scaled up and **pretrained** on huge unlabeled text via **masked language modeling** — predicting hidden tokens from bidirectional context.  Its `[CLS]` slot is the `<cls>` trick we used all module.
* As a **frozen feature extractor**, BERT matches our from-scratch model with far less data and no transformer training — an efficiency win.  The pooling (`[CLS]` vs mean) barely matters for a supervised probe, and the frozen accuracy has a ceiling.
* **Fine-tuning** — updating BERT's weights on the task — is where pretraining decisively wins on accuracy, and where `[CLS]` finally becomes a strong summary.
* Encoder-only (BERT), encoder–decoder (T5), and decoder-only (GPT) differ mainly in attention direction and objective; the decoder-only, next-token line scales into large language models — the next topic.

## Exercise: watching a word become contextual

In the RNN module we inspected learned word embeddings — but those were *static*: one vector per word type, identical in every sentence.  A subtlety noted earlier is that BERT's representations are **contextual** — a token's vector depends on the words around it.  This exercise makes that visible and turns it into a measurement.

The idea: take one word used in two *different senses* (a river **bank** vs. a savings **bank**), run each sentence through BERT, and pull out that word's vector at **every layer** — the input embedding, then after each of the six layers.  If the network is contextualizing, the two vectors (identical token, different context) should be *pulled apart* as they rise through the stack.

One caution keeps this honest: in deep transformers cosine similarities drift downward for *everything* (an anisotropy effect), so the absolute numbers are not the story.  The signal is the **gap** between two curves — the *same* word in *different* senses should separate faster and further than the same word used twice in the *same* sense.

```python
import torch, numpy as np, torch.nn.functional as F
import matplotlib.pyplot as plt

@torch.no_grad()
def word_layer_vectors(sentence, word):
    """(n_layers+1, hidden) -- `word`'s vector at every layer of DistilBERT
    (row 0 = input embedding, then each layer), averaged over its sub-tokens."""
    enc = tokenizer(sentence, return_tensors='pt').to(device)
    ids = enc['input_ids'][0].tolist()
    wp = tokenizer(word, add_special_tokens=False)['input_ids']   # word -> sub-token ids
    span = next((list(range(i, i + len(wp)))
                 for i in range(len(ids) - len(wp) + 1)
                 if ids[i:i + len(wp)] == wp), None)
    if span is None:
        raise ValueError(f'"{word}" not found as tokens in: {sentence!r}')
    hs = bert(**enc, output_hidden_states=True).hidden_states     # embeddings + each layer
    return torch.stack([h[0, span].mean(0) for h in hs]).cpu()

def cos_by_layer(a, b):
    return F.cosine_similarity(a, b, dim=1).numpy()

# same word "bank", DIFFERENT senses -> should diverge with depth
diff = cos_by_layer(
    word_layer_vectors('i sat on the river bank and watched the water.', 'bank'),
    word_layer_vectors('i deposited a check at the bank downtown.',      'bank'))
# same word "bank", SAME sense -> should stay closer
same = cos_by_layer(
    word_layer_vectors('i deposited a check at the bank downtown.',    'bank'),
    word_layer_vectors('she withdrew cash from the bank this morning.', 'bank'))

print('different senses:', np.round(diff, 3))
print('same sense      :', np.round(same, 3))

plt.figure(figsize=(6, 4))
plt.plot(diff, 'o-', label='"bank": river vs. finance  (different senses)')
plt.plot(same, 's-', label='"bank": finance vs. finance (same sense)')
plt.xlabel('layer  (0 = input embedding)'); plt.ylabel('cosine similarity')
plt.xticks(range(len(diff))); plt.ylim(0, 1.02); plt.legend()
plt.title("How context reshapes a token's representation")
plt.tight_layout();
```

Read the plot as the *gap* between the two curves.  Both start high at layer 0 — the input embedding of "bank" is the same token, so only position differs — and by the upper layers the *different-senses* curve should sit clearly **below** the *same-sense* one: the network has rewritten one "bank" toward *river / water* and the other toward *money*.

**A negative control — the from-scratch encoder.**  Does *any* transformer do this, or does it take BERT's depth and pretraining?  The cell below runs the identical measurement on our 2-layer, sentiment-only model from notebook 5 — random-initialized embeddings, trained on 25,000 reviews to classify sentiment and nothing else.  It reuses the `cos_by_layer` helper defined just above, but needs notebook 5's `model`, `vocab`, and `tokenize` in scope, so **run it in notebook 5's namespace** (or bring those objects in).  Expect the two curves to stay high and nearly on top of each other: with only two blocks, and an objective that never required telling word senses apart, it barely contextualizes.  That flatness *is* the result — contextualization comes from depth and the right pretraining objective, not from the architecture alone.

```python
import math, numpy as np, torch
import matplotlib.pyplot as plt

# NEGATIVE CONTROL: the 2-layer, sentiment-only encoder from notebook 5.
# Needs that notebook's objects in scope -- run this there, or %run it in.
if not all(n in globals() for n in ('model', 'vocab', 'tokenize')):
    raise NameError("this cell needs notebook 5's `model`, `vocab`, and "
                    "`tokenize` -- run it in that notebook's namespace")

@torch.no_grad()
def word_layer_vectors_scratch(sentence, word):
    """(num_blks+1, num_hiddens) -- `word`'s vector at each stage of the
    from-scratch encoder: row 0 = input (embedding + positional), then the
    output of each encoder block.  Word-level, so one token per word."""
    model.eval()
    toks = ['<cls>'] + tokenize(sentence)
    if word not in toks:
        raise ValueError(f'"{word}" (lowercased) is not a token in: {sentence!r}')
    if vocab[word] == vocab.unk:
        raise ValueError(f'"{word}" is out-of-vocabulary (<unk>) -- pick a commoner word')
    pos = toks.index(word)
    X = torch.tensor(vocab[toks], device=device).reshape(1, -1)
    H = model.pos_encoding(model.embedding(X) * math.sqrt(model.num_hiddens))
    reps = [H[0, pos]]                      # layer 0: input representation
    for blk in model.blks:
        H = blk(H)
        reps.append(H[0, pos])             # after each encoder block
    return torch.stack(reps).cpu()

def cos_by_layer(a, b):
    return F.cosine_similarity(a, b, dim=1).numpy()

diff = cos_by_layer(
    word_layer_vectors_scratch('i sat on the river bank and watched the water', 'bank'),
    word_layer_vectors_scratch('i deposited a check at the bank downtown',      'bank'))
same = cos_by_layer(
    word_layer_vectors_scratch('i deposited a check at the bank downtown',    'bank'),
    word_layer_vectors_scratch('she withdrew cash from the bank this morning', 'bank'))

print('different senses:', np.round(diff, 3))
print('same sense      :', np.round(same, 3))

plt.figure(figsize=(6, 4))
plt.plot(diff, 'o-', label='"bank": river vs. finance  (different senses)')
plt.plot(same, 's-', label='"bank": finance vs. finance (same sense)')
plt.xlabel('block  (0 = input embedding)'); plt.ylabel('cosine similarity')
plt.xticks(range(len(diff))); plt.ylim(0, 1.02); plt.legend()
plt.title('From-scratch encoder: does context reshape the token?')
plt.tight_layout();
```
