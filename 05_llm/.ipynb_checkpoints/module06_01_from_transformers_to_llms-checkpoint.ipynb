# From Transformers to Large Language Models

In the previous module we built multi-head attention and positional encoding, assembled encoder blocks into a from-scratch sentiment classifier, and then met BERT — that same encoder, scaled up and pretrained on billions of words.  In this module we transition from encoder models to decoder-only models.  Those are trained to do nothing but **predict the next token**.  This is the architecture behind today's large language models.

This notebook is the conceptual map for the construction of such models.  No new architecture will be introduced — the pieces are the ones we already used and built.  What is new is the *objective* (language modeling), the *input representation* (subword tokens), the *scale*, and what falls out of combining the three.  Along the way we will play with a real pretrained model — GPT-2, small enough to run on any laptop — to keep the concepts attached to something that actually works.

Everything here is designed to run on a laptop CPU or Apple-silicon MPS; nothing in this module requires a GPU cluster.

## Three ways to use a transformer

The original transformer model (Vaswani et al., 2017) was an encoder–decoder architecture designed for machine translation.  Nearly all variants since keep the same transformer building blocks and differs mainly in **which half is kept, and how attention is allowed to flow**.  That single choice sorts pretrained transformers into three families:

* **Encoder-only** (BERT and its descendants).  Every token attends to every other: the model reads its whole input *bidirectionally* and produces one representation per token.  Pretrained by filling in masked blanks.  Built for *understanding* a given text and addresses tasks such as classification, tagging, and retrieval (covered in the previous module).
* **Encoder–decoder** (the original translation model; T5, BART).  An encoder reads the input; a decoder generates an output sequence token by token, attending *causally* to what it has generated so far and, through cross-attention, to the entire encoded input.  T5's contribution was to cast every task — translation, summarization, even classification — as text in, text out.
* **Decoder-only** (the GPT family — and essentially every model now called an "LLM").  Keep only the decoder stack.  Each token may attend only to tokens *before* it —the **causal attention** pattern;  the model is trained to predict the next token at every position.

The attention patterns are the real signature of the three families, and they are worth seeing side by side.

```python
# render figures as crisp vector graphics (SVG) in the notebook
try:
    from matplotlib_inline.backend_inline import set_matplotlib_formats
    set_matplotlib_formats('svg')
except ImportError:
    pass

# This module has its own environment.yml (a superset of the attention
# module's).  To update the course environment:
#     conda env update -f environment.yml --prune
import numpy as np
import torch
import matplotlib.pyplot as plt

device = ('cuda' if torch.cuda.is_available()
          else 'mps' if torch.backends.mps.is_available() else 'cpu')
print(f'Using {device} device')
```

```python
from matplotlib.colors import ListedColormap

BLUE, GREEN, IO = '#cdddec', '#d0e3d4', '#ededed'
AMBER, TX = '#c0662f', '#333333'
allowed = ListedColormap([IO, BLUE])   # 0 = blocked (gray), 1 = allowed (blue)

def show_pattern(ax, mask, toks, title):
    ax.imshow(mask, cmap=allowed, vmin=0, vmax=1)
    ax.set_xticks(range(len(toks)), toks, rotation=60, fontsize=7)
    ax.set_yticks(range(len(toks)), toks, fontsize=7)
    ax.set_title(title, fontsize=9)
    ax.set_xlabel('attends to (key position)', fontsize=7)
    for spine in ax.spines.values():
        spine.set_visible(False)
    ax.set_xticks(np.arange(-.5, len(toks)), minor=True)
    ax.set_yticks(np.arange(-.5, len(toks)), minor=True)
    ax.grid(which='minor', color='white', linewidth=1.2)
    ax.tick_params(which='both', length=0)

toks = ['<cls>', 'I', 'love', 'this', 'red', 'car']       # encoder input
dec_toks = ['<bos>', 'I', 'love', 'this', 'red', 'car']   # decoder input
src, tgt = ['I', 'love', 'cars'], ['<bos>', "j'aime", 'les', 'voitures']
m, k = len(src), len(tgt)
ed = np.zeros((m + k, m + k))
ed[:m, :m] = 1                          # encoder: source attends to all source
ed[m:, :m] = 1                          # cross-attention: target sees all source
ed[m:, m:] = np.tril(np.ones((k, k)))   # decoder: causal over the target

fig, axes = plt.subplots(1, 3, figsize=(8.5, 3.4))
show_pattern(axes[0], np.ones((len(toks), len(toks))), toks,
             'encoder-only (BERT)\nbidirectional')
show_pattern(axes[1], ed, src + tgt,
             'encoder–decoder (T5)\nbidirectional source, causal target')
show_pattern(axes[2], np.tril(np.ones((len(dec_toks), len(dec_toks)))), dec_toks,
             'decoder-only (GPT)\ncausal')
axes[0].set_ylabel('query position', fontsize=7)
fig.tight_layout();
```

In the figure above each row is a query position, and the blue cells mark the positions it is allowed to attend to.  The BERT encoder on the left is a full square and represents the pattern of bidirectional attention where every token sees every other.  This is also what makes it unable to generate text.  The encoder–decoder model (center) is a composite: a bidirectional encoder block for the source, a triangle for the target, and a filled rectangle of cross-attention connecting them.  The decoder-only pattern (right) is just the lower triangle: token $t$ sees tokens $1, \dots, t$ and nothing else.

Why has the field converged on the third — seemingly the most restricted — pattern?  A few reasons compound:

* **Simplicity.**  One stack, one attention pattern, one objective.  No masking scheme to design, no special corruption tokens, no encoder/decoder split to manage.
* **Dense training signal.**  With a causal mask, *every* position of *every* document is a prediction target, all computed in a single forward pass.  BERT, by contrast, learns only from the ~15% of tokens it masks.
* **Unification.**  Any task whose input and output can be written as text becomes a *continuation* problem for one and the same model — and generation, which BERT structurally cannot do, comes built in.
* **It scales.**  The next-token objective improves smoothly and predictably as models, data, and compute grow — a point we return to below.

Encoders are not obsolete — embedding and retrieval models are typically encoder-only to this day — but generation is where this module is headed, so from here on, *decoder-only* is the architecture.

## Language modeling: predict the next token

A **language model** assigns a probability to a sequence of tokens.  The chain rule of probability factors any such distribution — exactly, with no assumptions — into one conditional per position:

$$P(x_1, x_2, \dots, x_T) \;=\; \prod_{t=1}^{T} P(x_t \mid x_1, \dots, x_{t-1}).$$

An **autoregressive** language model implements the right-hand side directly: a network that, given a prefix, outputs a distribution over what comes next.  Notice what each factor is: a plain **classification problem** — softmax over the vocabulary, one $|V|$-way classifier per position.  The loss is the familiar cross-entropy, averaged over positions:

$$\mathcal{L} \;=\; -\frac{1}{T} \sum_{t=1}^{T} \log P(x_t \mid x_{<t}).$$

Machinery-wise, nothing here is new to us; what is new is where the labels come from.  Like BERT's masked-word objective, this is **self-supervised** — the "label" at each position is simply the next token of the text itself, so every book, webpage, and source file is free training data.  But it is even simpler than BERT's scheme: **no masking is performed**, simply the text shifted by one.

That shift is exactly how training is implemented.  Feed the model $x_1, \dots, x_T$; ask it, at every position to simultaneously, to predict $x_2, \dots, x_{T+1}$.  The causal mask is what makes this work: position $t$ must not peek at $x_{t+1}$ — its own label, or any subsequent token in the sequence.

Two pointers forward.  The quantity $\exp(\mathcal{L})$ has a name — **perplexity** — and an interpretation worth a notebook of its own (notebook 2).  And this loss, computed over a causally-masked stack of the blocks we built in the attention module, is the *entire* pretraining recipe of the GPT family — implemented from scratch in notebook 3.

```python
from matplotlib.patches import FancyBboxPatch

inp = ['<bos>', 'the', 'cat', 'sat', 'on', 'the', 'mat']
tgt = ['the', 'cat', 'sat', 'on', 'the', 'mat', '<eos>']

fig, ax = plt.subplots(figsize=(8.2, 3.4))
BW = 0.88
for i, t in enumerate(inp):
    ax.add_patch(FancyBboxPatch((i - BW/2, -0.28), BW, 0.56,
                 boxstyle='round,pad=0.02', facecolor=IO, edgecolor='none'))
    ax.text(i, 0, t, ha='center', va='center', fontsize=8, color=TX)
ax.add_patch(FancyBboxPatch((-0.5, 0.95), len(inp) - 0.1, 0.7,
             boxstyle='round,pad=0.02', facecolor=BLUE, edgecolor='none'))
ax.text((len(inp) - 1) / 2, 1.3,
        'transformer decoder  (causal self-attention)',
        ha='center', va='center', fontsize=8.5, color=TX)
for i in range(len(inp)):
    ax.annotate('', xy=(i, 0.95), xytext=(i, 0.30),
                arrowprops=dict(arrowstyle='-', color='0.6', lw=0.9))
    ax.annotate('', xy=(i, 2.02), xytext=(i, 1.65),
                arrowprops=dict(arrowstyle='-|>', color='0.6', lw=0.9))
for i, t in enumerate(tgt):
    ax.add_patch(FancyBboxPatch((i - BW/2, 2.02), BW, 0.56,
                 boxstyle='round,pad=0.02', facecolor=GREEN, edgecolor='none'))
    ax.text(i, 2.3, t, ha='center', va='center', fontsize=8, color=TX)
ax.text((len(inp) - 1) / 2, 2.95,
        'targets = the input, shifted one position to the left',
        ha='center', va='center', fontsize=8, style='italic', color=AMBER)
ax.text(-0.62, 0, 'input', ha='right', va='center', fontsize=8, color=TX)
ax.text(-0.62, 2.3, 'predict', ha='right', va='center', fontsize=8, color=TX)
ax.set_xlim(-1.7, len(inp) - 0.4)
ax.set_ylim(-0.75, 3.3)
ax.axis('off')
plt.tight_layout();
```

A word about this figure, because it invites a misreading.  It can look as if the network's job is to *copy its input one slot to the left* — and, when all we ultimately care about is the continuation of the whole prompt (the `<eos>` slot above), as if the outputs at the earlier positions were some strange overhead.  Neither is right.

The output at position $t$ is a full distribution over the vocabulary, computed — thanks to the causal mask — from the prefix $x_1, \dots, x_t$ *alone*.  Each column of the figure is therefore a self-contained instance of the same problem: *given this prefix, what comes next?*  Our one sentence poses that problem seven times — $P(\text{the} \mid \texttt{<bos>})$, then $P(\text{cat} \mid \texttt{<bos>}\,\text{the})$, all the way to $P(\texttt{<eos>} \mid \texttt{<bos>}\,\text{the cat sat on the mat})$.  The "shift" is not an operation the model performs, and not something it must learn; it only describes where the *loss* finds each column's correct answer — the label for the prefix ending at position $t$ happens to be stored at position $t+1$ of the same tensor.  The model learns one thing, the conditional distribution $P(x_t \mid x_{<t})$, and training simply evaluates it at all $T$ prefixes of the sequence in a single forward pass.  That is the "dense training signal" promised earlier: one sentence, $T$ gradient-carrying predictions.

At *generation* time, by contrast, only the last column is consulted — its prefix is the entire prompt, so its output is the prediction we want.  The predictions at earlier positions are computed and ignored (or, with a KV cache, not recomputed at all).

## A real next-token distribution

Rather than train a language model (that is notebook 3), let us borrow one that has already been trained and inspect the object at its heart: the conditional distribution $P(x_t \mid x_{<t})$.

**GPT-2** (Radford et al., 2019) is the natural specimen: a decoder-only transformer trained on 40 GB of web text, state of the art in 2019, the direct ancestor of every model with "GPT" in its name — and its smallest version, at 124M parameters, runs comfortably on a laptop.  As with DistilBERT in the previous module, the first call downloads the weights (~550 MB) and caches them; later calls are instant.

```python
from transformers import AutoTokenizer, AutoModelForCausalLM

LM_NAME = 'gpt2'   # the 124M-parameter "small" — GPT-2 came in 4 sizes up to 1.5B
tok = AutoTokenizer.from_pretrained(LM_NAME)
lm = AutoModelForCausalLM.from_pretrained(LM_NAME).to(device).eval()

n_params = sum(p.numel() for p in lm.parameters())
print(f'{n_params / 1e6:.0f}M parameters, vocabulary of {len(tok)} tokens')
```

```python
@torch.no_grad()
def next_token_probs(prompt, top_k=5):
    """Print the model's top-k candidates for the single next token."""
    ids = tok(prompt, return_tensors='pt').to(device)
    logits = lm(**ids).logits[0, -1]      # one score for every vocabulary token
    probs = logits.softmax(-1)
    top = probs.topk(top_k)
    print(f'{prompt!r}')
    for p, i in zip(top.values, top.indices):
        print(f'    {tok.decode(i)!r:16s} {p.item():.3f}')
    print()

for prompt in ['The capital of France is',
               'I sat down on the river bank and watched the',
               'import numpy as',
               'The opposite of hot is']:
    next_token_probs(prompt)
```

Look at where the probability mass goes: geography (*Paris*), word-sense disambiguation (*river* bank pulls the continuation toward water, not money — the contextualization we measured layer by layer in the BERT notebook), Python idiom (`np`), lexical relations (*cold*).  None of this was annotated.  It all condensed out of one objective — predict the next token — applied to a lot of text.  But note the model's size before being too impressed: swap in a less world-famous fact — say, `'The capital of Colorado is'` — and *Denver* may not even crack the top five at 124M parameters.  Factual recall at that grain is among the things that improve most dramatically with scale; keep this experiment in mind when we meet larger, better-trained models in notebook 4.  The distribution *is* the model: everything it "knows" is visible only as an arrangement of probability mass over continuations.

This should feel like BERT's `fill-mask` demo, with one crucial difference: the prediction uses only the *left* context.  And that difference is everything, because a model that predicts what comes next can **write**.  Feed it a prompt, pick a token from its distribution, append that token to the prompt, and repeat:

```python
@torch.no_grad()
def generate_greedy(prompt, n_tokens=40):
    """The autoregressive loop, in its simplest (greedy) form."""
    ids = tok(prompt, return_tensors='pt')['input_ids'].to(device)
    for _ in range(n_tokens):
        logits = lm(ids).logits[0, -1]        # distribution over the next token
        next_id = logits.argmax()             # greedy: always take the top one
        ids = torch.cat([ids, next_id.view(1, 1)], dim=1)
    return tok.decode(ids[0])

print(generate_greedy('In deep learning, the transformer architecture'))
```

Fluent, on-topic — and, if you let it run longer, it will eventually start repeating itself, a characteristic failure of always taking the single most likely token (try it: raise `n_tokens`).  The fix is to *sample* from the distribution instead, and how exactly to sample — temperature, top-$k$, top-$p$ — is a genuinely consequential design space that notebook 4 explores with the logits in hand.

Do not let the simplicity fool you: this ten-line loop **is** the inference engine of every chatbot you have used.  A production system swaps in a bigger model and a smarter selection rule (and a *KV cache* so the forward pass reuses attention computed for earlier tokens rather than redoing it), but the mechanism — one full forward pass per generated token, each token conditioned on all the ones before it — is exactly this.  It is also why generation is priced and metered *per token*, and why long answers take visibly longer than short ones.

## Tokens: the model's alphabet

We have been saying "token" without ever defining it, and the demos above quietly used GPT-2's tokenizer.  Time to look, because the choice of token is consequential.

The two obvious options both have serious problems.  **Words**, which we used in the RNN and transformer modules, give short sequences but an unbounded vocabulary: any word not in it becomes `<unk>` (recall how rare words starved for gradient in the IMDb models), and morphology is wasted — *slow*, *slower*, *slowest* are unrelated integers.  **Characters** solve coverage with a tiny alphabet, but sequences get several times longer — painful when self-attention costs $O(T^2)$ — and each position carries almost no meaning on its own.

**Subword** tokenization is the compromise every modern LLM uses, and **byte-pair encoding** (BPE; Sennrich et al., 2016, adapting a 1994 compression algorithm) is its standard construction.  The algorithm is almost embarrassingly simple:

1. Start with a base alphabet (characters, or raw bytes) and split the corpus into it.
2. Count all adjacent symbol pairs; **merge** the most frequent pair into a new symbol.
3. Repeat step 2 for a fixed number of merges.  The vocabulary is the base alphabet plus every merged symbol.

Frequent words end up as single tokens; rare words decompose into meaningful pieces; and *nothing* is ever out-of-vocabulary.  The algorithm fits in a screenful of code, so let us run it on a toy corpus and watch what it discovers:

```python
from collections import Counter

corpus = 'low lower lowest new newer newest wide wider widest slow slower slowest'
# each word starts as a sequence of characters, plus an end-of-word marker
words = Counter(tuple(w) + ('_',) for w in corpus.split())

def most_frequent_pair(words):
    pairs = Counter()
    for w, c in words.items():
        for pair in zip(w, w[1:]):
            pairs[pair] += c
    return max(pairs, key=pairs.get)

def apply_merge(words, pair):
    a, b = pair
    out = {}
    for w, c in words.items():
        new, i = [], 0
        while i < len(w):
            if i + 1 < len(w) and (w[i], w[i + 1]) == pair:
                new.append(a + b); i += 2
            else:
                new.append(w[i]); i += 1
        out[tuple(new)] = c
    return out

for step in range(1, 11):
    pair = most_frequent_pair(words)
    words = apply_merge(words, pair)
    print(f'merge {step:2d}: {pair[0]!r} + {pair[1]!r}  ->  {pair[0] + pair[1]!r}')

print('\nresulting segmentations:')
for w in sorted(words, key=lambda w: ''.join(w)):
    print('   ', ' | '.join(w))
```

The merge list is a mix of stems and suffix fragments — pieces like `low` and the endings that build *-er* and *-est* forms — discovered purely by counting, with no notion of morphology built in.  Ten merges on twelve words is a toy; a real tokenizer runs the same loop for tens of thousands of merges over gigabytes of text, starting from raw *bytes* so that any UTF-8 string whatsoever can be tokenized.  GPT-2's tokenizer has 50,257 tokens (50,000 merges, 256 byte symbols, and one end-of-text marker) — which is exactly the size of the softmax in the demos above.

Here is how it carves up real text (in the printout, `Ġ` marks a token that *begins with a space* — the space is part of the token):

```python
def show_tokens(text):
    pieces = tok.tokenize(text)
    print(f'{len(pieces):3d} tokens |', ' '.join(pieces))

show_tokens('The quick brown fox jumps over the lazy dog.')
show_tokens('Transformers are surprisingly effective.')
show_tokens('Antidisestablishmentarianism!')
show_tokens('strawberry Strawberry STRAWBERRY')
show_tokens('pi = 3.14159265358979')
show_tokens('El zorro marrón salta sobre el perro perezoso.')

print()
print("'hello'  ->", tok('hello')['input_ids'],
      "   ' hello' ->", tok(' hello')['input_ids'],
      "   'Hello' ->", tok('Hello')['input_ids'])
```

A short safari, but every finding matters in practice:

* **Common English words are single tokens; rare words shatter.**  *Antidisestablishmentarianism* becomes a string of fragments — the model must reassemble its meaning from pieces, as we hoped BPE would allow.
* **Case and leading spaces change the tokens.**  `hello`, ` hello`, and `Hello` are three different token sequences; *strawberry* tokenizes differently in each casing.  Models can be oddly sensitive to such surface details, and prompt formatting bugs (a stray trailing space) are a real failure mode.
* **The model never sees letters.**  If *strawberry* reaches the model as two or three opaque fragments, counting the r's in it requires knowledge the input representation actively hides.  The famous failures of LLMs at letter-counting and spelling games start here, at tokenization — not in the transformer.
* **Digits are chopped arbitrarily**, so multi-digit arithmetic must be learned over inconsistent chunks — one reason it is hard for language models.
* **Non-English text costs more.**  The Spanish sentence needs far more tokens than its English counterpart (note the byte-fallback pieces for *ó*), because GPT-2's merges were learned mostly on English.  More tokens per sentence means higher cost, slower generation, and less effective context for other languages — a fairness issue baked in below the model.
* **Token counts are tokenizer-relative.**  A "per-token" quantity — cost, context length, or a per-token loss — means nothing across models with different tokenizers.  This will matter immediately in notebook 2, because perplexity is exactly such a per-token quantity.

## Scaling: GPT to GPT-3, and the laws behind it

The objective and the tokens above were all in place by 2018–2019.  What turned this recipe into the center of the field was the discovery of how far — and how *predictably* — it improves with scale.

The **GPT** lineage tells the story in three steps.  GPT (Radford et al., 2018; 117M parameters) showed that a decoder pretrained on next-token prediction, then *fine-tuned* per task — BERT-style, one adapted copy per task — beat task-specific architectures.  **GPT-2** (2019; up to 1.5B) showed something stranger: pretrained on 40 GB of web text, it could perform translation, summarization, and question answering *without any fine-tuning*, just by being prompted — a hint that one generalist might replace the one-model-per-task workflow.  **GPT-3** (Brown et al., 2020; 175B, trained on ~300B tokens) turned the hint into a paradigm, as we will see in a moment.

What justified betting ever-larger compute budgets on one simple objective was an empirical regularity (Kaplan et al., 2020): language-modeling loss falls as a **power law** in each of model size, dataset size, and training compute, over many orders of magnitude —

![Transformer language-model performance improves smoothly as model size, dataset size, and training compute increase, following a power law in each factor when not bottlenecked by the other two (figure from Kaplan et al., 2020).](https://github.com/d2l-ai/d2l-en/raw/refs/heads/master/img/scaling-power-law.png)

— which converts "scale it up" from a gamble into something close to an engineering projection: you can *forecast* the loss of a model before spending the compute to train it.  GPT-3 landed on the predicted curve two orders of magnitude further out (Brown et al., 2020).

The scaling story has a second chapter worth knowing.  **Chinchilla** (Hoffmann et al., 2022) showed that the first-generation giants had the *balance* wrong: for a fixed compute budget, parameters and training tokens should grow together — roughly 20 tokens per parameter — and a 70B model trained on 1.4T tokens beat much larger models trained on less data.  Modern practice has moved past even that: today's small open models (the 0.5–1B ones we will run in notebook 4) are trained on *trillions* of tokens, far beyond their "compute-optimal" share, because training compute is spent once while inference cost is paid forever — a small model trained longer is a better model to *serve*.

## In-context learning

Scale did not just lower the loss; it changed how the models are *used*.  GPT-3's headline result was **in-context learning**: describe a task in the prompt — optionally with a few worked input–output examples — and the model performs it on the next input, *with no parameter updates at all*.

The canonical illustration is the one from the GPT-3 paper: translating a single English word into French, with the task specified — and optionally demonstrated — entirely in the prompt.

```python
# zero-, one- and few-shot prompting, after Brown et al. (2020), figure 2.1
from matplotlib.patches import FancyArrowPatch

def in_context_learning_figure():
    """Zero-, one- and few-shot prompting: the same frozen model, three prompts."""
    TASK = 'Translate English to French:'
    EXAMPLES = ['sea otter => loutre de mer',
                'peppermint => menthe poivrée',
                'plush giraffe => girafe en peluche']
    PROMPT = 'cheese =>'

    panels = [('zero-shot', 0), ('one-shot', 1), ('few-shot', 3)]
    LH, W, CX = 1.25, 14.0, [8.0, 23.0, 38.0]      # line height, box width, centres
    Y_BAR, Y_OUT = -3.4, -6.3

    fig, ax = plt.subplots(figsize=(9.2, 5.0))

    def mono(x, y, s, color=TX, ha='left', size=7.4, weight='normal'):
        ax.text(x, y, s, family='monospace', fontsize=size, color=color,
                ha=ha, va='center', weight=weight)

    for (name, n_ex), cx in zip(panels, CX):
        rows = [(TASK, GREEN)] + [(e, BLUE) for e in EXAMPLES[:n_ex]] \
               + [(PROMPT, IO)]
        left = cx - W / 2
        top = (len(rows) - 1) * LH
        # the prompt as one block, with each line tinted by its role
        ax.add_patch(FancyBboxPatch((left, -0.62), W, top + 1.24,
                                    boxstyle='round,pad=0.25', facecolor='white',
                                    edgecolor='0.75', linewidth=1.0))
        for k, (text, fc) in enumerate(rows):
            y = top - k * LH
            ax.add_patch(FancyBboxPatch((left + 0.35, y - 0.44), W - 0.7, 0.88,
                                        boxstyle='round,pad=0.06', facecolor=fc,
                                        edgecolor='none'))
            mono(left + 0.75, y, text)
        ax.text(cx, top + 1.5, name, ha='center', va='bottom', fontsize=10.5,
                color=TX, fontweight='bold')
        ax.add_patch(FancyArrowPatch((cx, -1.0), (cx, Y_BAR + 0.75),
                                     arrowstyle='-|>', mutation_scale=11,
                                     color='0.55', linewidth=1.3, shrinkA=0,
                                     shrinkB=0))

    # one model, three prompts
    ax.add_patch(FancyBboxPatch((CX[0] - W / 2, Y_BAR - 0.75),
                                CX[2] - CX[0] + W, 1.5,
                                boxstyle='round,pad=0.25', facecolor=BLUE,
                                edgecolor='none'))
    ax.text((CX[0] + CX[2]) / 2, Y_BAR,
            'the same pretrained model  —  one forward pass, weights frozen',
            ha='center', va='center', fontsize=9.5, color=TX)

    for cx in CX:
        ax.add_patch(FancyArrowPatch((cx, Y_BAR - 1.0), (cx, Y_OUT + 0.62),
                                     arrowstyle='-|>', mutation_scale=11,
                                     color='0.55', linewidth=1.3, shrinkA=0,
                                     shrinkB=0))
        ax.add_patch(FancyBboxPatch((cx - 2.4, Y_OUT - 0.55), 4.8, 1.1,
                                    boxstyle='round,pad=0.14', facecolor=GREEN,
                                    edgecolor='none'))
        mono(cx, Y_OUT, 'fromage', ha='center', size=8.4)

    # what the three colours mean
    key = [(GREEN, 'task description'), (BLUE, 'worked examples'),
           (IO, 'the prompt to complete')]
    for k, (fc, label) in enumerate(key):
        y = 7.3 - k * 0.95
        ax.add_patch(FancyBboxPatch((1.2, y - 0.32), 0.9, 0.64,
                                    boxstyle='round,pad=0.05', facecolor=fc,
                                    edgecolor='0.8', linewidth=0.6))
        ax.text(2.5, y, label, fontsize=8, va='center', color='0.35')

    ax.set_xlim(-1.0, 46.0)
    ax.set_ylim(Y_OUT - 1.6, 8.2)
    ax.axis('off')
    fig.tight_layout()


in_context_learning_figure()
```

Everything above the frozen bar is *text*.  Moving from zero-shot to few-shot does not change a single weight — it only makes the document the model is completing longer and more informative.  That is the whole mechanism.

The terminology: **zero-shot** (task description only), **one-shot** (one example), **few-shot** (a handful).  Pause on how strange this is.  Everywhere else in this course, "learning" has meant gradient descent on parameters.  Here the parameters are frozen; the "learning" happens inside a single forward pass, conditioned on the prompt — because predicting the continuation of a document that contains worked examples *requires* inferring the pattern connecting them.  Larger models are dramatically better at this, and its emergence at scale was not predicted by the smooth loss curves above.

In-context learning is why **prompting** became a practical engineering discipline: the prompt is the program.  One technique deserves special mention: **chain-of-thought** prompting (Wei et al., 2022) — showing or simply asking for intermediate reasoning steps before the answer ("let's think step by step") — substantially improves performance on math and multi-step reasoning.  Its descendants are the *reasoning models* of the last two years, trained to produce long internal chains of thought before answering.  We will do all of this with our own hands, on our own laptop, in notebook 4.

## From next-token predictor to assistant

One gap remains between everything above and the chatbots everyone knows.  A pretrained language model is an *autocomplete*, not an assistant.  Ask a **base model** "What is the capital of Colorado?" and it may well continue with *nine more quiz questions* — because a list of questions is a perfectly likely document.  It gives the most probable continuation, not the most helpful response.  (You can try this: put a question to `generate_greedy` above and watch what GPT-2 — a pure base model — does with it.)

Turning a base model into an assistant is **post-training**, and the modern pipeline has two stages stacked on top of pretraining:

1. **Pretraining** — everything this notebook described: next-token prediction on trillions of tokens.  Nearly all of the compute, and the source of essentially all of the model's capability.
2. **Instruction tuning (supervised fine-tuning, SFT)** — continue training, with the *same* next-token objective, on a curated corpus of instruction → response pairs.  No new math; the data alone reshapes behavior from "continue the document" to "answer the request" (Wei et al., 2021; Sanh et al., 2021).
3. **Preference tuning** — collect human (or AI) judgments comparing candidate responses, and optimize the model toward preferred ones: **RLHF**, reinforcement learning from human feedback (Ouyang et al., 2022 — the InstructGPT recipe that begat ChatGPT), or simpler direct methods such as **DPO** (Rafailov et al., 2023).  This is what tunes helpfulness, format, and refusals.

Open-weight model families (Llama, Qwen, Gemma, Mistral, and the rest) typically publish both the **base** and the **instruct** checkpoints of the same model.  That pair is a perfect controlled experiment — same architecture, same pretraining, different post-training — and comparing their answers to identical prompts, in notebook 4, is the clearest demonstration of what post-training does that you can run anywhere.  In notebook 5 we will perform a small instruction-tuning run ourselves, with LoRA.

One closing reassurance.  The models we will load in notebook 4 are, architecturally, the decoder-only transformer of this notebook with incremental refinements — rotary position embeddings instead of sinusoids, RMSNorm instead of LayerNorm, grouped-query attention to shrink the KV cache — every one of them a variation on a component we have already built by hand.  Nothing ahead is architecturally mysterious.

## Summary

* Pretrained transformers come in three modes distinguished by their **attention pattern**: encoder-only (bidirectional; BERT, previous module), encoder–decoder (T5), and **decoder-only** (causal; GPT) — the LLM architecture, and this module's focus.
* **Language modeling** is next-token prediction: the chain rule factors any sequence distribution into per-position conditionals, each a $|V|$-way classification trained with cross-entropy.  Self-supervised, so all text is training data; the causal mask plus shifted targets make every position a training example in one parallel forward pass.  $\exp(\text{loss})$ is **perplexity** (notebook 2).
* **Generation** is the autoregressive loop: forward pass, pick a token, append, repeat.  Greedy picking degenerates into repetition; principled sampling is notebook 4's topic.
* **BPE tokenization** builds a subword vocabulary by iteratively merging frequent pairs: no out-of-vocabulary words, but real artifacts — case/whitespace sensitivity, hidden letters, chopped digits, higher token costs for non-English text.
* **Scaling laws** make loss a predictable power law in model size, data, and compute; Chinchilla set the parameter–token balance, and modern small models deliberately train far past it.  At scale, **in-context learning** emerges: tasks specified in the prompt, no weight updates.
* A base model autocompletes; **post-training** (instruction tuning + preference tuning) makes it an assistant.  Base-vs-instruct pairs of open models let us see exactly what that changes.
