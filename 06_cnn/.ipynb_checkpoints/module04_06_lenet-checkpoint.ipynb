# Convolutional Neural Networks: the LeNet Architecture

We now have every ingredient needed to assemble a working CNN: the convolution operation and the principles behind it (notebook 2), control over output size (notebook 3), channels (notebook 4), and pooling (notebook 5).  This notebook puts them together, resulting in a more accurate and parsimonious model than our MLP from notebook 1.

In this notebook we present the **LeNet** architecture, among the first CNNs to attract wide attention for its performance on computer vision tasks.  It was introduced by Yann LeCun, then at AT&T Bell Labs:

> LeCun, Y., Bottou, L., Bengio, Y., & Haffner, P. (1998). Gradient-based learning applied to document recognition. *Proceedings of the IEEE*, 86(11), 2278–2324.

That paper was the culmination of a decade of work; LeCun's group had published the first successful training of CNNs by backpropagation nearly ten years earlier (LeCun et al., *Neural Computation* 1(4), 1989, 541–551).  At the time LeNet matched the performance of support vector machines, then the dominant approach in supervised learning, with an error rate below 1% per digit, and it was eventually deployed to read digits on deposits in ATMs.

We will build it exactly as published — sigmoid activations, average pooling, plain SGD — because the ways in which the 1998 design differs from what we would write today are themselves instructive.

```python
# render figures as crisp vector graphics (SVG) in the notebook
try:
    from matplotlib_inline.backend_inline import set_matplotlib_formats
    set_matplotlib_formats('svg')
except ImportError:
    pass

import matplotlib
matplotlib.rcParams['svg.fonttype'] = 'none'   # keep text as text in the SVG

import time

import numpy as np
import torch
from torch import nn
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
from matplotlib import pyplot as plt
```

```python
# if you have a GPU we should enable it
# mps is the pytorch device that supports GPU computations on MacOS

device = (
    'cuda' if torch.cuda.is_available()
    else 'mps' if torch.backends.mps.is_available()
    else 'cpu'
)
print(f'Using {device} device')
```

## The data

Fashion-MNIST: 60,000 training and 10,000 validation images, $28 \times 28$ grayscale, in ten clothing categories.  `transforms.ToTensor()` gives us `(1, 28, 28)` float tensors with values in $[0, 1]$ — note that the channel axis is already there, which is what a convolutional layer expects.

```python
batch_size = 128
data_dir = './data'
transform = transforms.ToTensor()

train_set = datasets.FashionMNIST(data_dir, train=True, download=True,
                                  transform=transform)
val_set = datasets.FashionMNIST(data_dir, train=False, download=True,
                                transform=transform)
train_iter = DataLoader(train_set, batch_size, shuffle=True)
val_iter = DataLoader(val_set, batch_size)

labels = train_set.classes
print(f'{len(train_set)} training and {len(val_set)} validation images, '
      f'shape {tuple(train_set[0][0].shape)}')
print('classes:', ', '.join(labels))
```

## The architecture

At a high level, **LeNet-5** has two parts: a *convolutional encoder* of two convolutional layers, and a *dense block* of three fully connected layers.

![Data flow in LeNet. The input is a handwritten digit, the output is a probability over 10 possible outcomes.](https://github.com/d2l-ai/d2l-en/raw/refs/heads/master/img/lenet.svg)

The basic unit of each convolutional block is a convolutional layer, a sigmoid activation, and an average pooling operation.  Both convolutions use a $5\times 5$ kernel; the first produces 6 output channels, the second 16.  Each $2\times2$ pooling with stride 2 quarters the spatial resolution.  This is the pattern notebooks 4 and 5 pointed at: resolution goes down, channel depth goes up.

Two of these choices are of their time.  **Sigmoid** activations were standard in 1998; ReLU and max-pooling, which work better, had not yet been established.  We keep the originals, and you will see the cost in the training curves below.

Passing the encoder's output to the dense block requires flattening each example in the minibatch: the four-dimensional `(batch, channels, height, width)` tensor becomes a two-dimensional one whose first axis indexes examples and whose second is the flattened representation of each.  The dense block then has three fully connected layers of 120, 84, and 10 units — the last matching the number of classes.

Here is the whole thing:

```python
def make_lenet(num_classes=10):
    """LeNet-5, as published: 5x5 kernels, sigmoid activations, average pooling."""
    return nn.Sequential(
        nn.LazyConv2d(6, kernel_size=5, padding=2), nn.Sigmoid(),
        nn.AvgPool2d(kernel_size=2, stride=2),
        nn.LazyConv2d(16, kernel_size=5), nn.Sigmoid(),
        nn.AvgPool2d(kernel_size=2, stride=2),
        nn.Flatten(),
        nn.LazyLinear(120), nn.Sigmoid(),
        nn.LazyLinear(84), nn.Sigmoid(),
        nn.LazyLinear(num_classes))
```

We have taken one liberty: the original's Gaussian output layer is replaced by the softmax we get from a linear layer plus cross-entropy loss.  This greatly simplifies the implementation, and Gaussian decoders are rarely used today.  Otherwise this matches LeNet-5.

Note also the `Lazy` layers.  `nn.LazyConv2d` and `nn.LazyLinear` infer the size of their input the first time they see data, so we need only state the *output* size of each.  The convenience is real — nobody wants to compute that the flattened encoder output is 400-dimensional by hand — but it means the parameters do not exist until the first forward pass.  We will run one deliberately before initializing or counting anything.

## Following the shapes

Let's push a single-channel $28 \times 28$ image through the network and print the shape after every layer.  This is the check that the architecture does what we think it does, and it is worth doing for any network you write.

![Compressed notation for LeNet-5.](https://github.com/d2l-ai/d2l-en/raw/refs/heads/master/img/lenet-vert.svg)

```python
def layer_summary(net, X_shape):
    """Print the output shape of every layer for an input of the given shape."""
    X = torch.randn(*X_shape)
    for layer in net:
        X = layer(X)
        print(f'{layer.__class__.__name__:12s} output shape: {tuple(X.shape)}')

model = make_lenet()
layer_summary(model, (1, 1, 28, 28))
```

The first convolutional layer uses two pixels of padding, which is the shape-preserving choice for a $5\times5$ kernel from notebook 3 — so $28 \times 28$ stays $28 \times 28$.  Incidentally, the $28 \times 28$ size of MNIST is itself an artifact of trimming two rows and columns from the original $32 \times 32$ scans, to save 30% of the space at a time when megabytes mattered.

The second convolutional layer forgoes padding, so height and width each are reduced by four pixels.  Each pooling layer halves both.  And as spatial extent shrinks, channel count grows: 1, then 6, then 16.  By the flatten operation, each image has become a 400-dimensional vector — $16 \times 5 \times 5$, which the dense block narrows to 120, 84, and finally 10.

Let's take a look at parameter counts to see the contribution of each layer to the overall count.  The code uses the same `named_parameters()` listing as in the transformer classifier.

```python
def parameter_summary(net):
    """Print the parameter count of every tensor in the network."""
    total = 0
    for name, param in net.named_parameters():
        print(f'{name:16s} {str(tuple(param.shape)):>18s} {param.numel():>10,d}')
        total += param.numel()
    print('-' * 48)
    print(f'{"total":16s} {"":18s} {total:>10,d}')
    return total

num_params = parameter_summary(model)
```

Around 62,000 parameters in total — and look at how they are distributed.  The two convolutional layers, which do nearly all the perceptual work, hold about 2,500 of them between them.  The single fully connected layer that follows the flatten holds 48,000, nearly 80% of the network.

That is the parameter-efficiency argument of notebook 1 made concrete: weight sharing makes convolution *extremely* cheap in parameters, while a dense layer over a 400-dimensional input is not cheap at all.  It is also a preview of where later architectures found their savings — the fully connected head is the first thing modern designs shrink, usually by replacing it with global average pooling.

For comparison, the MLP from notebook 1 had over 200,000 parameters.

### Initialization

Deep networks with sigmoid activations are sensitive to the scale at which their weights start, so we use **Xavier initialization**, which sets the initial scale from the number of inputs and outputs of each layer so that activations neither shrink nor blow up as they propagate.  `net.apply(fn)` walks the module tree and calls `fn` on every submodule.

```python
def init_cnn(module):
    """Xavier-initialize the weights of convolutional and linear layers."""
    if type(module) in (nn.Linear, nn.Conv2d):
        nn.init.xavier_uniform_(module.weight)
        nn.init.zeros_(module.bias)

model = make_lenet().to(device)
# the Lazy layers need to see data once before their parameters exist
model(next(iter(train_iter))[0].to(device))
model.apply(init_cnn);
```

## Training

The training loop is the same one we have used since the recurrent module — a model is a model, whatever is inside it.  We minimize cross-entropy with minibatch stochastic gradient descent, tracking training loss, validation loss and validation accuracy per epoch.

CNNs have fewer parameters than comparable MLPs but can cost *more* to compute, because each parameter participates in many more multiplications: the $150$ weights of the first convolutional layer are applied at all $784$ spatial positions.  If you have a GPU, this is a good time to use it.

```python
def accuracy(Y_hat, Y):
    """Fraction of correct predictions."""
    return (Y_hat.argmax(dim=1) == Y).float().mean()

optimizer = torch.optim.SGD(model.parameters(), lr=0.1)
loss_fn = nn.CrossEntropyLoss()
```

```python
epochs = 30
loss_train, loss_valid, accuracy_valid = [], [], []

for epoch in range(epochs):
    start = time.time()
    model.train()
    loss_values = []
    for batch_X, batch_y in train_iter:
        batch_X, batch_y = batch_X.to(device), batch_y.to(device)
        preds = model(batch_X)
        loss = loss_fn(preds, batch_y)
        loss_values.append(loss.item())
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
    loss_train.append(np.mean(loss_values))

    model.eval()
    loss_values, accuracy_values = [], []
    for batch_X, batch_y in val_iter:
        batch_X, batch_y = batch_X.to(device), batch_y.to(device)
        with torch.no_grad():
            preds = model(batch_X)
            loss_values.append(loss_fn(preds, batch_y).item())
            accuracy_values.append(accuracy(preds, batch_y).item())
    loss_valid.append(np.mean(loss_values))
    accuracy_valid.append(np.mean(accuracy_values))

    print(f'epoch {epoch + 1:2d}  train loss {loss_train[-1]:.3f}  '
          f'val loss {loss_valid[-1]:.3f}  '
          f'val accuracy {accuracy_valid[-1]:.3f}  '
          f'({time.time() - start:.1f} s)')
```

```python
# loss and accuracy live on different scales, so give each its own y-axis
fig, ax1 = plt.subplots(figsize=(5.5, 3.5))
epochs_range = range(1, len(loss_train) + 1)
ax1.plot(epochs_range, loss_train, color='C0', label='training loss')
ax1.plot(epochs_range, loss_valid, color='C1', label='validation loss')
ax1.set_xlabel('epoch')
ax1.set_ylabel('loss')
ax2 = ax1.twinx()
ax2.plot(epochs_range, accuracy_valid, color='C2',
         label='validation accuracy')
ax2.set_ylabel('accuracy')
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='center right');
```

The shape of these curves is the most interesting thing in the notebook, and it is a property of the 1998 design.

For the first several epochs the network sits near 10% accuracy — chance, for ten classes — and the loss barely moves.  Then it breaks free and improves rapidly.  This **plateau** is the sigmoid's doing.  A sigmoid saturates: for inputs far from zero its output is nearly flat, so its derivative is nearly zero, and gradients arriving from above are multiplied by that near-zero number at every one of the network's several sigmoid layers.  Until the weights happen to move into the regime where the units are not saturated, almost no gradient signal reaches the early convolutional layers, and the network learns nothing.  Xavier initialization is what makes the plateau a few epochs long rather than effectively permanent.

This is the concrete form of the *vanishing gradient* problem, and dealing with it is much of what separates the architectures of the 1990s from those that followed.  ReLU does not saturate for positive inputs, so its derivative is exactly 1 there and gradients pass through undiminished; max-pooling routes the gradient to the single unit that produced the maximum rather than spreading it thin. Swapping both into `make_lenet` — two edits — removes the plateau entirely.  Exercise 2 asks you to try it.

## Looking at the predictions

Accuracy is a summary; it helps to see individual cases.  Below, the model's prediction for a batch of validation images, with mistakes labelled by both the prediction and the truth:

```python
model.eval()
batch_X, batch_y = next(iter(val_iter))
with torch.no_grad():
    preds = model(batch_X.to(device)).argmax(dim=1).cpu()

fig, axes = plt.subplots(2, 8, figsize=(12, 3.6))
for ax, image, y, pred in zip(axes.flatten(), batch_X, batch_y, preds):
    ax.imshow(image.squeeze(), cmap='Greys', interpolation='nearest')
    correct = (y == pred)
    title = labels[pred] if correct else f'{labels[pred]}\n(is {labels[y]})'
    ax.set_title(title, fontsize=7.5, color='black' if correct else 'C3')
    ax.axis('off')
fig.tight_layout()
```

The confusions are worth reading: shirts, coats, and pullovers get mixed up with each other far more often than any of them gets mixed up with a sandal.  The classes the model finds hard are the ones that are genuinely similar at $28 \times 28$ resolution.

## What the first layer learned

The first convolutional layer has six $5 \times 5$ kernels operating directly on pixels, so we can look at them.  These are the learned equivalents of the hand-designed edge detector from notebook 2:

```python
kernels = model[0].weight.data.cpu()          # (6, 1, 5, 5)
image = val_set[0][0]                          # one validation image

fig, axes = plt.subplots(2, 7, figsize=(11, 3.4))
axes[0, 0].imshow(image.squeeze(), cmap='Greys', interpolation='nearest')
axes[0, 0].set_title('input', fontsize=8)
axes[1, 0].axis('off')
with torch.no_grad():
    feature_maps = model[0](image.unsqueeze(0).to(device)).cpu()[0]
for i in range(6):
    axes[0, i + 1].imshow(kernels[i, 0], cmap='RdBu', interpolation='nearest')
    axes[0, i + 1].set_title(f'kernel {i}', fontsize=8)
    axes[1, i + 1].imshow(feature_maps[i], cmap='Greys', interpolation='nearest')
    axes[1, i + 1].set_title(f'feature map {i}', fontsize=8)
for ax in axes.flatten():
    ax.axis('off')
fig.tight_layout()
```

Top row: the six kernels, red for positive weights and blue for negative.  Bottom row: what each produces on the image above.  Several are recognizable as oriented edge or contrast detectors — a bright side and a dark side, the two-dimensional relatives of the $[1, -1]$ filter we designed by hand and then learned in notebook 2.  Nobody told the network to look for edges.  It found them because edges are what is worth measuring in images, which is the same conclusion Hubel and Wiesel reached about the visual cortex.

## Summary

We have covered a lot of ground in this module: from the MLPs of the 1980s to the CNNs of the 1990s, derived from first principles and assembled into a working network.

* **LeNet-5** is a convolutional encoder — convolution, activation, pooling, twice — followed by a dense block of three fully connected layers.  The architecture remains meaningful today; on Fashion-MNIST it is much closer to a modern residual network than to an MLP.
* Resolution falls and channel depth rises as the network deepens, the pattern established in notebooks 3 through 5.
* Convolution is remarkably parameter-efficient: the two convolutional layers hold about 4% of the network's 62,000 parameters, while the first dense layer holds 80%.  This imbalance is what later architectures attack.
* Sigmoid activations and average pooling produce a multi-epoch plateau at chance accuracy — a visible instance of vanishing gradients, and the reason ReLU and max-pooling became standard.
* The first layer learns oriented edge detectors on its own, reproducing what we hand-designed in notebook 2.
* Finally, note the *effort*: what once took months of C++ and assembly, and a bespoke Lisp-based deep learning system, is now forty lines of PyTorch.  That productivity gain is as much a part of the story as the architecture itself.
