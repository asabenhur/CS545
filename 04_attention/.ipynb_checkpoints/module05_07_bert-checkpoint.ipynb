# BERT: Pretraining an Encoder Model

In the previous notebook we built a transformer *encoder-only* classifier that uses a stack of encoder blocks, and a readout from the representation of the `<cls>` token into a linear classifier.  We trained it from scratch on 25,000 IMDb reviews and it worked.  But it is a narrow specialist, and its knowledge of English is only whatever those reviews could teach it.  Rare words never got enough gradient to mean anything (we saw this in the RNN module), and very little of what it learned will transfer to other tasks.

**BERT** (Bidirectional Encoder Representations from Transformers; Devlin et al., 2018) uses the same architecture: encoder-only, `<cls>` readout, with two differences: it is *bigger* (12 layers and a 768-dimensional model), and, crucially, it is *pretrained* on billions of words of general text before it ever sees a labeled task.  The `<cls>` token we introduced in notebook 1 as a sequence summary was the innovation in BERT's design.

In this notebook we will load a real, pretrained BERT model and use it to beat our from-scratch IMDb classifier.  We finish by looking beyond encoder-only models toward the large language models that are the next topic.

## The pretraining objective: masked language modeling

BERT is trained in a self-supervised manner on a large corpus of text without using any external labels.  But how can a model learn with no labels?  BERT's answer is **masked language modeling** (MLM): take a piece of text, and hide a random subset of tokens behind a `[MASK]` symbol, and train the encoder to predict those hidden tokens using the words on both sides as context (this bidirectionality is what the "B" in BERT stands for).  Every sentence in every book becomes free supervision, so the training data is effectively unlimited.

The following figure demonstrates this concept:

## Loading a pretrained model

Now let's watch BERT in action.  We use the Hugging Face `transformers` library - the standard way to load pretrained models.

**Models live on the Hub.**  `MODEL_NAME = "distilbert-base-uncased"` is an identifier on the [Hugging Face Hub](https://huggingface.co/models), a public registry of pretrained models.  The name follows a simple convention: the *architecture* (`distilbert`), the *size* (`base` or `large`), and a training detail — `uncased` means the tokenizer lowercases everything, so this model cannot tell *Apple* from *apple*.  **DistilBERT** is a compressed version of BERT, trained to imitate the full model: about 97% of the accuracy at 60% of the size, and is usable on a laptop.  The first call downloads the weights (around 250 MB) into a local cache under `~/.cache/huggingface`; later calls read them from disk.

**`pipeline` is the library's high-level interface.**  Putting a pretrained model to work takes three pieces: a *tokenizer* to turn text into token ids, the *model* with an output head suited to the task, and *post-processing* to turn logits back into something readable.  `pipeline` assembles all three from a task name and a model name and returns a callable object:

```python
fill = pipeline('fill-mask', model=MODEL_NAME, device=device)
```

The task string `'fill-mask'` selects which **head** to attach — here the masked-language-model head, the same softmax over the vocabulary that BERT was pretrained with, which is why no training is needed to run this.  It tells the pipeline to look for the special `[MASK]` token in the input.  The output is a list of candidate completions ordered by probability, each one a dictionary.

````python
# render figures as crisp vector graphics (SVG) in the notebook
try:
    from matplotlib_inline.backend_inline import set_matplotlib_formats
    set_matplotlib_formats('svg')
except ImportError:
    pass

# one-time install if needed:  pip install transformers
import os
import re
import tarfile
import urllib.request

import numpy as np
import torch
from torch import nn
from matplotlib import pyplot as plt

device = ('cuda' if torch.cuda.is_available()
          else 'mps' if torch.backends.mps.is_available() else 'cpu')
print(f'Using {device} device')

MODEL_NAME = "distilbert-base-uncased"
````

````python
from transformers import pipeline

fill = pipeline('fill-mask', model=MODEL_NAME, device=device)
# device is 'mps' on Apple Silicon; switch to device=-1 (CPU) if MPS errors out
for sentence in ['this movie was absolutely [MASK].',
                 'the acting was [MASK] and the plot made no sense.',
                 'a [MASK] is a small burrowing mammal.']:
    print(sentence)
    for p in fill(sentence, top_k=5):
        print(f'    {p["token_str"]:12s} {p["score"]:.3f}')
        # That p["token_str"] and p["score"] access a token and its score;
        # `top_k=5` is how many candidates to ask for.

    print()
````

Two things are worth noticing.  First, the predictions are *sensible*: BERT completes "absolutely ___" with words like *fantastic*, *hilarious*, *terrible* — it has learned which words fit.  Second, it fills "a ___ is a small burrowing mammal" with *mole* (or *rabbit*, *rodent*), pulling on context from *both* directions.  This is exactly the disambiguation-from-context idea from notebook 2, now learned from data at scale.  None of this required a labeled dataset; it fell out of predicting masked words over a huge corpus.

That is the whole trick of pretraining: a task you can generate for free (fill the blank) forces the model to build general representations of language, which we can then repurpose.

````python
# a pipeline is a thin wrapper -- look inside it
print('model     :', type(fill.model).__name__)
print('tokenizer :', type(fill.tokenizer).__name__)
print('mask token:', fill.tokenizer.mask_token,
      '(id', fill.tokenizer.mask_token_id, ')')
print('vocabulary:', f'{fill.tokenizer.vocab_size:,}', 'tokens')
````

A practical note.  The mask token is **model-specific**: BERT and DistilBERT spell it `[MASK]`, RoBERTa spells it `<mask>`.  Asking the tokenizer, as above, is the reliable way to find out.

### How BERT was actually trained

Here is the method that produced the trained BERT model we experimented with; several of the choices are instructive.

**Data.**  BERT was pretrained on the **BooksCorpus** (\~800M words) plus **English Wikipedia** (\~2.5B words) — about **3.3 billion words of unlabeled text**.  No annotation: the supervision is manufactured from the text itself, which is what lets pretraining use essentially unlimited data.

**Objective 1 — masked language modeling, and the 80/10/10 trick.**  BERT selects **15%** of the tokens in each sequence for prediction.  It would be natural to replace all of them with `[MASK]`, but that creates a subtle train/serve mismatch: `[MASK]` never appears in the real text you later fine-tune on, so a model that only ever predicts *at* `[MASK]` symbols is being trained for an input it will not see again.  BERT's fix: of the selected tokens, **80% become `[MASK]`, 10% are replaced by a random token, and 10% are left unchanged** — and the model is asked to predict the original in all three cases.  Because it cannot tell which token was corrupted and which is genuine, it is forced to build a useful representation at *every* position, not just decode the `[MASK]` symbol.

**Objective 2 — next-sentence prediction (and its retirement).**  Original BERT was trained on a second task at the same time: given two segments packed as `[CLS] A [SEP] B [SEP]`, predict whether B genuinely followed A in the corpus, the idea being to teach inter-sentence relationships for downstream pair tasks (NLI, QA).  Later work (**RoBERTa**) showed NSP contributes little and *dropped it entirely* with no loss — and **DistilBERT**, the model we used earlier, has no NSP either.  This is exactly why its raw `[CLS]` vector is an unremarkable summary until a task trains it.

**Tokenization.**  BERT uses **WordPiece** subword tokens with a \~30,000-token vocabulary, so a rare or unseen word is split into known pieces (`playing` -> `play` + `##ing`).  There is effectively **no out-of-vocabulary problem**.  This is in contrast with the fixed word vocabulary and `<unk>` bucket we used in the RNN module.

**Size and optimization.**  BERT-base has 12 layers, a hidden layer size of 768, 12 heads, with a total of **110M parameters** (BERT-large: 24/1024/16, 340M).  It was trained with **Adam**, a peak learning rate of 1e-4 under a **warmup-then-linear-decay** schedule, weight decay, dropout value of 0.1, GELU activations, and batches of 256 sequences x 512 tokens for **\~1 million steps** (\~40 passes over the corpus).  Training took roughly four days on TPUs.  This is characteristic for training transformers.

**DistilBERT.**  DistilBERT is a 6-layer, **66M-parameter** student trained by *knowledge distillation*: it learns to match BERT-base's output distribution (soft targets) alongside the MLM objective, dropping NSP.  It keeps \~97% of BERT-base's benchmark score at \~60% of the size and about twice the speed, which is why the cells here run on a laptop.

**A few subtleties worth knowing.**

* **Contextual, not static, embeddings.**  Unlike GloVe (one fixed vector per word, as the CNN used), BERT gives each token a vector that depends on the *whole* sentence — *bank* in "river bank" and "savings bank" come out different.  That contextualization is the entire reason to run the encoder rather than just look words up in a table.
* **Learned positional embeddings, not sinusoids.**  In notebook 4 we built *sinusoidal* position encodings for their relative-offset and extrapolation properties; BERT instead simply **learns** a separate position vector for each of its 512 slots.  Simpler and effective, but it hard-caps inputs at 512 tokens and does not extrapolate beyond them — a real limitation that later models revisit (RoPE, ALiBi).
* **Bidirectional means BERT cannot generate.**  Because every token attends to the future as well as the past, BERT is built to *read* a whole input at once, not to write text left to right.  That is precisely why the next topic — generative LLMs — switches to *causal* (decoder-only) attention, where a token may see only what came before it.

## Reusing BERT for sentiment

Here is the payoff.  BERT has already learned to turn a piece of text into a rich vector — so instead of learning that from 25,000 reviews as we did in notebook 6, we can *borrow* it.  The simplest way, and the one we use here, is **feature extraction**: run each review through the frozen (untrained-by-us) BERT, read a single fixed vector out of its output for each review, and train only a small linear classifier on that.  BERT's weights never change; all the learning happens in one `Linear` layer.  (Which vector to read out — BERT's `[CLS]` slot, or an average over the tokens — turns out to matter, and we compare the two.)

First, the IMDb reviews as raw text (BERT brings its own tokenizer, so we skip the vocabulary we built earlier):

````python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

BLUE, GREEN, IO = '#cdddec', '#d0e3d4', '#ededed'
MASKC, AMBER, TX = '#f0c674', '#c0662f', '#333333'

fig, ax = plt.subplots(figsize=(6.8, 4.4))
toks = ['[CLS]', 'a', 'brilliant', 'film']
BW = 0.9
for i, t in enumerate(toks):
    c = MASKC if t == '[CLS]' else IO
    ax.add_patch(FancyBboxPatch((i - BW/2, -0.28), BW, 0.56,
                 boxstyle='round,pad=0.02', facecolor=c, edgecolor='none'))
    ax.text(i, 0, t, ha='center', va='center', fontsize=8, color=TX,
            fontweight='bold' if t == '[CLS]' else 'normal')
for i in range(len(toks)):
    ax.annotate('', xy=(i, 0.85), xytext=(i, 0.30),
                arrowprops=dict(arrowstyle='-', color='0.6', lw=0.9))
ax.add_patch(FancyBboxPatch((-0.55, 0.85), 4.1, 0.75, boxstyle='round,pad=0.02',
             facecolor=BLUE, edgecolor='none'))
ax.text(1.5, 1.22, 'pretrained BERT encoder', ha='center', va='center',
        fontsize=8.5, color=TX)
ax.annotate('', xy=(0, 2.35), xytext=(0, 1.6),
            arrowprops=dict(arrowstyle='-|>', color=AMBER, lw=1.8))
ax.text(0.15, 1.97, 'the [CLS] vector\n(768-dim summary)', ha='left',
        va='center', fontsize=7, color=AMBER)
ax.add_patch(FancyBboxPatch((-0.9, 2.35), 1.8, 0.6, boxstyle='round,pad=0.02',
             facecolor=GREEN, edgecolor='none'))
ax.text(0, 2.65, 'Linear head', ha='center', va='center', fontsize=8, color=TX)
ax.annotate('', xy=(0, 3.5), xytext=(0, 2.95),
            arrowprops=dict(arrowstyle='-|>', color='0.55', lw=1.5))
ax.text(0, 3.62, 'neg / pos', ha='center', va='bottom', fontsize=8.5, color=TX)
ax.text(3.75, 1.22, 'feature extraction:\nweights FROZEN,\ntrain only the head',
        ha='left', va='center', fontsize=7, color='#3a6ea5')
ax.text(3.75, 2.62, 'fine-tuning:\ngradients flow\ninto BERT too', ha='left',
        va='center', fontsize=7, color='#b5651d')
#ax.annotate('', xy=(-0.78, 1.1), xytext=(-0.78, 2.6),
#            arrowprops=dict(arrowstyle='<->', color='#b5651d', lw=1.0,
#                            linestyle='dashed', alpha=0.7))
ax.set_xlim(-1.7, 6.8); ax.set_ylim(-0.9, 3.95); ax.axis('off')
ax.set_title('Reusing a pretrained BERT for classification', fontsize=9)
plt.tight_layout();
````

````python
def download_imdb(data_dir='../data'):
    url = 'https://d2l-data.s3-accelerate.amazonaws.com/aclImdb_v1.tar.gz'
    os.makedirs(data_dir, exist_ok=True)
    archive = os.path.join(data_dir, 'aclImdb_v1.tar.gz')
    if not os.path.exists(archive):
        print('downloading', url, '...'); urllib.request.urlretrieve(url, archive)
    extracted = os.path.join(data_dir, 'aclImdb')
    if not os.path.exists(extracted):
        with tarfile.open(archive) as tar:
            tar.extractall(data_dir)
    return extracted

def read_imdb(data_dir, is_train):
    data, labels = [], []
    for label in ('pos', 'neg'):
        folder = os.path.join(data_dir, 'train' if is_train else 'test', label)
        for file in os.listdir(folder):
            with open(os.path.join(folder, file), 'rb') as f:
                data.append(f.read().decode('utf-8').replace('<br />', ' '))
                labels.append(1 if label == 'pos' else 0)
    return data, labels

def subsample(texts, labels, n, seed=0):
    """A balanced random subset of size n (encoding all 25k on CPU is slow)."""
    rng = np.random.default_rng(seed)
    idx = rng.permutation(len(texts))[:n]
    return [texts[i] for i in idx], [labels[i] for i in idx]

data_dir = download_imdb()
train_texts, train_labels = subsample(*read_imdb(data_dir, True), n=2000)
test_texts, test_labels = subsample(*read_imdb(data_dir, False), n=2000)
print(len(train_texts), 'train /', len(test_texts), 'test reviews')
````

Now the encoder.  We load DistilBERT and its tokenizer, put it in eval mode, and run the reviews through it *without* gradients — BERT's weights never change.  That leaves one question: how to turn each review's *sequence* of token vectors into the single fixed vector a classifier needs.

For now we take the **`[CLS]` vector** — the final-layer output at position 0 — because it is the slot BERT reserves for exactly this purpose, and the direct descendant of the `<cls>` trick we have used all module.  It is one choice among several, and whether it is a good one is a question we come back to later in the notebook, once there is something to compare it against.

For the purpose of classifier training we need a different `transformers` API which provides access to the token representation produced by BERT.  We will make use of `AutoTokenizer` and `AutoModel`.

The `Auto` prefix names no architecture: it means *look up which architecture this identifier refers to, and build the matching class* — here the tokenizer and model associated with  `bert-base-uncased`.
`AutoModel` returns the **bare encoder**: it outputs hidden states and carries no task head, which is exactly what we need here.   The `AutoModelFor...` variants attach a head instead — the fill-mask pipeline chose `DistilBertForMaskedLM` for us, and we will ask for `AutoModelForSequenceClassification` when we fine-tune.

````python
from transformers import AutoTokenizer, AutoModel

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
bert = AutoModel.from_pretrained(MODEL_NAME).to(device).eval()

@torch.no_grad()
def embed(texts, batch_size=16, max_len=256):
    """One frozen DistilBERT vector per review: the [CLS] position."""
    feats = []
    for i in range(0, len(texts), batch_size):
        enc = tokenizer(texts[i:i + batch_size], padding=True, truncation=True,
                        max_length=max_len, return_tensors='pt').to(device)
        h = bert(**enc).last_hidden_state               # (B, T, 768)
        feats.append(h[:, 0].cpu())                     # the [CLS] position
    return torch.cat(feats)

cls_train = embed(train_texts)
cls_test = embed(test_texts)
y_train = torch.tensor(train_labels)
y_test = torch.tensor(test_labels)
print('feature shape:', cls_train.shape)   # (2000, 768)
````

Each review is now a point in a 768-dimensional space that BERT arranged.  All that remains is to separate positive from negative with a single linear layer; nothing about BERT is updated.

````python
def train_probe(X_train, X_test, epochs=500, lr=1e-3):
    """Train a linear classifier on fixed features; return test accuracy."""
    clf = nn.Linear(X_train.shape[1], 2)
    opt = torch.optim.Adam(clf.parameters(), lr=lr)
    loss_fn = nn.CrossEntropyLoss()
    for _ in range(epochs):
        opt.zero_grad()
        loss_fn(clf(X_train), y_train).backward()
        opt.step()
    with torch.no_grad():
        return (clf(X_test).argmax(1) == y_test).float().mean().item()

acc_cls = train_probe(cls_train, cls_test)
print(f'[CLS] feature  test accuracy: {acc_cls:.3f}')
````

About 0.83 is roughly the linear-separability ceiling of *frozen* DistilBERT features for this task, and it is close to what our from-scratch model reached in notebook 6.  So on accuracy, frozen features do not really *beat* training from scratch — what they buy is **data efficiency and simplicity**: we got there with 2,000 reviews and a single linear layer, where the from-scratch model used all 25,000 and a full transformer training loop.

Whether `[CLS]` was the right place to read from is a fair question, and one we take up later in the notebook.

The accuracy win comes from **fine-tuning**: letting gradients flow back into DistilBERT so the representation itself adapts to sentiment.  The `[CLS]` token, finally given a task signal, becomes the summary it was designed to be.  The cell below does this the *realistic* way — AdamW with a learning-rate **warmup and linear decay**, gradient clipping, and more training data than the frozen probe used.  All three matter: fine-tuning is data-hungry, and without the schedule it is unstable — a naive constant-rate fine-tune can actually get *worse* after the first epoch.  Keep in mind that this will take a while on a CPU.

````python
from transformers import (AutoModelForSequenceClassification,
                          get_linear_schedule_with_warmup)

# OPTIONAL, and the slow path: fine-tune DistilBERT end-to-end, the realistic
# way -- AdamW with weight decay, a learning-rate *warmup + linear decay*, and
# gradient clipping.  (Skipping the schedule is why a naive fine-tune can get
# WORSE in later epochs.)  Fine-tuning also needs more data than the frozen
# probe, so we draw a larger sample; set n_ft = 25000 on a GPU for the full
# result.  On CPU this is slow -- a GPU / Apple-MPS is strongly recommended.
n_ft = 5000
ft_texts, ft_labels = subsample(*read_imdb(data_dir, is_train=True),
                                n=n_ft, seed=1)
ft_y = torch.tensor(ft_labels)

ft_model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME, num_labels=2).to(device)
opt = torch.optim.AdamW(ft_model.parameters(), lr=2e-5, weight_decay=0.01)
loss_fn = nn.CrossEntropyLoss()
bs, epochs = 16, 3
total_steps = (len(ft_texts) + bs - 1) // bs * epochs
scheduler = get_linear_schedule_with_warmup(
    opt, num_warmup_steps=int(0.1 * total_steps),
    num_training_steps=total_steps)

def encode(texts):
    return tokenizer(texts, padding=True, truncation=True,
                     max_length=256, return_tensors='pt').to(device)

for epoch in range(epochs):
    ft_model.train()
    order = torch.randperm(len(ft_texts)).tolist()
    for i in range(0, len(order), bs):
        idx = order[i:i + bs]
        loss = loss_fn(ft_model(**encode([ft_texts[j] for j in idx])).logits,
                       ft_y[idx].to(device))
        opt.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(ft_model.parameters(), 1.0)
        opt.step()
        scheduler.step()
    ft_model.eval()
    correct = 0
    with torch.no_grad():
        for i in range(0, len(test_texts), bs):
            pred = ft_model(**encode(test_texts[i:i + bs])).logits.argmax(1).cpu()
            correct += (pred == y_test[i:i + bs]).sum().item()
    print(f'epoch {epoch + 1}: fine-tuned test accuracy '
          f'{correct / len(test_texts):.3f}')
````

This results improves on both the frozen probe *and* the strong TF-IDF baseline from the recurrent module.  That gain is the result of the representation *adapting* to the task instead of being read off from the frozen pre-trained model.

## Was it the readout, or the representation?

Fine-tuning beat the frozen probe by several percentage points, and there you might be tempted to explain this as follows: `[CLS]` was never trained to summarize anything — masked language modeling never asks it to predict a token, and DistilBERT dropped even BERT's next-sentence objective — so reading from it underperforms, and fine-tuning repairs it.

But perhaps the information is there and we simply did not read it in the best way:
`[CLS]` and the mean are simple aggregation methods that have their issues.  Next, we will explore a third approach that uses the frozen encoder's token vectors, and lets the model learn **how to combine them**.
The tool is **attention pooling** — a standard way to reduce a variable-length sequence of vectors to one vector.

Attention pooling uses a single query, which is a learn vector $\mathbf{q}$.
It then scores the query against each key (the output of the encoder), softmaxes the scores into weights, and returns the weighted average:

$$\alpha_t \;=\; \frac{\exp\!\big(\mathbf{q}^\top \mathbf{h}_t / \sqrt{d}\big)}
{\sum_{s} \exp\!\big(\mathbf{q}^\top \mathbf{h}_s / \sqrt{d}\big)},
\qquad
\mathbf{z} \;=\; \sum_t \alpha_t \, \mathbf{h}_t$$

The keys and values are both the token vectors $\mathbf{h}_t$ themselves, so the entire pooling head is one vector $\mathbf{q}$ of $d$ parameters, asking every token the same question: *how much should you count?*

Notice that attention pooling **strictly contains** both aggregation rules.  A query of all zeros scores every token equally and reproduces mean pooling; a query that drives all the softmax mass onto position 0 reproduces the `[CLS]` readout.

We will now check what the extra flexibility buys in accuracy.

First, the frozen states.  Until now we stored one vector per review; attention pooling needs *all* of them, which is a different memory problem — two thousand reviews at 256 tokens of 768 floats would be over a gigabyte per split.  So we shorten reviews to 128 tokens and store half precision, and recompute every pooling from this one cache so the three are exactly comparable.

````python
# keep EVERY token vector, not one summary per review -- so memory matters:
# 128 tokens instead of 256, and float16 instead of float32.
POOL_LEN = 128

@torch.no_grad()
def encode_tokens(texts, batch_size=16, max_len=POOL_LEN):
    """Frozen DistilBERT token vectors, padded to max_len.

    Returns (H, mask): H is (n, max_len, 768) and mask is (n, max_len) with
    1 at real tokens and 0 at padding."""
    Hs, masks = [], []
    for i in range(0, len(texts), batch_size):
        enc = tokenizer(texts[i:i + batch_size], padding='max_length',
                        truncation=True, max_length=max_len,
                        return_tensors='pt').to(device)
        h = bert(**enc).last_hidden_state          # (B, max_len, 768)
        Hs.append(h.half().cpu())                  # half precision: 2 bytes/number
        masks.append(enc['attention_mask'].cpu())
    return torch.cat(Hs), torch.cat(masks)


H_train, M_train = encode_tokens(train_texts)
H_test, M_test = encode_tokens(test_texts)
print('token states:', tuple(H_train.shape), 'per split,',
      f'{H_train.numel() * H_train.element_size() / 1e6:.0f} MB each')
````

Next, we train three heads that differ only in how they pool.  Each uses the same `nn.Linear(768, 2)`, readout and all three are trained by the same loop on the same frozen states, so any difference between them is a result of the pooling.

(These numbers will not match the probes earlier in the notebook, which used 256-token inputs and full-batch training.  That is why we recompute all three here.)

````python
import math

class ClsPooling(nn.Module):
    """Read position 0 and classify.  The rule we have been using."""
    def __init__(self, num_hiddens, num_classes=2):
        super().__init__()
        self.classifier = nn.Linear(num_hiddens, num_classes)

    def forward(self, H, mask):
        return self.classifier(H[:, 0]), None


class MeanPooling(nn.Module):
    """Average the real tokens and classify."""
    def __init__(self, num_hiddens, num_classes=2):
        super().__init__()
        self.classifier = nn.Linear(num_hiddens, num_classes)

    def forward(self, H, mask):
        m = mask.unsqueeze(-1).to(H.dtype)                  # (B, T, 1)
        pooled = (H * m).sum(1) / m.sum(1).clamp(min=1)
        return self.classifier(pooled), None


class AttentionPooling(nn.Module):
    """Attention with a single *learned* query.

    Attention so far has taken its queries from the sequence itself.  Here
    there is one query vector, a parameter, the same for every input: it asks
    one fixed question of every token.  The keys and the values are the token
    vectors themselves (no projections), so the whole head is one vector plus
    the classifier."""
    def __init__(self, num_hiddens, num_classes=2):
        super().__init__()
        self.query = nn.Parameter(torch.randn(num_hiddens) * 0.02)
        self.classifier = nn.Linear(num_hiddens, num_classes)

    def forward(self, H, mask):
        # one score per position: how much does this token match the query?
        scores = (H @ self.query) / math.sqrt(H.shape[-1])          # (B, T)
        scores = scores.masked_fill(mask == 0, float('-inf'))       # skip padding
        weights = torch.softmax(scores, dim=-1)                     # (B, T)
        pooled = (weights.unsqueeze(-1) * H).sum(dim=1)             # (B, num_hiddens)
        return self.classifier(pooled), weights


def train_head(head, epochs=30, batch_size=64, lr=1e-3, seed=0):
    """Train a pooling head on the frozen token states; return it and accuracy."""
    torch.manual_seed(seed)
    head = head.to(device)
    opt = torch.optim.Adam(head.parameters(), lr=lr)
    loss_fn = nn.CrossEntropyLoss()
    for _ in range(epochs):
        head.train()
        for idx in torch.randperm(len(H_train)).split(batch_size):
            H = H_train[idx].to(device).float()          # back to float32 to train
            opt.zero_grad()
            logits, _ = head(H, M_train[idx].to(device))
            loss_fn(logits, y_train[idx].to(device)).backward()
            opt.step()
    head.eval()
    correct = 0
    with torch.no_grad():
        for idx in torch.arange(len(H_test)).split(batch_size):
            logits, _ = head(H_test[idx].to(device).float(),
                             M_test[idx].to(device))
            correct += (logits.argmax(1).cpu() == y_test[idx]).sum().item()
    return head, correct / len(H_test)


D = H_train.shape[-1]
_, acc_cls_tok = train_head(ClsPooling(D))
_, acc_mean_tok = train_head(MeanPooling(D))
pooler, acc_attn = train_head(AttentionPooling(D))

print(f'[CLS], fixed              {acc_cls_tok:.3f}')
print(f'mean, fixed               {acc_mean_tok:.3f}')
print(f'attention, learned        {acc_attn:.3f}')
````

Three things to take from these numbers.

**Truncation costs more than any pooling choice.**  The `[CLS]` probe scored 0.834 earlier on 256-token inputs and 0.792 here on 128 — four points for the second half of each review.

**`[CLS]` against the mean could be noise.**  0.792 and 0.781.  On 2,000 test reviews it's unclear if this difference is significant.

**Learning the pooling is worth about 1.6 points.**  Attention pooling reaches 0.808.  This suggests that `[CLS]` is not a trained summary, and learning the readout helps; but most of fine-tuning's advantage came from changing the representation, not from where it is read.

The weights are also instructive as we see next.

````python
# what did the single learned query decide to look at?
enc = tokenizer(test_texts[0], truncation=True, max_length=POOL_LEN,
                return_tensors='pt').to(device)
with torch.no_grad():
    H = bert(**enc).last_hidden_state
    _, w = pooler(H, enc['attention_mask'])

toks = tokenizer.convert_ids_to_tokens(enc['input_ids'][0].tolist())
w = w[0].cpu()
n_real = int(enc['attention_mask'].sum())
top = w.topk(12).indices.sort().values          # the heaviest, in reading order

print(f'true label: {"positive" if y_test[0] == 1 else "negative"}')
print(f'the 12 highest-weighted of {n_real} tokens:\n')
print('   '.join(f'{toks[i]} ({w[i]:.3f})' for i in top))
print(f'\nweight on [CLS]: {w[0]:.3f}       uniform would be {1 / n_real:.3f}')
print(f'weight on the top 12: {w[top].sum():.2f} of 1.00')
````

That is not what most people expect.

**The heaviest tokens are function words.**  `is` takes the largest single weight (0.067) and appears five times in the top twelve, alongside `the` and `it`; `terrible` — the one unambiguously evaluative word in the list ranks below several of them.  A pooler trained to detect sentiment is apparently not looking at the sentiment words.

To understand this observation recall that vectors are **built from context**.  After six layers of self-attention, the vector sitting at the position of `is` in *this is terrible* has already absorbed `terrible`.  That is exactly what the "bank" experiment demonstrated.  The pooler has no need to point at the evaluative word, because a position adjacent to it already carries the clause's meaning.  In a contextual model, *where the information is* and *where the word is* are different.  There is a second reason function words are attractive: they are semantically light, so their slot in the residual stream is relatively free to hold a summary of the material around it, which is precisely the argument for why `[CLS]` ought to have been a good place to read.

**`[CLS]` gets less weight than an average token.**  Offered a choice of where to read, the head puts *below-average* weight on the slot BERT nominally reserves for summarizing the sequence.  However, nothing in DistilBERT's pretraining asked `[CLS]` to summarize anything, and a head that is free to look elsewhere does.

**A caveat, the same one the attention maps came with.**  The weights show where information is being drawn *from*, not what is computed with it, and the distribution here is spread out rather than peaked — the top twelve of 128 positions hold 0.32 of the mass, about 3.4 times that of uniform. This is a diffuse read of the sequence, not a spotlight on a few words.

## Beyond BERT: toward large language models

BERT is one of three ways to use the transformer, distinguished only by *which half* of the original encoder–decoder they keep and how attention is allowed to flow:

* **Encoder-only** (BERT): bidirectional attention — every token sees every other — trained to fill in masked tokens.  Ideal for *understanding* a whole input at once and useful for sequence classification, tagging, and retrieval.  This is what we just built.
* **Encoder–decoder** (e.g. T5): an encoder reads the input, a decoder generates an output sequence, useful when the output is itself text, as in translation or summarization.
* **Decoder-only** (e.g. the GPT family): a single stack with *causal* attention — each token may attend only to earlier ones — trained simply to predict the next token.  Because "predict the next token" lets the model *generate* text, and because this objective scales smoothly with more data and parameters, decoder-only models are the architecture behind today's large language models.

The step from BERT to an LLM is mostly a change of objective and scale, not of mechanism: the attention, the residual blocks, the layer norms are the same pieces we built by hand.  What changes is that the model is trained to continue text rather than to fill blanks, and is grown by orders of magnitude — at which point new behaviors (following instructions, in-context learning, chain-of-thought reasoning) begin to appear.  That is where the course goes next.

## Summary

* BERT is the encoder-only transformer we built, scaled up and **pretrained** on huge unlabeled text via **masked language modeling** — predicting hidden tokens from bidirectional context.  Its `[CLS]` slot is the `<cls>` trick we used all module.
* As a **frozen feature extractor**, BERT matches our from-scratch model with far less data and no transformer training — an efficiency win, and one with a ceiling: a linear probe can do no better than the fixed features allow.
* **Attention pooling** — one learned query, a softmax over positions, a weighted average — is the general way to turn a sequence into a single vector, and it contains both `[CLS]` and mean pooling as special cases.  Applied to *frozen* features it beats either fixed rule by a point or two, recovering roughly a quarter of what fine-tuning gains: the readout is part of the story, the representation is more of it.  Its learned weights fall on *contextualized* positions rather than on sentiment words, and below uniform on `[CLS]`.
* **Fine-tuning** — updating BERT's weights on the task — is where pretraining decisively wins on accuracy, and where `[CLS]` finally becomes a strong summary.
* Encoder-only (BERT), encoder–decoder (T5), and decoder-only (GPT) differ mainly in attention direction and objective; the decoder-only, next-token line scales into large language models — the next topic.
