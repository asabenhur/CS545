# Preparing your system for this course

In order to get up and running, we will need an environment for running Python, the Jupyter Notebook, the relevant libraries, and support code for the book.

## Installing Miniconda

Your simplest option is to install
[Miniconda](https://conda.io/en/latest/miniconda.html).
Note that the Python 3.x version is required.
You can skip the following steps
if your machine already has conda installed.

Visit the Miniconda website and determine
the appropriate version for your system
based on your Python 3.x version and machine architecture.

It is advised that you create a conda environment in which you will install the needed libraries.
You should be able to create
a new environment as follows:

```bash
conda create --name 545 python=3.9 -y
```

Next, activate the `545` environment:

```bash
conda activate 545
```

Remember that you will need to activate the 545 environment before opening a jupyter notebook.


## Installing the Python libraries we will need

First, let's use conda to intstall jupyter and the libraries we will need in the environment we created:

```bash
conda install jupyter pytorch torchvision numpy pandas scikit-learn matplotlib
```


You are now ready to go!

To open a notebook run the following command in the directory where you have a notebook:

```bash
jupyter lab
```

Alternatively you can also specify the name of the notebook in the commandline, e.g.

```bash
jupyter lab notebook.ipynb
```

Whenever you open a new command line window,
you will need to execute `conda activate 545`
to activate the runtime environment.
