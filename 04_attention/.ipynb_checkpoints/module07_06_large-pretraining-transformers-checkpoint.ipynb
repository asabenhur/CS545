# Large-Scale Pretraining with Transformers

Over the last four notebooks we opened the transformer completely: we know what every weight matrix in our sentiment classifier does, and we verified our from-scratch implementation against the library's.  But that model, like every model we have trained in this course, was trained *from scratch* on labeled examples for one specific task.  The result is a narrow specialist: our classifier knows something about movie-review sentiment and nothing else, and its knowledge of English is limited to whatever 25,000 reviews could teach it.  Recall the embedding geometry we examined in the RNN module — rare words never learned useful representations because they were seen only a handful of times.

The transformer changed how this limitation is addressed, for a reason that is now easy to state with what we know.  Because it parallelizes completely and its representational capacity grows gracefully with depth and width, the transformer *scales*: performance improves predictably as models, data, and compute grow together (Kaplan et al., 2020).  This makes it worthwhile to **pretrain** one enormous model on essentially unlimited unlabeled text — with a self-supervised objective that manufactures labels from the text itself — and then adapt that one model to many downstream tasks.  This notebook surveys how that is done.  It is a reading notebook: no code, but everything described here is built from the components you have implemented.

Transformers are pretrained in three modes, differing only in which parts of the original encoder–decoder architecture they keep: **encoder-only**, **encoder–decoder**, and **decoder-only**.  We take them in turn.  Along the way we will need one mechanism this module has not used: *masked* (also called *causal*) attention, which is what turns a transformer from an encoder of text into a *generator* of text.  It is a one-line change to the attention operation from notebook 3, introduced below at the point where it becomes necessary.

## Encoder-Only Models

When only the transformer encoder is used, a sequence of input tokens is converted into an equal number of representation vectors, which can then be read out for prediction — precisely the architecture of our sentiment classifier.  The most influential encoder-only model is **BERT** (Bidirectional Encoder Representations from Transformers; Devlin et al., 2018), and you have already met two of its fingerprints in this module: the `<cls>` summary token we prepended to every review, and the readout of that position's output vector as the representation of the whole text.  Architecturally, BERT is our `TransformerClassifierScratch` scaled up: 12–24 encoder blocks instead of 2, hidden dimension 768–1024 instead of 100, learned positional embeddings instead of sinusoids, and a subword vocabulary instead of our word-level one.

### Pretraining BERT

![Left: Pretraining BERT with masked language modeling. Prediction of the masked "love" token depends on all input tokens before and after "love". Right: Attention pattern in the transformer encoder. Each token along the vertical axis attends to all input tokens along the horizontal axis.](https://github.com/d2l-ai/d2l-en/raw/refs/heads/master/img/bert-encoder-only.svg)

What can such a model be trained on without labels?  BERT's answer is **masked language modeling**: hide some tokens of the input and train the model to reconstruct them.  Given "I love this red car", replace a randomly chosen token — say "love" — with a special `<mask>` token, feed the corrupted sequence through the encoder, and train the model (via the usual cross-entropy loss) to predict "love" at the masked position.  Every stretch of text generates its own supervision, so books and all of Wikipedia become training data with no annotation effort.

Notice that nothing constrains the encoder's attention pattern: the prediction of "love" can draw on tokens both before *and after* the masked position — this is what "bidirectional" refers to, and it is exactly the unconstrained all-pairs attention we built.  Filling in a blank is easier when you can read the whole sentence, and the representations learned this way encode context from both directions.

### Fine-Tuning BERT

A pretrained BERT is then **fine-tuned** on each downstream task: attach a small task-specific layer (for sentiment analysis, a linear classifier on the `<cls>` output — literally our `self.decoder`), and continue training *all* parameters on the labeled task data.  The task layer starts from random initialization; the encoder starts from pretraining, and needs only to be nudged.  The division of labor solves the data problem our from-scratch model faced: general knowledge of language comes from billions of unlabeled tokens, and the 25,000 labeled reviews only need to teach the specifics of the task.  With this recipe the 350-million-parameter BERT, pretrained on 250 billion tokens, advanced the state of the art across single-text classification, text-pair classification, tagging, and question answering.

BERT set off a family of refinements — ALBERT (parameter sharing; Lan et al., 2019), SpanBERT (span prediction; Joshi et al., 2020), DistilBERT (distillation to a lightweight model; Sanh et al., 2019), ELECTRA (replaced-token detection; Clark et al., 2019) — and the encoder-only recipe crossed into vision: the vision transformer (ViT; Dosovitskiy et al., 2021) treats an image as a sequence of patch tokens with a `<cls>` token in front, and masked autoencoders (He et al., 2022) even reuse the reconstruction objective on image patches.

## Interlude: generating text, and masked attention

The encoder-only mode has a hard limitation: it always outputs exactly as many vectors as it received tokens.  It can classify, tag, and fill in blanks, but it cannot *generate* — it cannot produce an output sequence of arbitrary length, as translation or summarization requires.  Generation is done **autoregressively**: produce one token at a time, each conditioned on everything produced so far, exactly as our character-level RNNs generated text in the previous module.

Making a transformer autoregressive requires one modification, and now is the moment to introduce it.  At training time we would like to process a whole target sequence in parallel — that is the transformer's entire advantage — training the model to predict token $t+1$ at each position $t$.  But unconstrained self-attention lets position $t$ attend to positions after $t$, so the model could trivially "predict" the next token by looking at it.  **Masked (causal) attention** removes the temptation: before the softmax in the attention operation of notebook 3, add $-\infty$ (in practice, $-10^6$) to every score $s_{ij}$ with $j > i$.  After the softmax, those weights are exactly zero: each position attends only to itself and the past.  The attention weight maps we have been plotting all module become *lower-triangular*.  One line in `attention` — everything else in the architecture is unchanged.  (The name "causal" is standard but slightly misleading; it means "respects temporal order", and has little to do with the study of causality.)

At generation time, the model repeatedly predicts the next token from the sequence so far, appends it, and continues — sequential, like the RNN, which is why generating with transformers is slow even though training is parallel.

## Encoder–Decoder Models

The original transformer (Vaswani et al., 2017) was built for machine translation and combines both modes: an *encoder* reads the source sentence with unconstrained bidirectional attention, and a *decoder* generates the target sentence with causal self-attention.  The decoder needs one further ingredient: its blocks contain a third sublayer, **cross-attention**, which is ordinary attention where the queries come from the decoder's positions but the keys and values come from the *encoder's outputs* — each target token looks up relevant source tokens, in exactly the dictionary sense of notebook 3.

To pretrain encoder–decoder transformers beyond the limited supply of human-translated sentence pairs, BART (Lewis et al., 2019) and T5 (Raffel et al., 2020) both use reconstruction objectives on large text corpora: corrupt the input, train the model to restore it.

### Pretraining T5

T5 (Text-to-Text Transfer Transformer) unifies many tasks as one text-to-text problem: the encoder's input is a task description followed by the task input (e.g., "Summarize: " followed by an article), and the decoder generates the task output (the summary).

![Left: Pretraining T5 by predicting consecutive spans. The original sentence is "I", "love", "this", "red", "car", where "love" is replaced by a special "&lt;X&gt;" token, and consecutive "red", "car" are replaced by a special "&lt;Y&gt;" token. The target sequence ends with a special "&lt;Z&gt;" token. Right: Attention pattern in the transformer encoder–decoder. In the encoder self-attention (lower square), all input tokens attend to each other; in the encoder–decoder cross-attention (upper rectangle), each target token attends to all input tokens; in the decoder self-attention (upper triangle), each target token attends to present and past target tokens only (causal).](https://github.com/d2l-ai/d2l-en/raw/refs/heads/master/img/t5-encoder-decoder.svg)

For pretraining, T5 predicts *consecutive spans*: random spans of the text are each replaced by a special placeholder token, the corrupted text goes into the encoder, and the decoder is trained to produce the missing spans.  In the figure's example, the original "I love this red car" becomes the input "I &lt;X&gt; this &lt;Y&gt;" and the target "&lt;X&gt; love &lt;Y&gt; red car &lt;Z&gt;".  Note the attention pattern on the right of the figure — it summarizes everything in this notebook so far: bidirectional attention in the encoder (square block), cross-attention from every target token to all input tokens (rectangle), causal attention within the decoder (triangle).  T5 was pretrained this way on a trillion tokens from the C4 corpus of clean web text.

### Fine-Tuning T5

Like BERT, T5 is fine-tuned on task-specific data, but the text-to-text framing changes the mechanics: the task is specified in the input text itself, the output is generated as a sequence of arbitrary length, and no new layers are needed — the same language-in, language-out interface serves every task.

![Fine-tuning T5 for text summarization. Both the task description and article tokens are fed into the transformer encoder for predicting the summary.](https://github.com/d2l-ai/d2l-en/raw/refs/heads/master/img/t5-finetune-summarization.svg)

After fine-tuning, the 11-billion-parameter T5 achieved state-of-the-art results on both classification-style and generation benchmarks, and its encoder has had a productive afterlife of its own — the Imagen text-to-image model (Saharia et al., 2022) conditions image generation on text representations from a *frozen* T5 encoder, evidence that pretrained representations can be useful even with no fine-tuning at all.

![Text-to-image examples by the Imagen model, whose text encoder is from T5 (figures from Saharia et al., 2022).](https://github.com/d2l-ai/d2l-en/raw/refs/heads/master/img/imagen.png)

## Decoder-Only Models

The third mode removes the encoder entirely, along with the cross-attention sublayer that consumed its output.  What remains is a stack of blocks identical to the ones we built, except that the self-attention is causal: a pure autoregressive language model.  Decoder-only transformers are now the de facto architecture for large-scale language modeling — the simplest of the three recipes turned out to be the most scalable, feeding on the world's unlabeled text with the oldest self-supervised objective there is: predict the next token.

### GPT and GPT-2

GPT (Generative Pre-Training; Radford et al., 2018) chose the transformer decoder as its backbone with exactly this objective: the target sequence is the input sequence shifted by one token.

![Left: Pretraining GPT with language modeling. The target sequence is the input sequence shifted by one token. Both "&lt;bos&gt;" and "&lt;eos&gt;" are special tokens marking the beginning and end of sequences, respectively. Right: Attention pattern in the transformer decoder. Each token along the vertical axis attends to only its past tokens along the horizontal axis (causal).](https://github.com/d2l-ai/d2l-en/raw/refs/heads/master/img/gpt-decoder-only.svg)

The original GPT (about 100 million parameters) still needed per-task fine-tuning, like BERT.  Its successor GPT-2 (Radford et al., 2019) — 1.5 billion parameters, pretrained on 40 GB of text, with pre-normalization (the `norm_first` variant from this module's exercises) and improved initialization — showed something new: promising performance on multiple tasks *without any parameter updates*, purely by framing tasks as text continuation.

### GPT-3 and In-Context Learning

GPT-3 (Brown et al., 2020) scaled the same architecture roughly two orders of magnitude further (175 billion parameters, 300 billion training tokens) and turned that hint into a paradigm.  Because a language model generates text conditioned on a prefix, one can put a task description and a few input–output examples *in the prefix* and let the model complete the pattern:

![Zero-shot, one-shot, few-shot in-context learning with language models (transformer decoders). No parameter update is needed.](https://github.com/d2l-ai/d2l-en/raw/refs/heads/master/img/gpt-3-xshot.svg)

This is **in-context learning** — zero-shot, one-shot, or few-shot, depending on how many demonstrations the prefix contains.  No gradients, no parameter updates: the "learning" happens inside the forward pass, an ability nobody explicitly trained for.  Its successors, like GPT-4, extended the paradigm to multimodal input, accepting both text and images.

![Aggregate performance of GPT-3 for all 42 accuracy-denominated benchmarks (caption adapted and figure from Brown et al., 2020).](https://github.com/d2l-ai/d2l-en/raw/refs/heads/master/img/gpt3-xshot-scaling.png)

## Scalability

Why did scaling keep working?  Empirically, transformer language models improve according to remarkably regular **scaling laws** (Kaplan et al., 2020): the loss falls as a power law in each of model size, dataset size, and training compute, provided the other two are not the bottleneck.

![Transformer language model performance improves smoothly as we increase the model size, dataset size, and amount of compute used for training. For optimal performance all three factors must be scaled up in tandem. Empirical performance has a power-law relationship with each individual factor when not bottlenecked by the other two (caption adapted and figure from Kaplan et al., 2020).](https://github.com/d2l-ai/d2l-en/raw/refs/heads/master/img/scaling-power-law.png)

Larger models are also more *sample-efficient* — they reach a given loss having consumed fewer tokens than small models:

![Transformer language model training runs (figure from Kaplan et al., 2020).](https://github.com/d2l-ai/d2l-en/raw/refs/heads/master/img/scaling-sample-conv.png)

GPT-3 tested these predictions two orders of magnitude beyond the original measurements and stayed on the curve:

![GPT-3 performance (cross-entropy validation loss) follows a power-law trend with the amount of compute used for training (caption adapted and figure from Brown et al., 2020).](https://github.com/d2l-ai/d2l-en/raw/refs/heads/master/img/scaling-gpt3.png)

The regularity is what makes the enormous investments rational: one can forecast, from cheap small-scale runs, what a thousand-times-larger run will achieve.  Exactly *how* to split a fixed compute budget between model size and data remained contested; Chinchilla (Hoffmann et al., 2022) showed that models of that era were substantially undertrained relative to their size — a 70-billion-parameter model trained on 1.4 trillion tokens outperformed much larger models trained on fewer tokens — shifting the field's emphasis toward more data per parameter.

## Large Language Models

The scaling line from GPT continued through models such as Megatron-Turing NLG (530B parameters; Smith et al., 2022), Gopher (280B; Rae et al., 2021), Chinchilla, and PaLM (540B; Chowdhery et al., 2022), while open-weight releases — OPT (Zhang et al., 2022), BLOOM (Scao et al., 2022), Falcon (Penedo et al., 2023), and especially the Llama family (Touvron et al., 2023) — democratized research and deployment.  Llama's design point is worth noting in the light of the scaling laws: train a comparatively small model on far more tokens than compute-optimal, trading extra training cost for cheap inference.

A raw language model, however large, is just a next-token predictor: it continues text, which is not the same as being *useful*.  Two further training stages close the gap.  **Instruction tuning** (Wei et al., 2021; Sanh et al., 2021) fine-tunes the model on many tasks phrased as natural-language instructions, so that following instructions itself generalizes to unseen tasks.  **Reinforcement learning from human feedback** (RLHF; Ouyang et al., 2022) then optimizes the model against human preferences over candidate responses.  This recipe produced InstructGPT and, from it, ChatGPT — and the assistant-style models that followed from many labs (GPT-4, Claude, Gemini, Llama-family chat models); a variant using AI-generated feedback, constitutional AI (Bai et al., 2022), partially automates the human-labeling step.

Large models also exhibit capabilities their smaller versions lack (Wei et al., 2022, on "emergent abilities" — though how much is emergence and how much is measurement remains debated).  The most consequential for practice is that *prompting* becomes a programming interface.  **Chain-of-thought prompting** (Wei et al., 2022) — demonstrating intermediate reasoning steps in the prefix, or even just appending "Let's think step by step" (Kojima et al., 2022) — elicits multi-step mathematical, commonsense, and symbolic reasoning, and refinements such as sampling multiple reasoning paths and self-consistency (Wang et al., 2023) improve it further.  The field continues to move quickly; treat this section as a snapshot of a moving frontier rather than its current edge.

## Summary and Discussion

Transformers are pretrained as encoder-only (BERT: bidirectional attention, masked-token reconstruction, fine-tuned per task), encoder–decoder (T5: full attention in the encoder, causal in the decoder, cross-attention between; span reconstruction; text-to-text interface), and decoder-only (GPT: causal attention, next-token prediction; adapted by fine-tuning or by in-context learning with no parameter update at all).  Performance scales as a power law with model size, data, and compute, which justifies the pretraining paradigm economically; instruction tuning and RLHF convert scaled language models into usable assistants.  Every architecture in this story is assembled from the components you implemented in notebooks 2–4, plus exactly one new mechanism — the causal mask, a single line added to `attention`.

The ideas also travel across modalities: vision transformers apply the encoder recipe to image patches; CLIP (Radford et al., 2021) aligns transformer text and image encoders; systems like Imagen and DALL-E 2 (Ramesh et al., 2022) build image generation on pretrained text representations; and text-to-image transformers such as Parti (Yu et al., 2022) show the same scaling behavior as their language counterparts.

![Image examples generated from the same text by the Parti model of increasing sizes (350M, 750M, 3B, 20B) (examples from Yu et al., 2022).](https://github.com/d2l-ai/d2l-en/raw/refs/heads/master/img/parti.png)
