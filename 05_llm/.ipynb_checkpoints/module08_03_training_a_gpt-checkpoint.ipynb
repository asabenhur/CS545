# Training a GPT from Scratch

We have measured other people's language models; now we build one.  Everything needed is already in hand: the decoder-only architecture from notebook 1, perplexity from notebook 2, and our transformer layers.  This notebook assembles those parts into a **Generative Pretrained Transformer (GPT)**, trains it on real text, and watches English emerge as training progresses.

Two decisions shape everything that follows.

**The corpus.**  Train a small model on a novel and you get what the previous notebook's trigram gave us: locally plausible phrases that fall apart across a sentence.  The problem is not really the model, it is the mismatch — ordinary English text presumes a vast amount of world knowledge and vocabulary that a few-million-parameter model trained for twenty minutes cannot acquire.  **TinyStories** (Eldan & Li, 2023) was built to remove exactly that mismatch: a synthetic corpus of short children's stories, generated by GPT-3.5/4 under the constraint that they use only the vocabulary a three-to-four-year-old understands.  Its authors showed that models *far* below the usual size threshold — millions, not billions of parameters — produce fluent, coherent, grammatical stories when trained on it.  That makes it the rare corpus where a laptop-scale training run yields a model whose output is genuinely readable rather than merely suggestive.

**The scale.**  Our default model is about 4M parameters trained on a few million tokens — minutes on Apple silicon, well under an hour on a CPU.  That is roughly 1/40,000th of GPT-3, and the point is not to compete.  The point is that the *recipe* is identical: same architecture, same objective, same optimizer, same loss curve shape.  What separates our model from a frontier one is data, parameters, and compute — the three axes of notebook 1's scaling laws — not ideas.

The plan: prepare TinyStories, train a BPE tokenizer on it (the real version of notebook 1's toy), build the model, verify it before training it, train while tracking validation perplexity, and generate stories from checkpoints along the way.

## The data

TinyStories lives on the Hugging Face Hub.  It's a large multi-gigabyte dataset.  So rather than downloading the entire thing, we stream it, pulling stories one at a time until we have as much text as we intend to train on, caching the result to a local file.

```python
N_STORIES = 40_000        # ~4M words; raise for a better model, lower to go faster

def load_tinystories(n_stories=N_STORIES, path='./data/tinystories.txt'):
    """Stream `n_stories` stories from the Hub; cache them to a local file."""
    if not os.path.exists(path):
        from datasets import load_dataset
        os.makedirs(os.path.dirname(path), exist_ok=True)
        print(f'streaming {n_stories} stories from the Hub ...')
        stream = load_dataset('roneneldan/TinyStories', split='train',
                              streaming=True)
        with open(path, 'w') as f:
            for i, row in enumerate(stream):
                if i >= n_stories:
                    break
                # <|endoftext|> marks a document boundary, as in GPT-2 training
                f.write(row['text'].strip() + '\n<|endoftext|>\n')
    with open(path) as f:
        return f.read()

text = load_tinystories()
stories = text.split('<|endoftext|>')
print(f'{len(text):,} characters, {len(text.split()):,} words, '
      f'{len(stories) - 1:,} stories\n')
print(stories[0].strip()[:600])
```

Simple vocabulary, short sentences, a beginning and an end — and, importantly, *narrative structure*: a character, a problem, a resolution.  That structure is what will make success visible.  A model that has learned only local grammar will produce sentences that go nowhere; a model that has learned more will keep a character's name consistent and finish the story it started.

## The tokenizer

In notebook 1 we implemented byte-pair encoding on twelve words to see the algorithm; now we run the real thing on our actual corpus, using Hugging Face's `tokenizers` library.

Why train our own rather than borrow GPT-2's?  Because vocabulary size is a design parameter, and GPT-2's 50,257 tokens would be absurd here: the embedding table scales with $|V|$, so GPT-2's vocabulary would cost $50{,}257 \times 256 \approx 13$M parameters which is going to be totally unnecessary to represent words TinyStories never uses.  A 4,096-token vocabulary trained on this corpus covers its simple language efficiently and keeps the parameters where they do the work: in the transformer blocks.

Note the tradeoff, though: a smaller vocabulary means *more tokens per sentence*, so each story consumes more context and more compute.  Vocabulary size trades embedding parameters against sequence length, and 4,096 is a reasonable compromise for a corpus this simple.

```python
from tokenizers import Tokenizer, models, trainers, pre_tokenizers, decoders

VOCAB_SIZE = 4096
TOK_PATH = './data/tinystories-bpe.json'

def train_tokenizer(text, vocab_size=VOCAB_SIZE, path=TOK_PATH):
    if os.path.exists(path):
        return Tokenizer.from_file(path)
    tok = Tokenizer(models.BPE())
    # byte-level: any UTF-8 string is representable, nothing is ever <unk>
    tok.pre_tokenizer = pre_tokenizers.ByteLevel(add_prefix_space=False)
    tok.decoder = decoders.ByteLevel()
    trainer = trainers.BpeTrainer(vocab_size=vocab_size,
                                  special_tokens=['<|endoftext|>'],
                                  initial_alphabet=pre_tokenizers.ByteLevel.alphabet(),
                                  show_progress=False)
    tok.train_from_iterator([text], trainer)     # the merge loop from notebook 1
    tok.save(path)
    return tok

tokenizer = train_tokenizer(text)
EOT = tokenizer.token_to_id('<|endoftext|>')
print(f'vocabulary: {tokenizer.get_vocab_size()} tokens; <|endoftext|> id = {EOT}\n')

sample = 'Once upon a time, a curious little rabbit found a shiny key.'
ids = tokenizer.encode(sample).ids
print('tokens :', tokenizer.encode(sample).tokens)
print(f'{len(sample.split())} words -> {len(ids)} tokens; round-trip ok:',
      tokenizer.decode(ids) == sample)
```

Look at which words survived as single tokens and which fragmented (`Ġ` marks a leading space, as in notebook 1).  The staples of this corpus — *once*, *upon*, *time*, *little* — are whole, while rarer words break into pieces; with only 4,096 slots, BPE spends them on what the corpus actually repeats.  Compare the token count to the word count: roughly 1.3–1.5 tokens per word, versus GPT-2's ~1.3 on ordinary English.  A specialized vocabulary a twelfth the size costs us almost nothing *on this corpus* — and would do badly on anything else, which is the tradeoff in a sentence.

Now encode the whole corpus once, into one long stream of token ids, and split off the last 5% for validation.  Note *what* we are splitting: a language model's training example is simply a contiguous window of the stream, so the "dataset" is a single array and a batch is a handful of random windows into it.

```python
ids = np.array(tokenizer.encode(text).ids, dtype=np.uint16)   # 4096 < 65536
n_val = int(0.05 * len(ids))
train_ids = torch.from_numpy(ids[:-n_val].astype(np.int64))
val_ids = torch.from_numpy(ids[-n_val:].astype(np.int64))
print(f'{len(ids):,} tokens total  ({len(train_ids):,} train / {len(val_ids):,} val)')
print(f'{len(ids) / len(text.split()):.2f} tokens per word')

CONTEXT = 256      # how many tokens the model can see at once

def get_batch(data, batch_size, context=CONTEXT, generator=None):
    """A batch of random windows: inputs x and next-token targets y."""
    i = torch.randint(len(data) - context - 1, (batch_size,), generator=generator)
    x = torch.stack([data[j:j + context] for j in i])
    y = torch.stack([data[j + 1:j + context + 1] for j in i])   # shifted by one
    return x.to(device), y.to(device)

xb, yb = get_batch(train_ids, 4)
print('\nx:', xb.shape, ' y:', yb.shape)
print('x[0][:8] =', xb[0][:8].tolist())
print('y[0][:8] =', yb[0][:8].tolist(), ' <- x shifted left by one')
```

There is the shifted-target construction from notebook 1, in code: `y` is `x` moved one position left, so the model's output at every position has a label sitting at the same index of `y`.  One batch of 4 windows yields $4 \times 256 = 1024$ next-token predictions, every one of them contributing to the loss.

## The model

Now the architecture.  We start from module 7's components and make **three** changes, each worth understanding.

**Change 1: the causal mask.**  Our attention function let every query attend to every key.  A language model must not: position $t$ predicting $x_{t+1}$ cannot be allowed to look at $x_{t+1}$, which sits right there in the input.  The fix is to set the pre-softmax scores for all future positions to $-\infty$, so softmax assigns them exactly zero weight.  This is the lower-triangular pattern from notebook 1's figure, now as three lines of code.

**Change 2: pre-norm instead of post-norm.**  Module 7's `AddNorm` applied layer normalization *after* the residual addition — `LayerNorm(x + Sublayer(x))` — following the original 2017 paper.  GPT-2 and everything since instead normalize the *input* to each sublayer and leave the residual path untouched: `x + Sublayer(LayerNorm(x))`.  The difference matters for deep stacks: with pre-norm there is a clean, unnormalized path from input to output (a "residual highway"), which makes gradients well-behaved and lets you train without the learning-rate warmup that post-norm transformers require to avoid diverging.  We adopt it — one small reordering, a large gain in stability.

**Change 3: weight tying.**  The input embedding maps token id → vector; the output layer maps vector → scores over tokens.  These are inverse operations over the same vocabulary, and *sharing the same matrix* for both (Press & Wolf, 2017) saves $|V| \times d$ parameters — here about 1M, a fifth of what the model would otherwise be — and typically improves perplexity.  GPT-2 does this too.

Everything else — multi-head attention, sinusoidal positions, the FFN, residuals — is the code from module 7, unchanged.

```python
class CausalSelfAttention(nn.Module):
    """Module 7's MultiHeadAttention, plus a causal mask.

    Only `forward` differs from the version we built for the classifier: the
    scores for future positions are set to -inf before the softmax."""
    def __init__(self, num_hiddens, num_heads, dropout=0.0, bias=True):
        super().__init__()
        assert num_hiddens % num_heads == 0
        self.num_heads = num_heads
        self.W_q = nn.Linear(num_hiddens, num_hiddens, bias=bias)
        self.W_k = nn.Linear(num_hiddens, num_hiddens, bias=bias)
        self.W_v = nn.Linear(num_hiddens, num_hiddens, bias=bias)
        self.W_o = nn.Linear(num_hiddens, num_hiddens, bias=bias)
        self.attn_dropout = nn.Dropout(dropout)
        self.resid_dropout = nn.Dropout(dropout)

    def split_heads(self, X):
        b, seq, _ = X.shape
        return X.reshape(b, seq, self.num_heads, -1).transpose(1, 2)

    def merge_heads(self, X):
        b, h, seq, d_head = X.shape
        return X.transpose(1, 2).reshape(b, seq, h * d_head)

    def forward(self, X):
        Q = self.split_heads(self.W_q(X))          # (batch, heads, seq, d_head)
        K = self.split_heads(self.W_k(X))
        V = self.split_heads(self.W_v(X))

        d = Q.shape[-1]
        scores = Q @ K.transpose(-2, -1) / math.sqrt(d)   # (b, h, seq, seq)
        # --- the causal mask: forbid attending to later positions ---
        seq = X.shape[1]
        future = torch.triu(torch.ones(seq, seq, dtype=torch.bool,
                                       device=X.device), diagonal=1)
        scores = scores.masked_fill(future, float('-inf'))
        # ------------------------------------------------------------
        self.attention_weights = F.softmax(scores, dim=-1)
        out = self.attn_dropout(self.attention_weights) @ V
        return self.resid_dropout(self.W_o(self.merge_heads(out)))
```

```python
# Does the mask actually work?  Two checks before we build anything on it.
torch.manual_seed(0)
att = CausalSelfAttention(32, 4).eval()
X = torch.randn(1, 6, 32)

W = att(X) is not None and att.attention_weights[0, 0]
print('attention weights, head 0 (rows = queries, cols = keys):')
print(W.detach().round(decimals=2).numpy())
assert torch.allclose(W.triu(1), torch.zeros_like(W)), 'attends to the future!'
assert torch.allclose(W.sum(-1), torch.ones(6)), 'rows must be distributions'

# The decisive test: changing a LATER token must not change an EARLIER output.
X2 = X.clone()
X2[0, 4] = torch.randn(32)                    # perturb position 4
out1, out2 = att(X), att(X2)
print(f'\nmax change at positions 0-3: {(out1[0, :4] - out2[0, :4]).abs().max():.2e}')
print(f'max change at positions 4-5: {(out1[0, 4:] - out2[0, 4:]).abs().max():.2e}')
assert torch.allclose(out1[0, :4], out2[0, :4], atol=1e-6)
print('\ncausal masking verified: no information flows backwards in time')
```

That second test is the one that matters, and it is the test to remember whenever you write an autoregressive model: perturb a token and confirm that *earlier* outputs are bit-for-bit unchanged.  A subtly broken mask leaks the answer into the question — training loss plummets, generation is gibberish, and the bug is invisible in the loss curve.  (If the assertion above ever fires while you are modifying this code, stop and fix it before training.)

Now the block and the full model.

```python
class PositionWiseFFN(nn.Module):
    """Module 7's FFN, with GELU in place of ReLU (as in GPT-2)."""
    def __init__(self, num_hiddens, ffn_num_hiddens, dropout=0.0):
        super().__init__()
        self.dense1 = nn.Linear(num_hiddens, ffn_num_hiddens)
        self.dense2 = nn.Linear(ffn_num_hiddens, num_hiddens)
        self.dropout = nn.Dropout(dropout)

    def forward(self, X):
        return self.dropout(self.dense2(F.gelu(self.dense1(X))))


class GPTBlock(nn.Module):
    """A decoder block: causal self-attention + FFN, both pre-normalized.

    Compare module 7's encoder block, which was post-norm:
        Y = LayerNorm(X + Attention(X));  return LayerNorm(Y + FFN(Y))"""
    def __init__(self, num_hiddens, ffn_num_hiddens, num_heads, dropout):
        super().__init__()
        self.ln1 = nn.LayerNorm(num_hiddens)
        self.attention = CausalSelfAttention(num_hiddens, num_heads, dropout)
        self.ln2 = nn.LayerNorm(num_hiddens)
        self.ffn = PositionWiseFFN(num_hiddens, ffn_num_hiddens, dropout)

    def forward(self, X):
        X = X + self.attention(self.ln1(X))    # residual path stays un-normalized
        return X + self.ffn(self.ln2(X))


class PositionalEncoding(nn.Module):
    """Module 7's sinusoidal positional encoding, unchanged."""
    def __init__(self, d_model, max_len=5000):
        super().__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float()
                             * (-math.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        self.register_buffer('pe', pe.unsqueeze(0))

    def forward(self, X):
        return X + self.pe[:, :X.size(1), :]
```

```python
class GPT(nn.Module):
    """A decoder-only transformer language model."""
    def __init__(self, vocab_size, num_hiddens=256, ffn_num_hiddens=1024,
                 num_heads=8, num_blks=4, dropout=0.1, max_len=CONTEXT):
        super().__init__()
        self.num_hiddens = num_hiddens
        self.embedding = nn.Embedding(vocab_size, num_hiddens)
        self.pos_encoding = PositionalEncoding(num_hiddens, max_len)
        self.dropout = nn.Dropout(dropout)
        self.blks = nn.ModuleList([
            GPTBlock(num_hiddens, ffn_num_hiddens, num_heads, dropout)
            for _ in range(num_blks)])
        self.ln_f = nn.LayerNorm(num_hiddens)          # final norm (pre-norm nets need it)
        self.lm_head = nn.Linear(num_hiddens, vocab_size, bias=False)
        self.lm_head.weight = self.embedding.weight    # weight tying
        self.apply(self._init_weights)

    def _init_weights(self, m):
        if isinstance(m, nn.Linear):
            nn.init.normal_(m.weight, std=0.02)        # GPT-2's initialization
            if m.bias is not None:
                nn.init.zeros_(m.bias)
        elif isinstance(m, nn.Embedding):
            nn.init.normal_(m.weight, std=0.02)

    def forward(self, X, targets=None):
        H = self.pos_encoding(self.embedding(X) * math.sqrt(self.num_hiddens))
        H = self.dropout(H)
        for blk in self.blks:
            H = blk(H)
        logits = self.lm_head(self.ln_f(H))            # (batch, seq, vocab)
        if targets is None:
            return logits, None
        loss = F.cross_entropy(logits.reshape(-1, logits.size(-1)),
                               targets.reshape(-1))
        return logits, loss

model = GPT(VOCAB_SIZE).to(device)
n_params = sum(p.numel() for p in model.parameters())
n_embed = model.embedding.weight.numel()
print(f'{n_params/1e6:.2f}M parameters '
      f'({n_embed/1e6:.2f}M in the (tied) embedding, '
      f'{(n_params-n_embed)/1e6:.2f}M in the blocks)')
```

### Sanity checks before training

Training a broken model wastes an hour and teaches nothing, so we spend a minute on two checks that catch most bugs (the causal-mask test above was the third).

The first is the most useful diagnostic in language modeling: **an untrained model must have perplexity $\approx |V|$**.  Random weights mean a near-uniform softmax, and notebook 2 fixed that anchor — uniform guessing over $|V|$ options costs exactly $\ln |V|$ nats.  If your initial loss is far from $\ln|V|$, something is wrong before you start (a common culprit: initialization at the wrong scale).

```python
model.eval()
with torch.no_grad():
    xb, yb = get_batch(val_ids, 8)
    logits, loss = model(xb, yb)

print(f'output shape {tuple(logits.shape)} = (batch, context, vocab)')
print(f'initial loss       {loss.item():.3f} nats')
print(f'ln(vocab_size)     {math.log(VOCAB_SIZE):.3f} nats   <- expected at random init')
print(f'initial perplexity {math.exp(loss.item()):.0f}  (vocabulary = {VOCAB_SIZE})')

# 2. Can it overfit a single batch?  If not, the training loop is broken.
#    (A short context and a high learning rate keep this to well under a minute.)
probe = GPT(VOCAB_SIZE).to(device)
opt = torch.optim.AdamW(probe.parameters(), lr=1e-3)
xb, yb = get_batch(train_ids, 4, context=64)
probe.train()
for step in range(300):
    _, l = probe(xb, yb)
    opt.zero_grad(); l.backward(); opt.step()
    if step % 100 == 0:
        print(f'  overfit probe, step {step:3d}: loss {l.item():.3f}')
print(f'after 300 steps on ONE batch: loss {l.item():.3f} '
      f'(should approach 0 -- the model can memorize)')
del probe, opt
```

Both checks pass: the model starts as ignorant as theory says it must, and it *can* learn when the gradient path is intact.  The single-batch overfit test is worth internalizing — a model that cannot memorize four sequences will never learn four million, and the failure is almost always a wiring bug rather than a hyperparameter.

## Training

The loop holds no surprises beyond what module 7 already used, plus two standard refinements:

* **AdamW with weight decay**, applied only to matrices — biases and LayerNorm gains are excluded by convention, since shrinking them toward zero has no regularizing benefit.
* **A cosine learning-rate schedule with a short warmup**: ramp up over the first few hundred steps, then decay smoothly toward zero.  (You saw warmup + linear decay in the BERT fine-tuning notebook; cosine decay is the standard for pretraining.)  Pre-norm makes warmup optional rather than essential, but it still helps.

We also evaluate periodically on held-out data — reporting **perplexity**, our instrument from notebook 2 — and sample a short story at each evaluation, so the model's progress is legible as text and not just as a number.

```python
@torch.no_grad()
def estimate_loss(model, data, n_batches=20, batch_size=32):
    model.eval()
    g = torch.Generator().manual_seed(1234)          # same windows every time
    total = 0.0
    for _ in range(n_batches):
        xb, yb = get_batch(data, batch_size, generator=g)
        total += model(xb, yb)[1].item()
    model.train()
    return total / n_batches


@torch.no_grad()
def generate(model, prompt='Once upon a time', max_new_tokens=100,
             temperature=0.8, top_k=None, seed=None):
    """Autoregressive sampling -- notebook 1's loop, now with temperature."""
    model.eval()
    # Seeding makes samples reproducible.  We save and restore the global RNG
    # state around it: otherwise calling this from inside the training loop
    # would reset the RNG and make every batch after an eval identical.
    rng_state = torch.get_rng_state()
    if seed is not None:
        torch.manual_seed(seed)          # works on cpu, cuda and mps alike
    ids = torch.tensor([tokenizer.encode(prompt).ids], device=device)
    for _ in range(max_new_tokens):
        logits, _ = model(ids[:, -CONTEXT:])          # never exceed the context
        logits = logits[:, -1, :] / temperature       # <1 sharpens, >1 flattens
        if top_k is not None:
            kth = logits.topk(top_k, dim=-1).values[:, [-1]]
            logits = logits.masked_fill(logits < kth, float('-inf'))
        nxt = torch.multinomial(logits.softmax(-1), 1)
        if nxt.item() == EOT:
            break
        ids = torch.cat([ids, nxt], dim=1)
    model.train()
    torch.set_rng_state(rng_state)
    return tokenizer.decode(ids[0].tolist())
```

```python
MAX_STEPS = 3000          # ~10 min on Apple MPS, ~40 min on CPU; 6000+ is better
BATCH_SIZE = 32
EVAL_EVERY = 250
WARMUP, LR, MIN_LR = 200, 3e-4, 3e-5

def lr_at(step):
    if step < WARMUP:                                   # linear warmup
        return LR * (step + 1) / WARMUP
    progress = (step - WARMUP) / max(1, MAX_STEPS - WARMUP)
    return MIN_LR + 0.5 * (LR - MIN_LR) * (1 + math.cos(math.pi * progress))

decay = [p for p in model.parameters() if p.dim() >= 2]   # matrices
no_decay = [p for p in model.parameters() if p.dim() < 2]  # biases, LayerNorm
opt = torch.optim.AdamW([{'params': decay, 'weight_decay': 0.1},
                         {'params': no_decay, 'weight_decay': 0.0}],
                        lr=LR, betas=(0.9, 0.95))

history = []
t0 = time.time()
model.train()
for step in range(MAX_STEPS + 1):
    for group in opt.param_groups:
        group['lr'] = lr_at(step)

    xb, yb = get_batch(train_ids, BATCH_SIZE)
    _, loss = model(xb, yb)
    opt.zero_grad(set_to_none=True)
    loss.backward()
    torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)   # guards against spikes
    opt.step()

    if step % EVAL_EVERY == 0:
        tr, va = estimate_loss(model, train_ids), estimate_loss(model, val_ids)
        history.append((step, math.exp(tr), math.exp(va)))
        print(f'step {step:5d} | train PPL {math.exp(tr):7.1f} | '
              f'val PPL {math.exp(va):7.1f} | {time.time() - t0:5.0f}s')
        print(f'   {generate(model, max_new_tokens=40, seed=0)!r}\n')

tokens_seen = MAX_STEPS * BATCH_SIZE * CONTEXT
print(f'done in {(time.time() - t0)/60:.1f} min; '
      f'{tokens_seen/1e6:.1f}M tokens seen '
      f'({tokens_seen/len(train_ids):.1f} epochs over the corpus)')
```

The printed samples are the real result of this notebook.  Read them in order and you can watch the model acquire language in stages: first token frequencies (a fog of *the*, *and*, *she*), then word-level grammar, then sentence structure and punctuation, then — remarkably, for a model this small — characters that persist across sentences and stories that actually resolve.  Nothing supervised this progression.  It is all "predict the next token", applied to enough text.

Note also *where the perplexity lands*.  Starting from $|V| = 4096$ (pure ignorance), a few thousand steps take it into the low tens — meaning the model has narrowed a four-thousand-way choice down to an effective handful of options at each step.  That is the same instrument we pointed at GPT-2 in notebook 2, now measuring something we built.

```python
steps, tr, va = zip(*history)
fig, ax = plt.subplots(figsize=(6.4, 4))
ax.plot(steps, tr, 'o-', color=AMBER, label='train', markersize=3)
ax.plot(steps, va, 's-', color='#3d6ea8', label='validation', markersize=3)
ax.axhline(VOCAB_SIZE, ls=':', color='0.6')
ax.text(steps[-1], VOCAB_SIZE * 0.95, f'random guessing (|V| = {VOCAB_SIZE})',
        ha='right', va='top', fontsize=7.5, color='0.4')
ax.set_yscale('log')
ax.set_xlabel('training step'); ax.set_ylabel('perplexity (log scale)')
ax.set_title('Learning to predict the next token', fontsize=10)
ax.legend(frameon=False); ax.spines[['top', 'right']].set_visible(False)
plt.tight_layout();
```

Train and validation curves that stay close together tell us this run is **not** overfitting — unsurprising, since with a few epochs over several million tokens the model sees most text only a handful of times.  This is the normal regime for language-model pretraining, and it is the opposite of the trigram's predicament in notebook 2: that model memorized because it had one parameter per observed context, while ours is forced to *generalize* because 4M parameters cannot store millions of tokens' worth of detail.  If you scale `MAX_STEPS` up far enough on a fixed corpus, the two curves will eventually separate — the classic sign that you need more data rather than more steps.

## Talking to our model

With training done, we can explore what the model learned.  The `temperature` knob is worth playing with: it divides the logits before the softmax, so low values sharpen the distribution toward the argmax (conservative, repetitive) and high values flatten it (creative, incoherent).  Notebook 4 examines this and its relatives, top-$k$ and top-$p$, in detail.

```python
for temp in [0.2, 0.8, 1.4]:
    print(f'--- temperature {temp} ---')
    print(generate(model, 'Once upon a time', max_new_tokens=120,
                   temperature=temp, seed=0), '\n')

print('--- prompted continuations (temperature 0.8) ---')
for prompt in ['Lily found a big red ball. She',
               'The dragon was very sad because',
               'Tom and his mother went to the park to']:
    print(f'{generate(model, prompt, max_new_tokens=80, seed=1)}\n')
```

Low temperature gives safe, repetitive prose; high temperature loses the thread; around 0.8 tends to read best — the same tradeoff practitioners tune for every deployed model.

Be precise about what we have and have not built.  This model writes fluent, mostly-grammatical children's stories, and that is a real achievement for 6M parameters and a few minutes of laptop compute — it is the TinyStories result reproduced.  But it knows nothing outside this corpus: it cannot answer questions, follow instructions, or discuss anything but simple stories, because its entire experience of language is 40,000 synthetic children's tales.  Capability follows data and scale, exactly as notebook 1's scaling laws describe.

Two experiments make that concrete, and both are worth running: train the same architecture on *The Time Machine* instead (the d2l original — you will get the trigram's problem back, since a novel's vocabulary and world knowledge exceed what this budget can absorb), and re-read the scaling-law section of notebook 1 with a loss curve of your own now in hand.

## What we would change to scale up

Our model is a faithful miniature, but the gap to a modern LLM is not only size.  If we were to grow this by orders of magnitude, these are the pieces that would change — each a refinement of something we built, none a new idea:

* **Positional encoding.**  Sinusoids are fine here; modern models use **RoPE** (rotary embeddings), which encodes *relative* position by rotating Q and K, and extrapolates better to longer contexts than the model was trained on.
* **Normalization and activation.**  **RMSNorm** (LayerNorm without the mean subtraction) is cheaper and works as well; **SwiGLU** replaces our GELU FFN with a gated variant that is consistently a little better per parameter.
* **Attention efficiency.**  Our explicit $T \times T$ score matrix costs $O(T^2)$ memory; **FlashAttention** computes the same result in tiles without ever materializing it, and **grouped-query attention** shares K/V across heads to shrink the KV cache during generation.  (PyTorch's `F.scaled_dot_product_attention` already dispatches to a fused implementation — a one-line substitution for our masked-softmax block, and a worthwhile exercise.)
* **Precision and parallelism.**  bfloat16 mixed precision, gradient accumulation for large effective batches, and — past a single machine — data and tensor parallelism across many GPUs.
* **Data, above all.**  Trillions of tokens of filtered, deduplicated, diverse text.  This is where most of the real work in frontier pretraining goes, and it is the axis that separates a story-teller from an assistant.

We have now been through the entire pretraining half of the pipeline: architecture, objective, tokenizer, training loop, evaluation.  In the next notebook we pick up models where trillions of tokens have already been spent — and find out what that buys.

## Summary

* **TinyStories** makes small-scale language modeling instructive: a corpus simple enough that a ~6M-parameter model trained for minutes produces coherent, grammatical prose.
* Our **GPT** is module 7's transformer with three changes: a **causal mask** (three lines), **pre-norm** blocks (better gradient flow in deep stacks), and **weight tying** between the embedding and the output layer.  Training data is one long token stream; a batch is a set of random windows with targets shifted by one.
* **Verify before training**: initial loss must be $\approx \ln |V|$; the model must be able to overfit a single batch; and perturbing a token must leave *earlier* outputs unchanged — the definitive test that the causal mask is intact.
* The training recipe is standard and small: AdamW (decay on matrices only), warmup + cosine decay, gradient clipping, periodic validation **perplexity** — which falls from $|V|$ to the low tens while the samples progress from noise to stories.
* Scaling further is a matter of engineering refinements (RoPE, RMSNorm, SwiGLU, FlashAttention, mixed precision) and, above all, **far more data** — not of new architectural ideas.
