# Accessing data files on google colab

This notebook is for those who have chosen to use google colab for this course.
The files on google colab do not persist across sessions.  Therefore, you will often find it useful to link colab to your google drive.  Let's explore the commands necessary for that.


```python
from google.colab import drive
drive.mount('/content/gdrive')
```

The first time you do this you will be prompted to authorize colab to access your google drive.  Once this step is done, you will be able to access your files on your google drive from colab.  For example, assume I have a directory called `CS545` with the file `biodeg.csv` in it.  

```python
!ls gdrive/MyDrive/CS545
```

With this module we can write code that accesses files on colab or on your local machine in a consistent way:

```python
use_colab = True
google_drive_path = 'CS545'

file_name = 'biodeg.csv'

if use_colab :
  from google.colab import drive
  drive.mount('/content/gdrive')
  import os
  os.chdir(os.path.join('gdrive/MyDrive', google_drive_path))
  # as an alternative to changing directory, you can copy the file from
  # your google drive to colab
# now we can access the file in the same way regardless of whether we are using
# colab or Jupyter on our local machine:
file_handle = open(file_name)
file_handle.readline()
```
