# Attention:  building meaning from context

The previous notebook established three things about the transformer: it works (matching the BiLSTM on sentiment), it is blind to word order, and each of its layers computes a grid of *attention weights* we could plot but did not yet explain.  The notebooks that follow will open the box and derive the attention operation precisely, in code and in math.

This one has no equations and trains nothing.  Its job is to answer a prior question — *what is attention for?* — with a picture, so that when the formula arrives in the next notebook, you already know what it is trying to accomplish.  The one-sentence version:

> The embedding layer gives each word a *fixed* vector, the same in every sentence.  Attention updates those vectors, layer by layer, so that each word's representation comes to reflect the *context* it appears in.

The rest of the notebook unpacks that sentence with two small animations.  Everything drawn here is a deliberate cartoon — a hand-built 2D "meaning space," not a trained model, chosen so the geometry is legible.  The real thing lives in hundreds of dimensions and is learned from data.

```python
import numpy as np
import torch
from torch import nn
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from matplotlib.patches import FancyArrowPatch
from IPython.display import HTML

# render animations as inline HTML players
plt.rcParams['animation.html'] = 'jshtml'
plt.rcParams['animation.embed_limit'] = 40  # MB
```

## Input embeddings:  One word, one vector

Our input tokens are converted to vectors using an embedding layer.  Those vectors are fixed, regardless of context.  In the following, we give three "sentences" that share the word *mole* used three different ways:

```python
torch.manual_seed(0)
vocab = {w: i for i, w in enumerate(
    ['<unk>', 'mole', 'avogadro', 'grams', 'burrowed', 'garden',
     'spy', 'agency', 'the', 'a', 'of', 'number'])}
embedding = nn.Embedding(len(vocab), 8)   # 8-dimensional toy embeddings

sentences = [
    ['a', 'mole', 'of', 'avogadro', 'number'],   # chemistry
    ['the', 'mole', 'burrowed', 'garden'],        # animal
    ['the', 'mole', 'spy', 'agency'],             # espionage
]

for sent in sentences:
    ids = torch.tensor([vocab[w] for w in sent])
    vecs = embedding(ids)
    mole_vec = vecs[sent.index('mole')]
    print(f'{" ".join(sent):32s}  mole -> {mole_vec[:4].detach().numpy().round(3)} ...')
```

The vector for *mole* is identical in all three sentences.  Yet the three moles could hardly be more different: a unit of $6\times10^{23}$ particles, a small burrowing mammal, an infiltrating agent.  A single fixed vector cannot be a good representation of all three at once.

This is the deficiency attention exists to fix.  We want the vector *at the mole position* to end up somewhere different depending on the company it keeps — pulled toward "chemistry" among *avogadro* and *grams*, toward "animal" among *burrowed* and *garden*.  That is what the next picture shows.

## The idea: context refines meaning

The following figure demonstrates the role of attention - to move the embedding of a vector to a region in the embedding space that better represents its meaning in the sentence. Each attention layer nudges the mole vector toward the region its neighbors imply, and stacking layers refines the estimate.

```python
# --- a hand-built 2D "meaning space" (illustration, not a trained model) ---
regions = {'CHEMISTRY': (np.array([3.2, 2.3]), 'C0'),
           'ANIMAL':    (np.array([3.0, -2.4]), 'C2'),
           'ESPIONAGE': (np.array([-3.2, 1.6]), 'C3')}
context_pos = {'avogadro': np.array([2.4, 2.9]), 'grams': np.array([3.9, 1.7]),
               'burrowed': np.array([2.3, -2.9]), 'garden': np.array([3.8, -1.8])}
demo = [("one mole contains avogadro's number of grams", ['avogadro', 'grams']),
        ("the mole burrowed under the garden",           ['burrowed', 'garden'])]
mole_start = np.array([0.2, 0.15])
n_layers, hold, fps = 3, 9, 12

def trajectory(ctx):
    """mole position per layer: move a fraction toward the mean of context."""
    pts, p = [mole_start.copy()], mole_start.copy()
    for _ in range(n_layers):
        p = p + 0.55 * (np.mean([context_pos[w] for w in ctx], axis=0) - p)
        pts.append(p.copy())
    return pts

trajs = [trajectory(ctx) for _, ctx in demo]
per = 6 + n_layers * hold + 12
fig, ax = plt.subplots(figsize=(6.5, 5))

def frame(f):
    ax.clear(); ax.set_xlim(-5, 5.5); ax.set_ylim(-4, 4)
    ax.set_xticks([]); ax.set_yticks([])
    for name, (c, col) in regions.items():
        ax.scatter(*c, s=2600, color=col, alpha=0.10)
        ax.text(*c, name, color=col, ha='center', va='center', alpha=0.6)
    s = min(f // per, len(demo) - 1); local = f - s * per
    sent, ctx = demo[s]; traj = trajs[s]
    if local < 6:                      # sit context-free
        pos, layer = traj[0], 0
    elif local < 6 + n_layers * hold:  # drift through layers
        t = (local - 6) / hold; layer = min(int(t) + 1, n_layers)
        a, b = traj[min(int(t), n_layers - 1)], traj[min(int(t) + 1, n_layers)]
        pos = a + (t - int(t)) * (b - a)
    else:                              # settle
        pos, layer = traj[-1], n_layers
    for w in ctx:
        wp = context_pos[w]
        ax.scatter(*wp, s=45, color='0.35')
        ax.text(wp[0], wp[1] + 0.28, w, ha='center', fontsize=8.5, color='0.35')
        if layer > 0:
            ax.annotate('', xy=pos, xytext=wp,
                        arrowprops=dict(arrowstyle='->', color='0.6', lw=1,
                                        alpha=0.5, shrinkA=6, shrinkB=12))
    ax.scatter(*pos, s=260, color='C1', zorder=5, edgecolor='white', lw=1.5)
    ax.text(pos[0], pos[1] - 0.42, 'mole', ha='center', fontsize=11,
            fontweight='bold', color='C1')
    ax.set_title(f'"{sent}"', fontsize=11)
    ax.text(0.5, 0.02, 'context-free embedding' if layer == 0
            else f'after attention layer {layer}', transform=ax.transAxes,
            ha='center', fontsize=9.5, color='0.4')
    return []

anim = FuncAnimation(fig, frame, frames=per * len(demo), blit=False)
html = HTML(anim.to_jshtml(fps=fps)); plt.close(fig); html
```

Same word, same starting vector, different destinations — determined entirely by the neighbors in the sentence.  Two ideas are doing the work here, and both are part of the real mechanism:

* **Representations become contextual.**  The output vector at a position is no longer the word's generic embedding; it has moved to reflect *this* sentence.  This is the sense in which a transformer "understands" a word in context.
* **Depth refines.**  One layer gets *mole* into the right neighborhood; further layers can sharpen it (and, in a real model, resolve subtler things — which noun a pronoun refers to, whether "bank" is a verb here).  Stacking attention layers is stacking rounds of contextual refinement.

Notebook 1's model had two layers for exactly this reason.  Next is *how* a token knows which way to move — that is the next picture.

## The mechanism: a weighted pull from every other word

How does the mole vector "know" to head toward chemistry?  It doesn't decide in advance; it *looks at the other words and moves a little toward each*, pulling hardest toward the ones most relevant to it.  Its updated vector is a **weighted average** of the surrounding vectors: relevant neighbors get large weights, irrelevant ones get nearly zero.  The animation below fixes a focus word and shows the pull from each other word, with arrow thickness proportional to the weight.

```python
# words of a sentence, placed for legibility (again: an illustration)
toks = ['the', 'mole', 'burrowed', 'under', 'the', 'garden']
pos2 = {i: np.array(p) for i, p in enumerate(
    [(-4, 1.2), (-2, -0.6), (0, 1.4), (1.4, -1.1), (3, 1.0), (4.4, -0.7)])}
focus = 1  # 'mole'
# hand-set "relevance" of each word to the focus (large for burrowed/garden)
weights = np.array([0.05, 0.10, 0.42, 0.06, 0.05, 0.32])
target = sum(weights[i] * pos2[i] for i in pos2)  # weighted average
fps2, grow, move = 12, 16, 14

fig, ax = plt.subplots(figsize=(7, 3.6))

def frame2(f):
    ax.clear(); ax.set_xlim(-5.5, 5.5); ax.set_ylim(-2.6, 2.6)
    ax.set_xticks([]); ax.set_yticks([])
    reveal = min(f / grow, 1.0)                 # arrows grow in
    slide = max(0.0, min((f - grow) / move, 1)) # focus slides to target
    fp = pos2[focus] + slide * (target - pos2[focus])
    for i in pos2:
        if i == focus:
            continue
        w = weights[i] * reveal
        ax.annotate('', xy=fp, xytext=pos2[i],
                    arrowprops=dict(arrowstyle='-', color='C0',
                                    lw=0.5 + 14 * w, alpha=0.35, shrinkA=10,
                                    shrinkB=14))
        ax.scatter(*pos2[i], s=40, color='0.4')
        ax.text(pos2[i][0], pos2[i][1] + 0.28, toks[i], ha='center',
                fontsize=9, color='0.4')
    ax.scatter(*fp, s=320, color='C1', zorder=5, edgecolor='white', lw=1.5)
    ax.text(fp[0], fp[1] - 0.4, toks[focus], ha='center', fontsize=11,
            fontweight='bold', color='C1')
    ax.set_title('each word pulls on "mole"; thicker = more relevant',
                 fontsize=11)
    return []

anim2 = FuncAnimation(fig, frame2, frames=grow + move + 8, blit=False)
html2 = HTML(anim2.to_jshtml(fps=fps2)); plt.close(fig); html2
```

That is the entire shape of attention: **compute a relevance weight from a query word to every other word, then replace the focus word's vector with the weighted average of the others.**  Do this for every word as query, and you have transformed the whole sequence, each vector now informed by the ones that matter to it.  Do it again in the next layer, on the already-contextual vectors, and you refine further.

In this animation, the weights were set by hand.  The real mechanism computes them from the vectors themselves, via a comparison (a dot product) passed through a softmax so they are nonnegative and sum to one — and it also lets the network *learn* what "relevant" should mean rather than hard-coding it.  That machinery is  the subject of the next notebook.

In this animation cartoon the embedding of *mole* moves but leaves the context words fixed.  In a real layer, every embedding is updated.

## Stacking layers: back to the architecture

So far the movement has lived in an abstract "meaning space," detached from the boxes and arrows of an actual network.  Time to connect the two.  Recall the encoder diagram from notebook 1: input tokens along the bottom, one output vector per position along the top, and every output drawing on every input through attention.  That diagram showed a *single* attention layer.  The real model *stacks* them — notebook 1's had two — and stacking is exactly what lets the refinement happen in steps.

The animation below puts the architecture and the meaning space side by side.  On the left, the encoder is drawn as a grid: the bottom row is the embeddings, and each row above is one attention layer, whose nodes are computed from all the nodes of the row below (the faint web is the all-pairs attention; the orange edges highlight what flows into *mole*).  On the right is the same meaning space as before, tracking *mole*'s vector.  The two are synchronized: **each step up a layer on the left is one step of the drift on the right.**


```python
# Playback speed for this animation, in frames per second.
# LOWER = SLOWER.  Try 4 (slow) up to 12 (the original).  Re-run the cell.
FPS = 6

from matplotlib.patches import FancyBboxPatch

toks = ['<cls>', 'the', 'mole', 'burrowed', 'garden']
FOCUS, CLS, n_pos, n_layers = 2, 0, 5, 2
wts = np.array([0.08, 0.06, 0.12, 0.40, 0.34]); wts = wts / wts.sum()

regions = {'CHEMISTRY': (np.array([2.6, 2.2]), 'C0'),
           'ANIMAL':    (np.array([2.9, -2.3]), 'C2'),
           'ESPIONAGE': (np.array([-2.8, 1.4]), 'C3')}
mole_traj = [np.array([0.1, 0.1]), np.array([1.5, -1.15]),
             np.array([2.2, -1.85])]
anchor = {'burrowed': np.array([1.4, -2.9]), 'garden': np.array([3.4, -1.4])}

INTRO, BUILD, HOLD, READ = 8, 16, 5, 10
per = BUILD + HOLD
total = INTRO + n_layers * per + READ

def stage_of(f):
    if f < INTRO:
        return 0, None, f / INTRO, 'intro'
    g = f - INTRO
    if g >= n_layers * per:
        return n_layers, None, min((g - n_layers * per) / READ, 1), 'read'
    L, local = g // per + 1, g % per
    if local < BUILD:
        return L - 1, L, local / BUILD, 'build'
    return L, None, 1.0, 'hold'

figA, (axL, axR) = plt.subplots(1, 2, figsize=(11, 4.6),
                                gridspec_kw=dict(width_ratios=[1.35, 1]))

def draw_arch(f):
    built, building, sub, phase = stage_of(f)
    # ---- left: architecture grid ----
    axL.clear(); axL.set_xlim(-1.7, n_pos - 0.4); axL.set_ylim(-0.6, 3.5)
    axL.axis('off')
    for L, nm in enumerate(['embeddings', 'attention layer 1',
                            'attention layer 2']):
        axL.text(-1.6, L, nm, va='center', ha='left', fontsize=8.5,
                 color='0.4')
    if building is not None:
        for p in range(n_pos):
            for q in range(n_pos):
                axL.plot([p, q], [building - 1, building], color='0.6',
                         lw=0.6, alpha=0.10 * sub, zorder=1)
        for p in range(n_pos):
            axL.plot([p, FOCUS], [building - 1, building], color='C1',
                     lw=0.5 + 7 * wts[p] * sub, alpha=0.55 * sub, zorder=2)
    for L in range(1, built + 1):
        for p in range(n_pos):
            axL.plot([p, FOCUS], [L - 1, L], color='C1',
                     lw=0.4 + 5 * wts[p], alpha=0.18, zorder=1)
    for L in range(n_layers + 1):
        vis = (min(sub, 1) if phase == 'intro' else 1.0) if L == 0 else (
            1.0 if L <= built else sub if L == building else 0.0)
        if vis <= 0:
            continue
        for p in range(n_pos):
            col = 'C2' if p == CLS else 'C1' if p == FOCUS else '0.75'
            axL.scatter(p, L, s=230 if p in (CLS, FOCUS) else 150, color=col,
                        alpha=vis, zorder=5 if p in (CLS, FOCUS) else 4,
                        edgecolor='white', linewidths=1.3)
        if L == 0:
            for p in range(n_pos):
                axL.text(p, -0.42, toks[p], ha='center', fontsize=8.5,
                         alpha=vis, color='C2' if p == CLS else
                         ('C1' if p == FOCUS else '0.5'))
    if phase == 'read':
        a = sub
        axL.annotate('', xy=(CLS, 3.15), xytext=(CLS, n_layers + 0.15),
                     arrowprops=dict(arrowstyle='->', color='C2', lw=2,
                                     alpha=a))
        axL.add_patch(FancyBboxPatch((CLS - 0.7, 3.15), 1.4, 0.3,
                      boxstyle='round,pad=0.05', facecolor='C2',
                      edgecolor='none', alpha=a))
        axL.text(CLS, 3.3, 'classifier', ha='center', va='center',
                 color='white', fontsize=8.5, alpha=a)
    axL.set_title('the architecture: attention layers, stacked', fontsize=10.5)
    # ---- right: meaning space ----
    axR.clear(); axR.set_xlim(-4, 4.2); axR.set_ylim(-3.4, 3.9)
    axR.set_xticks([]); axR.set_yticks([])
    for name, (c, col) in regions.items():
        axR.scatter(*c, s=2500, color=col, alpha=0.10)
        axR.text(c[0], c[1] + 0.95, name, color=col, ha='center',
                 va='bottom', alpha=0.6, fontsize=9)
    for wd, ap in anchor.items():
        axR.scatter(*ap, s=40, color='0.4')
        axR.text(ap[0], ap[1] + 0.26, wd, ha='center', fontsize=8, color='0.4')
    if phase == 'intro':
        pos, lab = mole_traj[0], 'embedding (context-free)'
    elif phase == 'build':
        a, b = mole_traj[building - 1], mole_traj[building]
        pos, lab = a + sub * (b - a), f'climbing to layer {building}'
    elif phase == 'hold':
        pos, lab = mole_traj[built], f'after layer {built}'
    else:
        pos, lab = mole_traj[-1], f'after layer {n_layers}'
    upto = built if phase in ('hold', 'read') else (building or 0)
    if upto >= 1:
        path = np.array(mole_traj[:upto + 1])
        axR.plot(path[:, 0], path[:, 1], color='C1', lw=1, alpha=0.3, ls='--')
    axR.scatter(*pos, s=300, color='C1', zorder=5, edgecolor='white', lw=1.5)
    axR.text(pos[0], pos[1] - 0.42, 'mole', ha='center', fontsize=11,
             fontweight='bold', color='C1')
    axR.text(0.5, 0.02, lab, transform=axR.transAxes, ha='center',
             fontsize=9.5, color='0.4')
    axR.set_title("the same token's vector, refined", fontsize=10.5)
    return []

animA = FuncAnimation(figA, draw_arch, frames=total, blit=False)
htmlA = HTML(animA.to_jshtml(fps=FPS)); plt.close(figA); htmlA
```

This is the whole module in one picture.  A few things to read from it:

* **A layer is a refinement step.**  Each attention layer reads the previous layer's vectors and writes a new set — for *mole*, one hop closer to its contextual meaning.  Depth is not decoration; it is how the representation is sharpened in stages.  (This is also why notebook 1's model needed more than one layer to be interesting; a single layer gets you one hop.)
* **Every position moves, in parallel.**  We follow *mole* for concreteness, but the grid computes a new vector at *every* position at once — *the*, *garden*, and the rest are being refined by their own incoming attention (edges not drawn, to keep the picture legible).  The whole row updates in a single matrix operation, which is the parallelism that made the transformer fast in notebook 1.
* **The classifier reads `<cls>`.**  The final classifier uses only the refined embedding of the `<cls>` token.  Its vector was refined the same way *mole*'s was — by attending to the whole sentence.  So after multiple layers, it has pooled the evidence the decision needs.  That is why, back in notebook 1, we prepended `<cls>` and read the answer out of position 0.

Only one thing in this picture is still a black box: the orange edges have thicknesses (the attention weights), and here they were chosen them by hand.  Computing them from the vectors is the operation we build in the next notebook.


### Back to the heatmaps from notebook 1

Recall the heatmaps we extracted in notebook 1: those *are* the weights of the second animation.  Row $i$ of a map is the set of pulls on word $i$ — how much it draws from each other word — which is why every row summed to one.  The `<cls>` row was the summary slot pulling hardest from the words most useful for the sentiment decision.  When we plotted that "not/bad" example and saw weight concentrate on "bad," we were watching exactly this weighted-pull mechanism decide what mattered.

## Summary

* An embedding layer is a fixed lookup: a word gets the same vector in every context, which cannot capture words whose meaning depends on their neighbors.
* Attention produces *contextual* representations: each layer moves a token's vector toward the meaning implied by the tokens around it, and stacking layers refines the result.
* The move is a **weighted average of the other tokens' vectors**, with weights that measure relevance — large for the tokens that matter, near zero for the rest.
* The attention maps of notebook 1 are these weights.  The only missing piece is how the weights are computed and learned — the operation we introduce next.
