# Exercises: What Can a Recurrent Network Remember?

These two exercises let you probe the difference between a vanilla RNN and an LSTM on tasks where we know exactly what the network *should* do, so you can check what it actually learned.  All the machinery — the model classes, the training loop, the visualization helpers — is provided; your job is to run the experiments, look inside the trained networks, and answer the questions.

Both exercises use the same style of toy data as the counting task from the earlier notebooks: tiny alphabets, one-hot inputs, sequences short enough to train in seconds on a CPU.

```python
import torch
from torch import nn
import torch.nn.functional as F
import numpy as np
from matplotlib import pyplot as plt
```

## Setup

The classes below are the ones developed in the RNN and LSTM notebooks, with one small addition: `LSTMScratch` accepts a `forget_bias` argument that initializes the forget-gate bias $\mathbf{b}_\textrm{f}$ to a positive value, so that the memory starts out in "retain" mode.  Exercise 3 of the LSTM notebook asked why this classic trick helps — Exercise 1 below will show you.

```python
class RNNScratch(nn.Module):
    """An Elman RNN implemented from scratch."""
    def __init__(self, num_inputs, num_hiddens, sigma=0.1):
        super().__init__()
        self.num_hiddens = num_hiddens
        self.W_xh = nn.Parameter(torch.randn(num_inputs, num_hiddens) * sigma)
        self.W_hh = nn.Parameter(torch.randn(num_hiddens, num_hiddens) * sigma)
        self.b_h = nn.Parameter(torch.zeros(num_hiddens))

    def forward(self, inputs, H=None):
        if H is None:
            H = torch.zeros(inputs.shape[1], self.num_hiddens,
                            device=inputs.device)
        outputs = []
        for X_t in inputs:
            H = torch.tanh(X_t @ self.W_xh + H @ self.W_hh + self.b_h)
            outputs.append(H)
        return torch.stack(outputs), H


class LSTMScratch(nn.Module):
    def __init__(self, num_inputs, num_hiddens, sigma=0.1, forget_bias=0.0):
        super().__init__()
        self.num_hiddens = num_hiddens
        init_weight = lambda *shape: nn.Parameter(torch.randn(*shape) * sigma)
        triple = lambda b=0.0: (
            init_weight(num_inputs, num_hiddens),
            init_weight(num_hiddens, num_hiddens),
            nn.Parameter(torch.full((num_hiddens,), float(b))))
        self.W_xi, self.W_hi, self.b_i = triple()            # Input gate
        self.W_xf, self.W_hf, self.b_f = triple(forget_bias) # Forget gate
        self.W_xo, self.W_ho, self.b_o = triple()            # Output gate
        self.W_xc, self.W_hc, self.b_c = triple()            # Input node

    def one_step(self, X, H, C):
        I = torch.sigmoid(X @ self.W_xi + H @ self.W_hi + self.b_i)
        F_ = torch.sigmoid(X @ self.W_xf + H @ self.W_hf + self.b_f)
        O = torch.sigmoid(X @ self.W_xo + H @ self.W_ho + self.b_o)
        C_tilde = torch.tanh(X @ self.W_xc + H @ self.W_hc + self.b_c)
        C = F_ * C + I * C_tilde
        H = O * torch.tanh(C)
        return H, C, I, F_, O

    def forward(self, inputs, H_C=None):
        if H_C is None:
            H = torch.zeros(inputs.shape[1], self.num_hiddens,
                            device=inputs.device)
            C = torch.zeros_like(H)
        else:
            H, C = H_C
        outputs = []
        for X_t in inputs:
            H, C, _, _, _ = self.one_step(X_t, H, C)
            outputs.append(H)
        return torch.stack(outputs), (H, C)


class Classifier(nn.Module):
    """Recurrent core + linear read-out of the final hidden state.
    Works with both cores: the RNN returns (outputs, H), the LSTM
    returns (outputs, (H, C))."""
    def __init__(self, core, num_hiddens, num_classes=2):
        super().__init__()
        self.core = core
        self.decoder = nn.Linear(num_hiddens, num_classes)

    def forward(self, X):
        outputs, state = self.core(X.permute(1, 0, 2))
        H = state[0] if isinstance(state, tuple) else state
        return self.decoder(H)


def evaluate_accuracy(model, X, y):
    model.eval()
    with torch.no_grad():
        return (model(X).argmax(dim=1) == y).float().mean().item()


def train_model(model, X_train, y_train, X_test, y_test,
                epochs=20, lr=0.01, batch_size=64, verbose=True):
    train_iter = torch.utils.data.DataLoader(
        torch.utils.data.TensorDataset(X_train, y_train),
        batch_size=batch_size, shuffle=True)
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    loss_fn = nn.CrossEntropyLoss()
    for epoch in range(epochs):
        model.train()
        losses = []
        for batch_X, batch_y in train_iter:
            loss = loss_fn(model(batch_X), batch_y)
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            losses.append(loss.item())
        if verbose and (epoch + 1) % 5 == 0:
            print(f'  epoch {epoch + 1:2d}  loss {np.mean(losses):.3f}  '
                  f'test accuracy {evaluate_accuracy(model, X_test, y_test):.3f}')
    return evaluate_accuracy(model, X_test, y_test)


def trace_lstm(lstm, X):
    """Run one sequence (a (seq_len, num_inputs) float matrix) step by step,
    recording the cell state, hidden state, and all three gates."""
    H = torch.zeros(1, lstm.num_hiddens)
    C = torch.zeros_like(H)
    rec = {k: [] for k in ['H', 'C', 'I', 'F', 'O']}
    with torch.no_grad():
        for X_t in X:
            H, C, I, F_, O = lstm.one_step(X_t.unsqueeze(0), H, C)
            for k, v in zip(['H', 'C', 'I', 'F', 'O'], [H, C, I, F_, O]):
                rec[k].append(v.squeeze(0))
    return {k: torch.stack(v) for k, v in rec.items()}
```

## Exercise 1: remembering the first token

**Task**: the *first* token of an `a`/`b` sequence is the label; every token after it is random noise drawn from the same alphabet.  To classify correctly, the network must do exactly one thing: store the first token, and protect that memory while the distractors stream past.

This deceptively simple problem is the *latch* task that Bengio, Simard & Frasconi (1994) used to demonstrate why learning long-term dependencies with gradient descent is hard — the teaching signal at the end of the sequence must reach a decision made at $t = 1$.  The optimal LSTM solution is easy to describe: write the first token into a memory cell (input gate open at $t=1$, closed after), and never flush it (forget gate at 1).

```python
def make_recall_data(num_examples, seq_len, seed=0):
    """First token is the label; the rest are random distractors."""
    g = torch.Generator().manual_seed(seed)
    tokens = torch.randint(0, 2, (num_examples, seq_len), generator=g)
    labels = tokens[:, 0].clone()
    return F.one_hot(tokens, num_classes=2).float(), labels, tokens

X, y, tokens = make_recall_data(5, 12, seed=2)
for seq, label in zip(tokens, y):
    print(''.join('ab'[t] for t in seq), ' label:', 'ab'[label])
```

Let's train both architectures at two sequence lengths.  (Training is deterministic given the seeds, so your numbers should match your neighbor's.)

```python
num_hiddens = 8
results = {}
for seq_len in [8, 20]:
    X_train, y_train, _ = make_recall_data(2000, seq_len, seed=0)
    X_test, y_test, _ = make_recall_data(500, seq_len, seed=1)
    for name, make_core in [
            ('RNN', lambda: RNNScratch(2, num_hiddens)),
            ('LSTM', lambda: LSTMScratch(2, num_hiddens, forget_bias=2.0))]:
        torch.manual_seed(0)
        model = Classifier(make_core(), num_hiddens)
        print(f'{name}, seq_len={seq_len}:')
        acc = train_model(model, X_train, y_train, X_test, y_test,
                          epochs=20)
        results[(name, seq_len)] = (model, acc)

print()
for (name, seq_len), (_, acc) in results.items():
    print(f'{name:5s} seq_len={seq_len:3d}  test accuracy {acc:.2f}')
```

At length 8 both architectures solve the task; at length 20 the vanilla RNN collapses to chance while the LSTM remains perfect.

Let's look inside the successful LSTM.  The cell below feeds it two sequences that are *identical except for the first token* and traces the memory cell of the unit the decoder relies on most, along with its gates.

```python
model_lstm = results[('LSTM', 20)][0]
u = (model_lstm.decoder.weight[1]
     - model_lstm.decoder.weight[0]).abs().argmax().item()

_, _, toks = make_recall_data(1, 20, seed=7)
seq_a, seq_b = toks[0].clone(), toks[0].clone()
seq_a[0], seq_b[0] = 0, 1     # same distractors, different first token

fig, axes = plt.subplots(2, 1, figsize=(7, 5), sharex=True)
for seq, style, label in [(seq_a, '-', 'first token a'),
                          (seq_b, '--', 'first token b')]:
    rec = trace_lstm(model_lstm.core, F.one_hot(seq, 2).float())
    axes[0].plot(rec['C'][:, u], style, marker='.', label=label)
    axes[1].plot(rec['I'][:, u], style, color='C2', marker='.',
                 label=f'input gate, {label}')
    axes[1].plot(rec['F'][:, u], style, color='C3', marker='.',
                 label=f'forget gate, {label}')
axes[0].axhline(0, color='gray', lw=0.5)
axes[0].set_ylabel(f'cell state $C_t$ (unit {u})')
axes[0].legend()
axes[1].set_ylim(-0.05, 1.05)
axes[1].set_xlabel('time step')
axes[1].set_ylabel('gate value')
axes[1].legend(fontsize=8);
```

**Questions**

1. Describe what you see in the two panels.  Where is the first token stored?  What do the input and forget gates do at $t = 1$ compared with the rest of the sequence?  Is this the "write once, hold" solution described above?
2. Why does the vanilla RNN succeed at length 8 but sit at chance at length 20?  Answer in terms of what happens to the hidden state each time a distractor arrives, and in terms of the gradient-distance figure from the RNN notebook: how many multiplications separate the loss from the only informative input?  (Try a few intermediate lengths to find where the RNN breaks — is the transition gradual or abrupt?)
3. Retrain the LSTM at `seq_len=20` with `forget_bias=0.0`.  What happens, and why does starting the forget gate near $\sigma(0) = 0.5$ doom the memory?  (How much of the cell state survives 19 steps of $F \approx 0.5$?)
4. Push `seq_len` to 30 and 40 and retrain the LSTM a few times (vary `torch.manual_seed`).  You should find that success becomes hit-or-miss: the LSTM *architecture* can represent the solution at any length, but gradient descent does not reliably *find* it.  This fragility is the historical reality documented by Bengio et al. — and part of why the field eventually moved to architectures with even shorter gradient paths, as we will see in the attention module.

## Exercise 2: selective counting

**Task**: sequences over the four-token alphabet $\{\texttt{a}, \texttt{b}, \texttt{[}, \texttt{]}\}$ contain two bracketed regions.  The label is 1 if the tokens *inside the brackets* contain more `b`s than `a`s.  The `a`s and `b`s outside the brackets are distractors — nothing about a token tells you whether it counts; only its *context* (inside or outside brackets) does.

This is the counting task from the RNN notebook plus one new requirement: the network must learn *when* to count.  The ideal LSTM solution uses the input gate as a switch — open inside brackets, closed outside — which is exactly the mechanism the gates were designed to provide.

One detail of the data generation deserves attention.  The imbalance of the *outside* tokens is drawn independently of the label and made *larger* than the inside imbalance ($\pm 5$ outside vs. $\pm 3$ inside).  Question 3 asks why.

```python
def make_selective_data(num_examples, seed=0, independent_outside=True):
    """Sequences of length 40 over {a=0, b=1, [=2, ]=3} with two bracketed
    regions.  Label: more b's than a's inside the brackets (imbalance +-3).
    Outside tokens (imbalance +-5) are independent of the label when
    independent_outside=True; otherwise they are sampled uniformly."""
    g = torch.Generator().manual_seed(seed)
    seqs, labels = [], []
    m, n_out = 13, 23          # tokens inside / outside brackets
    for i in range(num_examples):
        label = int(torch.randint(0, 2, (1,), generator=g))
        b_in = (m + 3) // 2 if label else (m - 3) // 2
        inside = torch.cat([torch.ones(b_in), torch.zeros(m - b_in)]).long()
        inside = inside[torch.randperm(m, generator=g)]
        if independent_outside:
            sign = int(torch.randint(0, 2, (1,), generator=g))
            b_out = (n_out + 5) // 2 if sign else (n_out - 5) // 2
            outside = torch.cat([torch.ones(b_out),
                                 torch.zeros(n_out - b_out)]).long()
            outside = outside[torch.randperm(n_out, generator=g)]
        else:
            outside = torch.randint(0, 2, (n_out,), generator=g)
        # layout: out [ in ] out [ in ] out
        s1 = int(torch.randint(3, m - 2, (1,), generator=g))
        c1 = int(torch.randint(0, n_out + 1, (1,), generator=g))
        c2 = int(torch.randint(0, n_out - c1 + 1, (1,), generator=g))
        seq = torch.cat([
            outside[:c1], torch.tensor([2]), inside[:s1], torch.tensor([3]),
            outside[c1:c1 + c2], torch.tensor([2]), inside[s1:],
            torch.tensor([3]), outside[c1 + c2:]])
        seqs.append(seq)
        labels.append(label)
    tokens = torch.stack(seqs)
    return (F.one_hot(tokens, num_classes=4).float(),
            torch.tensor(labels), tokens)

X, y, tokens = make_selective_data(3, seed=0)
for seq, label in zip(tokens, y):
    print(''.join('ab[]'[t] for t in seq), ' label:', int(label))
```

```python
num_hiddens = 16
X_train, y_train, _ = make_selective_data(3000, seed=0)
X_test, y_test, tokens_test = make_selective_data(500, seed=1)

models = {}
for name, make_core in [
        ('RNN', lambda: RNNScratch(4, num_hiddens)),
        ('LSTM', lambda: LSTMScratch(4, num_hiddens, forget_bias=1.0))]:
    torch.manual_seed(0)
    model = Classifier(make_core(), num_hiddens)
    print(f'{name}:')
    acc = train_model(model, X_train, y_train, X_test, y_test, epochs=10)
    models[name] = (model, acc)

print()
for name, (_, acc) in models.items():
    print(f'{name:5s} test accuracy {acc:.2f}')
```

The LSTM solves the task; the vanilla RNN gets stuck well short of it.  But *how* does the LSTM do it?  The cleanest way to see the gating is functionally: track the decoder's **evidence** (the difference between the two class scores, as in the RNN notebook) while the network reads, and measure how much each token moves it — inside versus outside the brackets.

```python
model_lstm = models['LSTM'][0]
w = model_lstm.decoder.weight.detach()
b = model_lstm.decoder.bias.detach()

def evidence_trace(model, seq):
    """Decoder evidence for class 1 after each time step of one sequence."""
    X = F.one_hot(seq, num_classes=4).float()
    H = torch.zeros(1, model.core.num_hiddens)
    C = torch.zeros_like(H)
    ev = []
    with torch.no_grad():
        for X_t in X:
            H, C, _, _, _ = model.core.one_step(X_t.unsqueeze(0), H, C)
            ev.append(float(H.squeeze(0) @ (w[1] - w[0]) + b[1] - b[0]))
    return ev

fig, axes = plt.subplots(1, 2, figsize=(11, 3.6), width_ratios=[2, 1])

# left: evidence over time for one test sequence, brackets shaded
seq = tokens_test[0]
axes[0].plot(evidence_trace(model_lstm, seq), marker='.', color='C0')
axes[0].axhline(0, color='gray', lw=0.5)
in_region = False
for t, tok in enumerate(seq):
    if tok == 2:
        start, in_region = t, True
    elif tok == 3 and in_region:
        axes[0].axvspan(start, t, color='gold', alpha=0.25)
        in_region = False
axes[0].set_xlabel('time step')
axes[0].set_ylabel("evidence for \"more b's inside\"")
axes[0].set_title('evidence while reading (brackets shaded)')

# right: average effect of each token on the evidence, by context
effects = {k: [] for k in [('in', 'a'), ('in', 'b'),
                           ('out', 'a'), ('out', 'b')]}
for s in range(100):
    seq = tokens_test[s]
    ev = evidence_trace(model_lstm, seq)
    depth, prev = 0, 0.0
    for t, tok in enumerate(seq.tolist()):
        if tok in (0, 1):
            effects[('in' if depth else 'out', 'ab'[tok])].append(ev[t] - prev)
        if tok == 2:
            depth = 1
        elif tok == 3:
            depth = 0
        prev = ev[t]
keys = [('in', 'a'), ('in', 'b'), ('out', 'a'), ('out', 'b')]
vals = [np.mean(effects[k]) for k in keys]
axes[1].bar(['a inside', 'b inside', 'a outside', 'b outside'], vals,
            color=['C3', 'C2', 'C3', 'C2'])
axes[1].axhline(0, color='gray', lw=0.5)
axes[1].tick_params(axis='x', rotation=20)
axes[1].set_ylabel('mean change in evidence per token')
axes[1].set_title('the same token, gated by context');
```

**Questions**

1. Interpret the two panels.  What does the bar chart say about how the network responds to an `a` or a `b` inside the brackets compared with the *same token* outside?  This *functional* gating — input effect switched on and off by context — is exactly what the task demands.  You may also notice that the evidence leaps just after the second bracket closes: which token drives that jump, and what might the network be doing there?
2. Now hunt for the mechanism.  Using `trace_lstm`, compare the input and forget gate values inside versus outside the brackets, averaged over test sequences, for each of the 16 units.  You will likely find the contrast modest and spread across several units — the network can implement gating through the input gate, the forget gate, or the candidate values $\tilde{C}_t$ (which also receive the hidden state).  Is there a single "gate unit", or is the strategy distributed?  This is typical of trained networks: the architecture provides mechanisms, and gradient descent uses whatever combination works.
3. The vanilla RNN scores well above chance but far below the LSTM.  What partial strategy might it have found?  (Hint: a saturating $\tanh$ unit driven by the bracket tokens can *approximately* modulate how strongly `a`/`b` inputs move the state — a soft, leaky imitation of a gate.)
4. Regenerate the data with `independent_outside=False`, so the outside tokens are uniform, and retrain the vanilla RNN.  Its accuracy should jump.  Explain why: what shortcut does uniform outside noise leave open, and why does the independent $\pm 5$ outside imbalance close it?  This is a general lesson: *a dataset teaches the easiest strategy it fails to rule out* — worth remembering whenever you design an evaluation.
5. (Optional) Modify the generator to allow *nested* brackets, counting only tokens at nesting depth $\geq 1$.  Does the LSTM still cope?  What extra quantity must the state now track?
