# Deep Convolutional Neural Networks: AlexNet

The previous module ended with a working convolutional network.  LeNet-5 reached
about 1% error on handwritten digits in 1998 and was deployed to read the amounts
on cheques — a real system, doing real work, built out of exactly the pieces we
derived.  And then, for roughly fourteen years, almost nothing happened.

Convolutional networks were known in the computer vision community, but they did
not take it over.  Through the 2000s, the methods that won benchmarks were kernel
machines and boosted ensembles applied not to pixels but to *hand-designed
features*.  Then, in 2012, a single result changed the field:

> Krizhevsky, A., Sutskever, I., & Hinton, G. E. (2012). ImageNet classification
> with deep convolutional neural networks. *Advances in Neural Information
> Processing Systems*, 25.

Their network — universally called **AlexNet**, after its first author — won the
ImageNet Large Scale Visual Recognition Challenge with a top-5 error rate of
15.3%, against 26.2% for the second-place entry.  A ten-point margin in a
competition where the previous years' progress had been measured in single
points is not an incremental result; it is an argument that the whole pipeline
was wrong.

What makes AlexNet worth a notebook is that, architecturally, it is *not* a new
idea.  It is LeNet, deeper and wider, with a handful of changes that each address
a specific failure we can already name.  The interesting question is therefore
not "what is AlexNet" but "why did it take fourteen years", and the answer to
that is most of what this notebook is about.

## Learned features against crafted features

Before 2012 a computer vision system was a pipeline, and the learning happened
only at the very end of it:

1. Collect a dataset.  In the 1990s this required expensive sensors: the Apple
   QuickTake 100 of 1994 offered 0.3 megapixels and storage for eight images, for
   \$1000.
2. Extract features with a hand-designed operator — SIFT, SURF, HOG and their
   relatives — each the product of a paper's worth of thinking about optics and
   geometry.
3. Feed the resulting fixed-length vectors to a classifier, typically a linear
   model or an SVM.

Only step 3 was learned.  A vision researcher of the era would have told you,
correctly for the time, that progress came from steps 1 and 2, and that the
choice of classifier barely mattered.

A minority position — held by LeCun, Hinton, Bengio and others — was that
**step 2 should be learned too**, and learned *hierarchically*: early layers
detecting edges and colours and textures, later layers composing those into
parts, and later ones still into objects.  This is exactly the argument we made
from the biology in notebook 04_01, and LeNet is exactly such a network.  So the
idea was available.  What was not available was everything needed to make it
work at scale.

### The missing data

Deep models earn their keep only when there is enough data to constrain their
parameters.  Through the 1990s and 2000s most vision research ran on datasets of
hundreds or a few thousand low-resolution images, often with artificially clean
backgrounds.  On data of that size a model with millions of parameters has
nothing to recommend it.

**ImageNet**, released in 2009, changed the scale by an order of magnitude: the
competition subset holds roughly 1.2 million training images across 1000
categories, drawn from web image search and verified by crowd workers, at
$224 \times 224$ resolution rather than the $32 \times 32$ thumbnails that other
large collections used.  Both numbers matter.  A thousand fine-grained
categories — ImageNet distinguishes dozens of dog breeds — force the network to
learn features that separate genuinely similar things, and the resolution leaves
enough detail for a hierarchy of features to have something to build on.

### The missing compute

The second obstacle was arithmetic.  Training a deep network means passing
millions of images through many layers of dense linear algebra, hundreds of
times over.

*Graphics* processors turned out to be the answer, for an accidental reason:
the operation a GPU was built to do quickly — many small matrix products, run
in parallel with almost no branching — is the operation a convolutional layer
needs.  Krizhevsky and Sutskever wrote convolution kernels that ran on two
NVIDIA GTX 580 cards, each with 3 GB of memory and about 1.5 TFLOPS, and trained
for around five days.  Their `cuda-convnet` code was the industry standard for
several years afterwards.

The gap in throughput between a CPU and a GPU is not a temporary engineering
detail; it follows from what each is for.  A CPU core spends most of its silicon
on branch prediction, speculative execution and cache, so that it can run
arbitrary control flow fast, and a machine has a handful of them.  A GPU spends
its silicon on thousands of weak arithmetic units with wide memory buses, which
is useless for general code and ideal for this one.

It is worth measuring how far that has come, because the training run later in
this notebook makes it concrete.  A GTX 580 was rated at 1.58 TFLOPS of
single-precision arithmetic and drew 244 W, and Krizhevsky used two of them.
The GPU in an M2 MacBook Air is rated at 2.9 TFLOPS for the eight-core part and
3.6 for the ten-core, and the entire machine — screen included — draws on the
order of 15 W under load.  A fanless laptop now has roughly the peak throughput
of the pair of cards AlexNet was trained on, at a small fraction of the power,
and with no CUDA kernels left to write.

And yet the run below takes the better part of an hour.  The arithmetic says why.
One forward pass through this network is about 0.9 billion multiply-accumulates
per image; the backward pass costs roughly twice that; an epoch is 60,000
images.  Ten epochs is on the order of $10^{15}$ floating-point operations.
Krizhevsky's run was almost two hundred times larger — 1.2 million images for 90
epochs — which at the rate you are about to observe would be about a week.  They
reported five to six days.

If you are on a Mac, open Activity Monitor's GPU history window while the
training cell runs.  You should see the GPU pinned near 100% and the CPU almost
idle.  That picture is the argument of this section in a single frame: nearly all
of the work is arithmetic of a shape a graphics processor was already built to
do, and almost none of it is the general-purpose control flow a CPU is good at.
LeNet, in 1998, ran on a CPU.

So the ingredients arrived together, and rather late: a dataset large enough to
justify a big model, and hardware fast enough to fit one.

## The architecture

AlexNet is an eight-layer network: five convolutional layers and three fully
connected ones.  Side by side with LeNet, the family resemblance is obvious.

![From LeNet (left) to AlexNet (right).](https://github.com/d2l-ai/d2l-en/raw/refs/heads/master/img/alexnet.svg)

The version we build below is slightly streamlined.  The original was split
across two GPUs, with only some layers exchanging information between the two
halves — a memory constraint of 2012 hardware, drawn explicitly in the paper's
own figure:



Nobody would do this today, and we do not.

Four things distinguish AlexNet from LeNet, and each of them answers a problem
we have already seen.

**It is deeper and much wider.**  Five convolutional layers instead of two, and
roughly ten times the channels — 96 where LeNet had 6.  Notebook 04_04 gave the
reason channels are the right axis to grow along: they are orthogonal to
locality and weight sharing, so adding them costs nothing conceptually while
letting each layer detect more kinds of thing.

**The first kernel is $11 \times 11$ with stride 4.**  ImageNet images are eight
times taller and wider than MNIST digits, so an object occupies far more pixels;
a $5 \times 5$ window would see almost nothing of it.  The large stride is a
resolution cut, exactly the mechanism from notebook 04_03 — it takes 224 down to
54 in one step and makes the rest of the network affordable.  Later kernels
shrink to $5 \times 5$ and then $3 \times 3$, and VGG will take that trend to its
conclusion.

**ReLU replaces sigmoid, and max-pooling replaces average pooling.**  This is
the plateau we watched LeNet sit in for several epochs.  A sigmoid saturates and
its gradient vanishes; ReLU has derivative exactly 1 on the positive side, so
gradients reach the early layers from the first step.  For a network of eight
layers rather than five this stops being an inconvenience and becomes the
difference between training and not training.

**Dropout regularizes the fully connected head.**  LeNet used weight decay.
AlexNet's head has tens of millions of parameters, and dropout — randomly zeroing
half the units during training, which you have already met in the transformer —
is what keeps it from memorizing the training set.  The authors also augmented
the data heavily, with random crops, horizontal flips and colour perturbations;
an image the network never sees twice in the same form is harder to memorize.

Here is the whole thing:

```python
def make_alexnet(num_classes=10):
    """AlexNet, streamlined: single stream rather than the original's two GPUs."""
    return nn.Sequential(
        nn.LazyConv2d(96, kernel_size=11, stride=4, padding=1), nn.ReLU(),
        nn.MaxPool2d(kernel_size=3, stride=2),
        nn.LazyConv2d(256, kernel_size=5, padding=2), nn.ReLU(),
        nn.MaxPool2d(kernel_size=3, stride=2),
        nn.LazyConv2d(384, kernel_size=3, padding=1), nn.ReLU(),
        nn.LazyConv2d(384, kernel_size=3, padding=1), nn.ReLU(),
        nn.LazyConv2d(256, kernel_size=3, padding=1), nn.ReLU(),
        nn.MaxPool2d(kernel_size=3, stride=2),
        nn.Flatten(),
        nn.LazyLinear(4096), nn.ReLU(), nn.Dropout(p=0.5),
        nn.LazyLinear(4096), nn.ReLU(), nn.Dropout(p=0.5),
        nn.LazyLinear(num_classes))
```

Forty lines shorter than the description that precedes it, which is the point
of the closing remark in the LeNet notebook: the 2012 implementation was months
of CUDA, and this is a `Sequential`.

Note the three max-pooling layers, all $3 \times 3$ with stride 2.  The window is
larger than the stride, so successive pooling windows *overlap* — the paper
reports this as slightly better than the non-overlapping pooling LeNet used.

As before, the `Lazy` layers infer their input size on the first forward pass,
so we state only output sizes and never have to work out by hand what the
flattened encoder produces.

## Following the shapes

The same walk we did for LeNet, now on a $224 \times 224$ single-channel image.

```python
def layer_summary(net, X_shape):
    """Print the output shape of every layer for an input of the given shape."""
    X = torch.randn(*X_shape)
    for layer in net:
        X = layer(X)
        print(f'{layer.__class__.__name__:12s} output shape: {tuple(X.shape)}')

model = make_alexnet()
layer_summary(model, (1, 1, 224, 224))
```

Read the height and width column downward: 224, then 54 after the strided first
convolution, 26, 12, and finally 5.  Read the channel column at the same time: 1,
96, 256, 384, 384, 256.  This is the pattern from the previous module, over a
longer run — spatial resolution down, channel depth up — and the network's job is
to trade one for the other until what remains is a description of the image
rather than a picture of it.

The flatten produces $256 \times 5 \times 5 = 6400$ numbers.  That is where the
trouble starts.

```python
def parameter_summary(net):
    """Print the parameter count of every tensor in the network."""
    total = 0
    for name, param in net.named_parameters():
        print(f'{name:16s} {str(tuple(param.shape)):>20s} {param.numel():>12,d}')
        total += param.numel()
    print('-' * 52)
    print(f'{"total":16s} {"":20s} {total:>12,d}')
    return total

num_params = parameter_summary(model)
```

Roughly 47 million parameters, against LeNet's 62 thousand — a factor of about
750.  Stored as 32-bit floats that is 187 MB for the weights alone, which is why
the model had to be split across two 3 GB cards in 2012.

And the distribution is the same one we found in LeNet, only more extreme.  The
two $4096$-wide layers after the flatten hold about 43 million parameters between
them, over 90% of the network.  The five convolutional layers that do all of the
perceptual work hold under 4 million.

Parameters are not the whole story, though.  Let's count the arithmetic as well,
using the cost formula from notebook 04_04 — a convolutional layer performs
$h \cdot w \cdot k^2 \cdot c_i \cdot c_o$ multiply-accumulates, one for every
weight at every output position, while a dense layer performs exactly one per
weight.

```python
def cost_summary(net, X_shape):
    """Parameters and multiply-accumulates per layer, for a single image."""
    X = torch.randn(*X_shape)
    print(f'{"layer":14s} {"output shape":>18s} {"params":>12s} {"MACs":>14s}')
    print('-' * 62)
    totals = {'conv': [0, 0], 'dense': [0, 0]}
    for layer in net:
        X = layer(X)
        params = sum(p.numel() for p in layer.parameters())
        if isinstance(layer, nn.Conv2d):
            # one multiply-accumulate per weight per output position
            macs = layer.weight.numel() * X.shape[-2] * X.shape[-1]
            kind = 'conv'
        elif isinstance(layer, nn.Linear):
            macs = layer.weight.numel()
            kind = 'dense'
        else:
            macs, kind = 0, None
        if kind:
            totals[kind][0] += params
            totals[kind][1] += macs
        print(f'{layer.__class__.__name__:14s} {str(tuple(X.shape)):>18s} '
              f'{params:>12,d} {macs:>14,d}')
    print('-' * 62)
    for kind, (params, macs) in totals.items():
        print(f'{kind + " layers":14s} {"":>18s} {params:>12,d} {macs:>14,d}')
    return totals

totals = cost_summary(make_alexnet(), (1, 1, 224, 224))

params_conv, macs_conv = totals['conv']
params_dense, macs_dense = totals['dense']
print()
print(f'convolutions: {params_conv / (params_conv + params_dense):.1%} of parameters, '
      f'{macs_conv / (macs_conv + macs_dense):.1%} of the arithmetic')
```

The two columns point in opposite directions.  The convolutional layers hold
about 8% of the parameters and do about 95% of the arithmetic; the dense head
holds 92% of the parameters and does 5% of the arithmetic.

That inversion is worth sitting with, because it explains what the next two
notebooks do.  Weight sharing is what causes it: a convolutional kernel's
parameters are reused at every spatial position, so its parameter count is tiny
relative to the work it performs, while a dense layer uses each weight exactly
once.  A network is therefore *memory*-limited at its head and *compute*-limited
in its body, and the two are optimized by different means.  Every architecture
after AlexNet attacks the head — VGG shrinks it, and ResNet removes it almost
entirely in favour of global average pooling.

## Training

AlexNet was trained on ImageNet.  We will not be: even on a modern GPU that is a
matter of hours, and on a laptop it is not a matter of hours.  Instead we keep
the architecture and change the data, training on Fashion-MNIST upsampled from
$28 \times 28$ to $224 \times 224$.

This is, to be clear, a silly thing to do.  Upsampling adds no information; it
just makes every convolution 64 times more expensive so that a network designed
for detailed photographs will accept a small grayscale thumbnail.  We do it
because the alternative is to change the architecture, and the architecture is
the thing we came to look at.  Exercise 2 asks you to design the sensible
version.

Be warned that this is by far the slowest training run in the course so far —
expect minutes per epoch rather than seconds.  Now is the time to use a GPU if
you have one.

```python
batch_size = 128
data_dir = './data'

# A full 10-epoch run over all 60,000 images takes the better part of an hour.
# Set train_subset to an integer for a smaller, class-balanced training set:
# 10_000 brings the run down to roughly ten minutes and costs a few points of
# accuracy, without changing the shape of the curves.  None uses everything.
train_subset = None

# Resize before ToTensor: the resize operates on the PIL image
transform = transforms.Compose([transforms.Resize((224, 224)),
                                transforms.ToTensor()])

train_set = datasets.FashionMNIST(data_dir, train=True, download=True,
                                  transform=transform)
val_set = datasets.FashionMNIST(data_dir, train=False, download=True,
                                transform=transform)
labels = train_set.classes

if train_subset is not None:
    # take the same number of examples from each class
    per_class = train_subset // len(labels)
    indices = torch.cat([torch.where(train_set.targets == c)[0][:per_class]
                         for c in range(len(labels))])
    train_set = torch.utils.data.Subset(train_set, indices.tolist())

train_iter = DataLoader(train_set, batch_size, shuffle=True)
val_iter = DataLoader(val_set, batch_size)

print(f'{len(train_set)} training and {len(val_set)} validation images, '
      f'shape {tuple(val_set[0][0].shape)}')
```

We keep the Xavier initialization from the LeNet notebook.  With ReLU
activations it matters much less than it did with sigmoids — that is rather the
point — but a sensible starting scale is never wasted.

The learning rate is 0.01 rather than the 0.1 we used for LeNet.  A deeper
network with unbounded ReLU activations takes larger steps for the same nominal
rate, and 0.1 is enough to diverge here.

```python
def init_cnn(module):
    """Xavier-initialize the weights of convolutional and linear layers."""
    if type(module) in (nn.Linear, nn.Conv2d):
        nn.init.xavier_uniform_(module.weight)
        nn.init.zeros_(module.bias)

def accuracy(Y_hat, Y):
    """Fraction of correct predictions."""
    return (Y_hat.argmax(dim=1) == Y).float().mean()

model = make_alexnet().to(device)
# the Lazy layers need to see data once before their parameters exist
model(next(iter(train_iter))[0].to(device))
model.apply(init_cnn)

optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
loss_fn = nn.CrossEntropyLoss()
```

```python
epochs = 10
loss_train, loss_valid, accuracy_valid = [], [], []

for epoch in range(epochs):
    start = time.time()
    model.train()
    loss_values = []
    for batch_X, batch_y in train_iter:
        batch_X, batch_y = batch_X.to(device), batch_y.to(device)
        preds = model(batch_X)
        loss = loss_fn(preds, batch_y)
        loss_values.append(loss.item())
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
    loss_train.append(np.mean(loss_values))

    model.eval()
    loss_values, accuracy_values = [], []
    for batch_X, batch_y in val_iter:
        batch_X, batch_y = batch_X.to(device), batch_y.to(device)
        with torch.no_grad():
            preds = model(batch_X)
            loss_values.append(loss_fn(preds, batch_y).item())
            accuracy_values.append(accuracy(preds, batch_y).item())
    loss_valid.append(np.mean(loss_values))
    accuracy_valid.append(np.mean(accuracy_values))

    print(f'epoch {epoch + 1:2d}  train loss {loss_train[-1]:.3f}  '
          f'val loss {loss_valid[-1]:.3f}  '
          f'val accuracy {accuracy_valid[-1]:.3f}  '
          f'({time.time() - start:.1f} s)')
```

```python
# loss and accuracy live on different scales, so give each its own y-axis
fig, ax1 = plt.subplots(figsize=(5.5, 3.5))
epochs_range = range(1, len(loss_train) + 1)
ax1.plot(epochs_range, loss_train, color='C0', label='training loss')
ax1.plot(epochs_range, loss_valid, color='C1', label='validation loss')
ax1.set_xlabel('epoch')
ax1.set_ylabel('loss')
ax2 = ax1.twinx()
ax2.plot(epochs_range, accuracy_valid, color='C2',
         label='validation accuracy')
ax2.set_ylabel('accuracy')
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='center right');
```

Two things about these curves are worth naming, and both are contrasts with
LeNet.

**There is no plateau.**  The loss falls from the first epoch.  This is the ReLU
doing its job: the network is deeper than LeNet, so a saturating activation would
have made the vanishing-gradient problem *worse*, and instead it has gone away.
Almost every architectural difference between the 1990s and now traces back to
some version of this — keeping a usable gradient flowing to the early layers is
the recurring problem, and we will meet it twice more, in batch normalization and
in residual connections.

**There is very little overfitting**, even though the model has 47 million
parameters and the training set has 60 thousand images — nearly 800 parameters
per example.  Classical learning theory says this should memorize the training
set completely.  Training and validation loss instead track each other closely.
Dropout and the implicit regularization of stochastic gradient descent are doing
the work, and the fact that they can do it at this ratio of parameters to data is
one of the genuinely surprising empirical facts about deep networks.

### Saving the trained model

An hour of training is worth keeping.  `state_dict()` returns the learned tensors
— weights and biases, without the code that defines the network — which is the
usual thing to save: the architecture lives in `make_alexnet`, so there is no
reason to serialize it as well.

The file comes to 187 MB, which is exactly the figure we computed from the
parameter count above.  A checkpoint really is nothing but the parameters.

```python
model_path = 'alexnet_fashion_mnist.pt'
torch.save(model.state_dict(), model_path)
print(f'saved {model_path}')
```

Loading it back has one wrinkle, and it is the price of the `Lazy` layers.  A
freshly constructed `make_alexnet()` has no parameters at all until it has seen
data, so `load_state_dict` would have nothing to fill.  One forward pass on a
dummy batch materializes them, and only then can the saved weights be copied in.

```python
reloaded = make_alexnet().to(device)
# materialize the Lazy layers' parameters before loading into them
reloaded(torch.zeros(1, 1, 224, 224, device=device))
reloaded.load_state_dict(torch.load(model_path, map_location=device))
reloaded.eval()

# check that it gives the same answers as the model we just trained
batch_X, batch_y = next(iter(val_iter))
with torch.no_grad():
    same = torch.allclose(model.eval()(batch_X.to(device)),
                          reloaded(batch_X.to(device)))
print('reloaded model reproduces the trained model:', same)
```

## What the first layer learned

The first convolutional layer has 96 kernels of size $11 \times 11$ applied
directly to pixels, so, as with LeNet, we can simply look at them.

```python
kernels = model[0].weight.data.cpu()          # (96, 1, 11, 11)
print('kernels:', tuple(kernels.shape))

fig, axes = plt.subplots(8, 12, figsize=(10, 6.8))
for i, ax in enumerate(axes.flatten()):
    limit = kernels[i].abs().max()
    ax.imshow(kernels[i, 0], cmap='RdBu', vmin=-limit, vmax=limit,
              interpolation='nearest')
    ax.axis('off')
fig.suptitle('the 96 first-layer kernels of AlexNet, trained on Fashion-MNIST',
             fontsize=9)
fig.tight_layout()
```

Red is positive weight, blue negative, each kernel scaled to its own range.
Many are oriented light-dark transitions at various angles and frequencies —
edge detectors, the same thing LeNet's six kernels learned, and the same thing we
designed by hand in notebook 04_02.  Others are blobs, or nearly uniform, or
noise; with 96 kernels and a dataset of small grayscale garments there is not
enough structure in the data to give all of them something to do.

Trained on ImageNet instead, the same layer produces this — the most reproduced
figure in the paper:

![Image filters learned by the first layer of AlexNet, trained on ImageNet.](https://github.com/d2l-ai/d2l-en/raw/refs/heads/master/img/filters.png)

Two families are visible: oriented edge and grating detectors in the top half,
and colour-opponent blobs in the bottom half.  This is a textbook diagram of
mammalian V1, and nobody put it there.  It was recovered from data by gradient
descent, on a network whose only prior was the one we derived in notebook 04_02 —
that the operation should be local and translation invariant.

The layers above these are harder to look at directly, but the hierarchy the
1980s neuroscience predicted, and that the crafted-feature pipeline never
achieved, is what the remaining seven layers build.

## Summary

* **AlexNet is LeNet, scaled up**, with a small number of targeted changes.  The
  architectural gap between 1998 and 2012 is much smaller than the fourteen years
  suggest; what changed was the availability of ImageNet-scale data and of GPUs
  fast enough to use it.
* **ReLU in place of sigmoid** removes the vanishing-gradient plateau, which is
  what makes eight layers trainable at all.  **Max-pooling in place of average
  pooling**, and overlapping pooling windows, are smaller wins in the same
  direction.
* **Dropout** regularizes a fully connected head with 43 million parameters,
  trained on far fewer examples than it has parameters, with almost no
  overfitting.
* **Parameters and compute are distributed inversely.**  The convolutional body
  holds 8% of the parameters and does 95% of the arithmetic; the dense head is
  the reverse.  Everything that follows in this module attacks the head.
* The first layer, trained on photographs, reconstructs the oriented-edge and
  colour-opponent detectors of visual cortex from nothing but data.
* AlexNet is a *design*, not a *principle*: its kernel sizes and channel counts
  are choices with no particular justification beyond having worked.  The next
  notebook asks what a network looks like when it is specified by a repeated
  pattern instead.
