# When the Transformer Earns Its Keep: Natural Language Inference

This module has delivered an uncomfortable result.  On IMDb sentiment, a linear model on word counts — TF-IDF + logistic regression — reaches ~0.91, and our recurrent network, our from-scratch transformer, and even a frozen BERT struggle to clearly beat it.  The reason is that movie-review sentiment is a *lexical* task: the answer is mostly in *which words appear* ("waste", "brilliant"), so counting the right words is nearly enough.  For a task like that, the transformer's machinery is expensive overkill.

So when does that machinery pay off?  When the answer depends on **structure and relationships**, not just word identity — when *which words* is not enough and *how they fit together* is the whole point.  This notebook shows one such task, and the gap between bag-of-words and a transformer goes from "a point or two" to "night and day."

## The task

**Natural language inference** (NLI): given two sentences — a *premise* and a *hypothesis* — decide whether the hypothesis is

* **entailment** — it must be true if the premise is,
* **contradiction** — it cannot be true if the premise is, or
* **neutral** — it might be either.

| premise | hypothesis | label |
|---|---|---|
| A soccer game with multiple males playing. | Some men are playing a sport. | entailment |
| A man inspects his uniform. | The man is sleeping. | contradiction |
| An older woman sits with a drink. | A woman is waiting for her order. | neutral |

Notice what makes this hard for bag-of-words.  In the contradiction example the two sentences *share almost no decisive words*, and in the entailment example the label hinges on "soccer" ⊑ "sport" and "males" ⊑ "men" — relationships between words, holding across two sentences.  The answer lives in *how the sentences relate*, which a model that only counts words cannot see.  A transformer, which relates every token in the hypothesis to every token in the premise, is built for exactly this.  We use the **Stanford Natural Language Inference (SNLI)** corpus.

*This notebook downloads a dataset and a pretrained model, so it needs an internet connection on first run (and `transformers` installed).*

```python
# render figures as crisp vector graphics (SVG) in the notebook
try:
    from matplotlib_inline.backend_inline import set_matplotlib_formats
    set_matplotlib_formats('svg')
except ImportError:
    pass

import os
import zipfile
import urllib.request

import numpy as np

def download_snli(data_dir='./data'):
    """Download and extract the SNLI corpus (a fast d2l mirror)."""
    url = 'https://d2l-data.s3-accelerate.amazonaws.com/snli_1.0.zip'
    os.makedirs(data_dir, exist_ok=True)
    archive = os.path.join(data_dir, 'snli_1.0.zip')
    if not os.path.exists(archive):
        print('downloading', url, '...')
        urllib.request.urlretrieve(url, archive)
    extracted = os.path.join(data_dir, 'snli_1.0')
    if not os.path.exists(extracted):
        with zipfile.ZipFile(archive) as z:
            z.extractall(data_dir)
    return extracted

LABELS = {'entailment': 0, 'contradiction': 1, 'neutral': 2}

def read_snli(data_dir, split):
    """Return (premises, hypotheses, labels) for 'train' or 'test'.
    SNLI rows are tab-separated; gold label is column 0, and the raw
    sentences are columns 5 and 6.  Rows with no gold label ('-') are
    dropped."""
    fname = os.path.join(data_dir, f'snli_1.0_{split}.txt')
    premises, hypotheses, labels = [], [], []
    with open(fname, encoding='utf-8') as f:
        for row in f.readlines()[1:]:                 # skip the header line
            cols = row.split('\t')
            if cols[0] in LABELS:
                premises.append(cols[5])
                hypotheses.append(cols[6])
                labels.append(LABELS[cols[0]])
    return premises, hypotheses, labels

def subsample(p, h, y, n, seed=0):
    rng = np.random.default_rng(seed)
    idx = rng.permutation(len(y))[:n]
    return ([p[i] for i in idx], [h[i] for i in idx],
            np.array([y[i] for i in idx]))

data_dir = download_snli()
p_train, h_train, y_train = subsample(*read_snli(data_dir, 'train'), n=5000)
p_test, h_test, y_test = subsample(*read_snli(data_dir, 'test'), n=2000)
print(len(p_train), 'train /', len(p_test), 'test pairs')
print('label balance (train):', np.bincount(y_train))   # ~even 3-way split
```

## The bag-of-words baseline

We repeat exactly the approach that dominated IMDb: TF-IDF features and a logistic regression.  The only wrinkle is that we have *two* sentences per example, so we concatenate them (with a separator) and let TF-IDF count words and bigrams across the pair.  If lexical content were enough, this would do well.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

def join(prem, hyp):
    return [p + ' [SEP] ' + h for p, h in zip(prem, hyp)]

vec = TfidfVectorizer(ngram_range=(1, 2), min_df=5, sublinear_tf=True)
X_train = vec.fit_transform(join(p_train, h_train))
X_test = vec.transform(join(p_test, h_test))

clf = LogisticRegression(C=10, max_iter=1000)
clf.fit(X_train, y_train)
print(f'TF-IDF + logistic regression accuracy: '
      f'{clf.score(X_test, y_test):.3f}   (3-way chance = 0.33)')
```

This lands well below what it managed on sentiment — typically ~0.55–0.65, not far above the majority/bias baseline and nowhere near usable.  The words are all there, but scrambled across two sentences with the crucial *relationship* thrown away, they simply do not carry the answer.  (In fact much of even this modest score comes from dataset artifacts — certain hypothesis words like "not" or "sleeping" correlate with contradiction regardless of the premise; exercise 2 exposes this.)

## A pretrained transformer

Now a model built to relate two sentences.  We do not train anything: we load a compact transformer that was *already* fine-tuned on NLI data and run it — the same "borrow a pretrained model" move as the BERT notebook, here for a task where it matters.  It reads the premise and hypothesis together, every token of one attending to every token of the other, and outputs one of the three labels.

```python
from transformers import pipeline

# a small NLI-fine-tuned transformer; runs on CPU.  (Alternatives: any
# *-mnli model, e.g. 'roberta-large-mnli' for higher accuracy but a big
# download.)  We print the model's own label names to map them safely.
NLI_MODEL = 'cross-encoder/nli-distilroberta-base'
nli = pipeline('text-classification', model=NLI_MODEL, top_k=None, device=-1)
print('model label names:', nli.model.config.id2label)

def nli_predict(premise, hypothesis):
    scores = nli({'text': premise, 'text_pair': hypothesis})   # list of dicts
    name = max(scores, key=lambda d: d['score'])['label'].lower()
    if 'entail' in name:
        return 0
    if 'contradict' in name:
        return 1
    return 2                                                    # neutral

# inference is slower than TF-IDF, so score a subset (raise on a GPU)
n_eval = 500
preds = np.array([nli_predict(p, h)
                  for p, h in zip(p_test[:n_eval], h_test[:n_eval])])
acc = (preds == y_test[:n_eval]).mean()
print(f'pretrained NLI transformer accuracy: {acc:.3f}   (on {n_eval} pairs)')
```

The transformer lands in the **mid-to-high 0.80s** — a chasm above the bag-of-words 0.6, the opposite of the near-tie we kept seeing on IMDb.  *This* is a task where the architecture earns its keep, and the reason is exactly what we spent the module building: the answer depends on relating the hypothesis to the premise word by word, which is what all-pairs attention does and what counting words cannot.

It closes a loop with notebook 3.  There we scrambled a review's words and the sentiment prediction did not budge — evidence the task was order-blind.  Scramble the words of an NLI pair and the answer *changes*: "the dog chased the cat" and "the cat chased the dog" entail different things.  Order and structure are the whole signal here, so the permutation-*equivariant*, all-pairs transformer is the right tool, and the permutation-*invariant* bag of words is the wrong one.

## The real lesson: match the tool to the task

Put the two tasks side by side:

| | IMDb sentiment | natural language inference |
|---|---|---|
| what carries the answer | which words appear | how the sentences relate |
| TF-IDF + logistic regression | ~0.91 | ~0.6 |
| pretrained transformer | ~0.91 (fine-tuned) | ~0.88 |

The transformer is not "better" in the abstract — it is better *when the task needs what it models*.  On a lexical task a linear bag-of-words is a ferociously strong, cheap baseline that deep learning barely improves; on a task about structure and relationships it is helpless, and the transformer is transformative.  The engineering lesson is to always run the cheap baseline first: it tells you whether your problem actually needs the expensive machinery, and it guards against being impressed by a complex model that is merely keeping pace with logistic regression.

## Summary

* Movie-review sentiment is largely *lexical*, so a TF-IDF + logistic-regression baseline is very hard to beat — deep models earn little there.
* Natural language inference is *relational*: the label depends on how a hypothesis relates to a premise, not on which words appear.  Bag-of-words collapses to ~0.6, while a pretrained transformer reaches the high 0.80s.
* The gap is the point: transformers pay off precisely when structure and word order carry the meaning — the setting their all-pairs, order-sensitive attention was designed for.
* Always benchmark against a strong simple baseline; it reveals whether the task needs deep learning at all.
