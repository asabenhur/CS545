# Assignment: Can a General LLM Replace a Specialist?

Module 7 closed with a ladder.  On **natural language inference** — given a premise and a hypothesis, decide *entailment*, *contradiction*, or *neutral* — four models spread out across forty accuracy points:

| rung | model | what it is | SNLI accuracy |
|---|---|---|---|
| 1 | TF-IDF + logistic regression | bag of words, no structure | ~0.48 |
| 2 | transformer + GloVe, from scratch | our architecture, trained by us | ~mid-0.60s |
| 3 | DistilBERT, fine-tuned | pretrained + our task training | ~high-0.70s |
| 4 | off-the-shelf NLI model | pretrained *and* task-tuned by others | ~mid-0.80s |

That spread is what makes NLI the right task for this assignment.  On IMDb sentiment every method ties near 0.90, because the answer is mostly in *which words appear* — notebook 4's in-context result sat in the same crowded band, below a bag-of-words baseline.  NLI is **relational**: the label depends on how two sentences relate, so a method either captures that structure or it does not, and the accuracy says which.

This module has since given you two ways to attack a task without building a model for it:

* **In-context learning** (notebook 4): describe the task in a prompt, supply a few worked examples, change no parameters at all.
* **LoRA fine-tuning** (notebook 5): freeze a pretrained model and train a small low-rank adapter on labeled data.

**Your question is where those two land on the ladder above.**  Rungs 1–4 were built by training a model *for this task*.  You now have a general-purpose language model that was trained for no task in particular.  Does generality substitute for specialization here — and if so, at what cost?

The notebook gives you working code for both methods, a fair evaluation harness, and a set of questions to answer with evidence.  It does not tell you how the comparison comes out.  I do not know either: whether a 360M decoder with an adapter can catch a fully fine-tuned 66M bidirectional encoder on sentence-pair inference is a genuinely open question at this scale, and finding out is the assignment.

```python
# render figures as crisp vector graphics (SVG) in the notebook
try:
    from matplotlib_inline.backend_inline import set_matplotlib_formats
    set_matplotlib_formats('svg')
except ImportError:
    pass

import gc
import math
import os
import time
import urllib.request
import zipfile

import numpy as np
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
from transformers import AutoTokenizer, AutoModelForCausalLM

BLUE, GREEN, IO = '#cdddec', '#d0e3d4', '#ededed'
AMBER, TX = '#c0662f', '#333333'

# Both parts of this assignment run on a CPU.  If you have a GPU, set this to
# False -- but see notebook 5's note on Apple MPS if the kernel dies.
FORCE_CPU = True

device = ('cpu' if FORCE_CPU
          else 'cuda' if torch.cuda.is_available()
          else 'mps' if torch.backends.mps.is_available() else 'cpu')
torch.manual_seed(0)
print(f'Using {device} device')

MODEL = 'HuggingFaceTB/SmolLM2-360M'    # a base model: no instruction tuning
```

## The data

Identical loading code to module 7's NLI notebook, so the two share a download and — importantly — the same label convention.  If you ran that notebook, SNLI is already cached.

```python
def _download(urls, archive):
    """Fetch the first working URL (browser User-Agent avoids many 403s)."""
    err = None
    for url in urls:
        try:
            print('downloading', url, '...')
            req = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0'})
            with urllib.request.urlopen(req) as r:
                payload = r.read()
            with open(archive, 'wb') as out:
                out.write(payload)
            return
        except Exception as e:
            print('  failed:', e)
            err = e
    raise RuntimeError(f'could not download ({err}); fetch it manually')

def download_snli(data_dir='./data'):
    os.makedirs(data_dir, exist_ok=True)
    archive = os.path.join(data_dir, 'snli_1.0.zip')
    extracted = os.path.join(data_dir, 'snli_1.0')
    if not os.path.exists(extracted):
        if not os.path.exists(archive):
            _download(['https://nlp.stanford.edu/projects/snli/snli_1.0.zip',
                       'https://d2l-data.s3-accelerate.amazonaws.com/snli_1.0.zip'],
                      archive)
        with zipfile.ZipFile(archive) as z:
            z.extractall(data_dir)
    return extracted

LABELS = {'entailment': 0, 'contradiction': 1, 'neutral': 2}
LABEL_NAMES = ['entailment', 'contradiction', 'neutral']

def read_snli(data_dir, split):
    """(premises, hypotheses, labels); rows with no gold label are dropped."""
    fname = os.path.join(data_dir, f'snli_1.0_{split}.txt')
    prem, hyp, lab = [], [], []
    with open(fname, encoding='utf-8') as f:
        for row in f.readlines()[1:]:
            c = row.split('\t')
            if c[0] in LABELS:
                prem.append(c[5]); hyp.append(c[6]); lab.append(LABELS[c[0]])
    return prem, hyp, lab

def subsample(p, h, y, n, seed=0):
    rng = np.random.default_rng(seed)
    idx = rng.permutation(len(y))[:n]
    return ([p[i] for i in idx], [h[i] for i in idx],
            np.array([y[i] for i in idx]))

data_dir = download_snli()
p_test, h_test, y_test = subsample(*read_snli(data_dir, 'test'), n=300, seed=0)
p_pool, h_pool, y_pool = subsample(*read_snli(data_dir, 'train'), n=4000, seed=1)

print(f'{len(y_test)} test pairs, {len(y_pool)} training pairs available')
print('test label balance:', np.bincount(y_test), '(entail / contra / neutral)')
print(f'\nmajority-class baseline: {np.bincount(y_test).max() / len(y_test):.3f}')
for i in range(2):
    print(f'\npremise   : {p_pool[i].strip()}')
    print(f'hypothesis: {h_pool[i].strip()}')
    print(f'label     : {LABEL_NAMES[y_pool[i]]}')
```

Note the two baselines you must clear before claiming anything works: **chance is 0.333**, and the **majority class** is whatever the printout says (SNLI is close to balanced, so they are similar).  A three-way task makes it easy to fool yourself — a model that always answers "neutral" will look like it is doing something.  Always print the confusion matrix before believing an accuracy.

## Part 1 — In-context learning

The method from notebook 4, adapted from two classes to three.  We build a prompt containing `k` worked examples and then the query, and read the model's answer from a **single forward pass** by comparing the logits of the three label tokens.  No generation, no parsing, no gradients.

The verbalizers — the actual words standing in for the classes — are a free choice, and a consequential one.  We start with the obvious mapping and you will test alternatives.

```python
tok = AutoTokenizer.from_pretrained(MODEL)
model = AutoModelForCausalLM.from_pretrained(MODEL, torch_dtype=torch.float32
                                             ).to(device).eval()

# Verbalizers: one word per class.  These must be SINGLE tokens for the
# one-forward-pass trick to work, and the leading space matters.
VERBALIZERS = [' true', ' false', ' maybe']      # entailment / contradiction / neutral

def check_verbalizers(words):
    ids = []
    for w in words:
        enc = tok(w, add_special_tokens=False)['input_ids']
        assert len(enc) == 1, f'{w!r} is {len(enc)} tokens: {tok.convert_ids_to_tokens(enc)}'
        ids.append(enc[0])
    return ids

LABEL_IDS = check_verbalizers(VERBALIZERS)
print('verbalizer token ids:', dict(zip(VERBALIZERS, LABEL_IDS)))

def format_example(premise, hypothesis, label=None):
    """One block of the prompt; `label=None` leaves the answer open."""
    s = (f'Premise: {premise.strip()}\n'
         f'Hypothesis: {hypothesis.strip()}\n'
         f'Question: Is the hypothesis true given the premise?\nAnswer:')
    return s if label is None else s + VERBALIZERS[label]

INSTRUCTION = ('Decide whether each hypothesis is true, false, or maybe true '
               'given its premise.\n\n')

def build_prompt(premise, hypothesis, shots=()):
    blocks = [format_example(p, h, y) for p, h, y in shots]
    blocks.append(format_example(premise, hypothesis))
    return INSTRUCTION + '\n\n'.join(blocks)

print('\n--- a 1-shot prompt ---\n')
print(build_prompt(p_test[0], h_test[0], shots=[(p_pool[0], h_pool[0], y_pool[0])]))
```

```python
@torch.no_grad()
def icl_predict(premise, hypothesis, shots=()):
    """One forward pass; the prediction is the highest-scoring label token."""
    ids = tok(build_prompt(premise, hypothesis, shots),
              return_tensors='pt')['input_ids'].to(device)
    logits = model(ids).logits[0, -1].float()
    return int(torch.tensor([logits[i] for i in LABEL_IDS]).argmax())

def make_shots(k, seed=0):
    """k demonstrations, class-balanced and in a fixed shuffled order.

    Balance matters: notebook 4's exercises showed that a model infers the
    label *distribution* from the demonstrations as readily as the task."""
    if k == 0:
        return []
    rng = np.random.default_rng(seed)
    per_class, chosen = max(1, k // 3), []
    for c in range(3):
        idx = [i for i in range(len(y_pool)) if y_pool[i] == c][:per_class]
        chosen += [(p_pool[i], h_pool[i], int(y_pool[i])) for i in idx]
    chosen = chosen[:k] if k < len(chosen) else chosen
    rng.shuffle(chosen)
    return chosen

def evaluate(predict_fn, n=None):
    """Accuracy plus a 3x3 confusion matrix (rows = truth, cols = prediction)."""
    n = len(y_test) if n is None else min(n, len(y_test))
    conf = np.zeros((3, 3), dtype=int)
    for i in range(n):
        conf[y_test[i], predict_fn(p_test[i], h_test[i])] += 1
    return np.trace(conf) / conf.sum(), conf

def show_result(name, acc, conf):
    print(f'{name}: accuracy {acc:.3f}  ({conf.sum()} examples)')
    print('        predicted:  ' + '  '.join(f'{c[:6]:>6s}' for c in LABEL_NAMES))
    for i, row in enumerate(conf):
        print(f'  true {LABEL_NAMES[i][:6]:>6s}:  ' + '  '.join(f'{v:6d}' for v in row))
    print()
```

Now the sweep.  Start small — `N_EVAL = 100` — to see how long one setting takes, then raise it once you know the cost.  Every accuracy you report should come with the `n` it was measured on, because the difference between 0.42 and 0.47 on 100 examples is not a difference at all (see the error-bar note below).

```python
N_EVAL = 100          # raise once you have timed one setting

icl_results = {}
for k in [0, 3, 6, 12]:
    shots = make_shots(k)
    t0 = time.time()
    acc, conf = evaluate(lambda p, h: icl_predict(p, h, shots), n=N_EVAL)
    icl_results[k] = acc
    show_result(f'{k}-shot  [{time.time() - t0:.0f}s]', acc, conf)

print('A useful sanity check: a binomial 95% interval on n examples is roughly')
print(f'+/- {1.96 * math.sqrt(0.4 * 0.6 / N_EVAL):.3f} at these accuracies, on n = {N_EVAL}.')
```

**Before going on, look at the confusion matrices, not just the accuracies.**  A model that has not understood the task usually reveals itself by collapsing onto one or two classes — and *neutral* is the classic attractor, because "maybe" is a safe continuation when the model has no idea.  If two of your rows are nearly empty, the accuracy number is measuring class priors, not inference.

## Part 2 — LoRA fine-tuning

Same model, same evaluation, but now we train an adapter on labeled examples, exactly as in notebook 5.  Two design decisions are already made for you, and both are worth understanding:

* We train on the **same prompt format** used above, so the two methods are compared on equal footing.  A fine-tune that changed the format would confound "training helped" with "the format changed".
* The loss is computed on the **label token only** — every other position is masked to `-100`.  Notebook 5 masked the whole prompt so the model would learn to answer rather than to ask; here the "answer" is a single token, so the masking is even more aggressive.  The model is being taught one thing: which of three words follows `Answer:`.

```python
from peft import LoraConfig, get_peft_model

N_TRAIN = 1000        # from the 4000-pair pool; raise if you have the time
MAX_LEN = 128
N_TRAIN = min(N_TRAIN, len(y_pool) - 200)   # always leave 200 for validation

def encode_for_training(premise, hypothesis, label):
    """Prompt + one label token; only the label token carries loss."""
    prompt_ids = tok(build_prompt(premise, hypothesis), add_special_tokens=False)['input_ids']
    prompt_ids = prompt_ids[-(MAX_LEN - 1):]           # keep the end if too long
    ids = prompt_ids + [LABEL_IDS[label]]
    labels = [-100] * len(prompt_ids) + [LABEL_IDS[label]]
    return {'input_ids': ids, 'labels': labels}

train_examples = [encode_for_training(p_pool[i], h_pool[i], int(y_pool[i]))
                  for i in range(N_TRAIN)]
val_examples = [encode_for_training(p_pool[i], h_pool[i], int(y_pool[i]))
                for i in range(N_TRAIN, min(N_TRAIN + 200, len(y_pool)))]

ex = train_examples[0]
print(f'{len(train_examples)} training / {len(val_examples)} validation examples')
print(f'example: {len(ex["input_ids"])} tokens, '
      f'{sum(1 for l in ex["labels"] if l != -100)} supervised '
      f'(the label token: {tok.decode([ex["input_ids"][-1]])!r})')

def collate(batch):
    n = max(len(b['input_ids']) for b in batch)
    pad = tok.pad_token_id if tok.pad_token_id is not None else tok.eos_token_id
    x, y, m = [], [], []
    for b in batch:
        k = n - len(b['input_ids'])
        x.append(b['input_ids'] + [pad] * k)
        y.append(b['labels'] + [-100] * k)
        m.append([1] * len(b['input_ids']) + [0] * k)
    return torch.tensor(x), torch.tensor(y), torch.tensor(m)
```

```python
lora_cfg = LoraConfig(r=8, lora_alpha=16, lora_dropout=0.05, bias='none',
                      task_type='CAUSAL_LM',
                      target_modules=['q_proj', 'k_proj', 'v_proj', 'o_proj'])
model = get_peft_model(model, lora_cfg)
model.print_trainable_parameters()

MICRO_BATCH, ACCUM_STEPS = 4, 2          # effective batch of 8
MAX_STEPS, WARMUP, LR = 120, 10, 2e-4
EVAL_EVERY = 20

@torch.no_grad()
def val_loss(examples, n_seqs=32):
    model.eval()
    total, seen = 0.0, 0
    for i in range(0, min(len(examples), n_seqs), MICRO_BATCH):
        xb, yb, mb = collate(examples[i:i + MICRO_BATCH])
        out = model(input_ids=xb.to(device), attention_mask=mb.to(device),
                    labels=yb.to(device))
        k = int((yb[:, 1:] != -100).sum())
        total += out.loss.item() * k; seen += k
    model.train()
    return total / max(seen, 1)

def lr_at(step):
    if step < WARMUP:
        return LR * (step + 1) / WARMUP
    prog = (step - WARMUP) / max(1, MAX_STEPS - WARMUP)
    return 0.1 * LR + 0.45 * LR * (1 + math.cos(math.pi * prog))

params = [p for p in model.parameters() if p.requires_grad]
opt = torch.optim.AdamW(params, lr=LR, weight_decay=0.0)
rng = np.random.default_rng(0)
history = [(0, val_loss(train_examples), val_loss(val_examples))]
print(f'step   0 | train {history[0][1]:.3f} | val {history[0][2]:.3f}')

model.train()
t0 = time.time()
for step in range(1, MAX_STEPS + 1):
    for g in opt.param_groups:
        g['lr'] = lr_at(step)
    opt.zero_grad(set_to_none=True)
    for _ in range(ACCUM_STEPS):
        idx = rng.integers(0, len(train_examples), MICRO_BATCH)
        xb, yb, mb = collate([train_examples[i] for i in idx])
        out = model(input_ids=xb.to(device), attention_mask=mb.to(device),
                    labels=yb.to(device))
        (out.loss / ACCUM_STEPS).backward()
    torch.nn.utils.clip_grad_norm_(params, 1.0)
    opt.step()
    if step == 2:
        per = (time.time() - t0) / 2
        print(f'  ~{per:.1f}s per step -> about {per * MAX_STEPS / 60:.0f} min total\n')
    if step % EVAL_EVERY == 0:
        tr, va = val_loss(train_examples), val_loss(val_examples)
        history.append((step, tr, va))
        print(f'step {step:3d} | train {tr:.3f} | val {va:.3f} '
              f'| {time.time() - t0:5.0f}s')

print(f'\nfinished in {(time.time() - t0)/60:.1f} min')
# Free the last step's graph before evaluating (notebook 5 explains why).
del out, xb, yb, mb
opt.zero_grad(set_to_none=True)
gc.collect()
if device == 'mps':
    torch.mps.empty_cache()
elif device == 'cuda':
    torch.cuda.empty_cache()
```

```python
# The tuned model, evaluated exactly as the in-context one was -- zero shots,
# because the training has taken the place of the demonstrations.
model.eval()
acc_lora, conf_lora = evaluate(lambda p, h: icl_predict(p, h, ()), n=N_EVAL)
show_result('LoRA fine-tuned, 0-shot', acc_lora, conf_lora)

# And the same model with the adapter switched off, as a control: this is the
# 0-shot in-context number again, and it should reproduce it.
with model.disable_adapter():
    acc_ctrl, conf_ctrl = evaluate(lambda p, h: icl_predict(p, h, ()), n=N_EVAL)
show_result('adapter disabled (control)', acc_ctrl, conf_ctrl)
```

```python
ladder = [('TF-IDF + logistic regression', 0.48, '#7a7a7a'),
          ('transformer + GloVe, from scratch', 0.65, '#7a7a7a'),
          ('DistilBERT, fine-tuned', 0.78, AMBER),
          ('off-the-shelf NLI model', 0.85, '#7a7a7a')]

fig, ax = plt.subplots(figsize=(7.2, 4.4))
ks = sorted(icl_results)
ax.plot(ks, [icl_results[k] for k in ks], 'o-', color='#3d6ea8',
        label=f'in-context ({MODEL.split("/")[-1]})', markersize=5)
ax.axhline(acc_lora, color=GREEN, lw=2.5, label='LoRA fine-tuned (0-shot)')
ax.axhline(1/3, ls=':', color='0.6')
ax.text(max(ks), 0.339, 'chance', fontsize=7.5, color='0.45', ha='right')
for name, y, col in ladder:
    ax.axhline(y, ls='--', color=col, lw=1)
    ax.text(max(ks), y + 0.006, f'{name} (module 7)', fontsize=7.2, color=col,
            ha='right', va='bottom')
ax.set_xlabel('demonstrations in the prompt')
ax.set_ylabel(f'SNLI accuracy  (n = {N_EVAL})')
ax.set_ylim(0.25, 0.95)
ax.set_title('Two ways to use a general model, against four specialists', fontsize=10)
ax.legend(frameon=False, loc='lower right', fontsize=8)
ax.spines[['top', 'right']].set_visible(False)
plt.tight_layout();

print('Replace the module 7 numbers above with YOUR measured values if you ran '
      'that notebook -- the defaults here are the ranges it reports.')
```

### A note on honest evaluation

Three habits this assignment is really about.  **Report `n`** — an accuracy without a sample size is not a measurement.  **Look at the confusion matrix** — on a three-way task, aggregate accuracy hides collapse onto a single class.  **Run the ablation that could embarrass you** — the hypothesis-only check exists because a result that survives it means much more than one that was never tested.  These are the same habits that separate a real experimental claim from a demo, in this course and outside it.
